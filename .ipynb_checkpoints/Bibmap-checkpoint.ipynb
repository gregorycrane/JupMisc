```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt

import csv, os
import pandas as pd

TSV_PATH = "../GRC_misc/aeschbib.tsv"   # your path

def read_tsv_ragged(path):
    rows = []
    with open(path, "r", errors="ignore", newline="") as f:
        rdr = csv.reader(f, delimiter="\t", quoting=csv.QUOTE_NONE, strict=False)
        for row in rdr:
            row = [(c or "").replace("\x00", "").strip() for c in row]
            rows.append(row)
    if not rows:
        return pd.DataFrame()
    ncols = len(rows[0])
    # pad/truncate to header width
    fixed = []
    for r in rows[1:]:
        if len(r) < ncols:
            r = r + [""]*(ncols - len(r))
        elif len(r) > ncols:
            r = r[:ncols]
        fixed.append(r)
    # ensure unique, non-empty column names
    cols = [c if c else f"col_{i}" for i, c in enumerate(rows[0])]
    seen = {}
    uniq = []
    for c in cols:
        if c in seen:
            seen[c] += 1
            uniq.append(f"{c}_{seen[c]}")
        else:
            seen[c] = 0
            uniq.append(c)
    return pd.DataFrame(fixed, columns=uniq)

df = read_tsv_ragged(TSV_PATH)

# Normalize known abbreviations
mapping = {
    "Lips.": "Leipzig, Germany",
    "Berol.": "Berlin, Germany",
    "Oxon.": "Oxford, United Kingdom",
    "Cantab.": "Cambridge, United Kingdom",
    "Cambr.": "Cambridge, United Kingdom",
    "Lond.": "London, United Kingdom",
    "Bonn": "Bonn, Germany",
    "Vindob.": "Vienna, Austria",
    "Hagae Comitis": "The Hague, Netherlands",
    "Lugd. Bat.": "Leiden, Netherlands",
    "Amst.": "Amsterdam, Netherlands",
    "Halle": "Halle (Saale), Germany",
    "Halle (Saale)": "Halle (Saale), Germany",
}

df["Place_norm"] = df["Place"].replace(mapping)
places = df["Place_norm"].dropna().value_counts()

# Quick geocoding (approximate coordinates)
geo_lookup = {
    "Leipzig, Germany": (51.34, 12.37),
    "Berlin, Germany": (52.52, 13.40),
    "Oxford, United Kingdom": (51.75, -1.26),
    "Cambridge, United Kingdom": (52.20, 0.12),
    "London, United Kingdom": (51.51, -0.13),
    "Bonn, Germany": (50.73, 7.10),
    "Vienna, Austria": (48.21, 16.37),
    "The Hague, Netherlands": (52.08, 4.30),
    "Leiden, Netherlands": (52.16, 4.49),
    "Amsterdam, Netherlands": (52.37, 4.89),
    "Halle (Saale), Germany": (51.48, 11.97),
}

geo_df = pd.DataFrame([
    {"City": city, "Count": count,
     "lat": geo_lookup[city][0],
     "lon": geo_lookup[city][1]}
    for city, count in places.items() if city in geo_lookup
])

# Plot
world = gpd.read_file(gpd.datasets.get_path("naturalearth_lowres"))
ax = world.plot(color="whitesmoke", edgecolor="gray", figsize=(12,8))
plt.scatter(
    geo_df["lon"], geo_df["lat"],
    s=geo_df["Count"]*30, color="royalblue", alpha=0.7, edgecolor="black"
)
for _, r in geo_df.iterrows():
    plt.text(r["lon"]+0.3, r["lat"]+0.3, r["City"].split(",")[0], fontsize=8)
plt.title("Publication Geography of Aeschylus Scholarship (1700–1950)")
plt.xlabel("Longitude"); plt.ylabel("Latitude")
plt.show()
```
