```python
import re

curtg = '0032'
curwork = '006'

fname = '/Users/gcrane/github/glaux-trees/public/xml/' + curtg + '-' + curwork + '.xml'

f = open(fname)

curcit = ''
cursent = ''
sentcits = {}
senttexts = {}
curform = ''
sents2cit = {}
sentsincit = {}
citswithsents = {}

def addentry(curdict,curentry):
    if(curentry in curdict):
        curdict[curentry] = curdict[curentry] + 1
    else:
        curdict[curentry] = 1

for l in f:
    l = re.sub('\s+$','',l)
    m = re.search('subdoc="([^"]+)"',l)
    if(m):
        curcit = m[1]
        addentry(sents2cit,curcit)
        if(not curcit in sentsincit):
            sentsincit[curcit]  = []
        if(not curcit in citswithsents):
            citswithsents[curcit] = ''
        
        
    #if(not re.search('^1\.3\.',curcit)):
    #    continue

    m = re.search('sentence id="([^"]+)"',l)
    if(m):
        cursent = m[1]
        senttexts[cursent] = ''
        sentcits[cursent] = curcit
        sentsincit[curcit].append(cursent)
        citswithsents[curcit] = citswithsents[curcit] + ' <s n="'+cursent+'"/> '


    m = re.search('form="([^"]+)"',l)
    if(m):
        curform = m[1]
        citswithsents[curcit] = citswithsents[curcit] + ' ' + curform
        if(senttexts[cursent]):
            senttexts[cursent] = senttexts[cursent] + ' ' + curform
        else:
            senttexts[cursent] = curform


    #if(not re.search('^1\.1.',curcit)):
    #    continue
        
engcittext = {}
f = open('/Users/gcrane/github/anabasis/tlg0032.tlg006.rouse-eng2.xml')
curbook = ''
curchap = ''
cursect = ''
curcit = ''
for l in f:
    l = re.sub('\s+$','',l)
    l = re.sub('<note[^>]*>[^<]+</note>',' ',l)
    m = re.search('type="book" n="([^"]+)',l)
    if(m):
        curbook = m[1]
    m = re.search('type="chapter" n="([^"]+)',l)
    if(m):
        curchap = m[1]
        
    m = re.search('type="section" n="([^"]+)',l)
    if(m):
        cursect = m[1]
        curcit = curbook + '.' + curchap + '.' + cursect
        engcittext[curcit] = ''
        continue
    if(re.search('(</div>|<head>)',l)):
       continue
    if(not curcit):
        continue
    engcittext[curcit] = engcittext[curcit] + ' ' + l


    


    
```
