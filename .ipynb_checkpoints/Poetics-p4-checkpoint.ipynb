```python
import os
import re
import time
from pathlib import Path
from collections import OrderedDict
import xml.etree.ElementTree as ET

# ── Local Configuration for Aristotle's Poetics ────────────────
OUT_DIR = Path("/Users/gcrane/Downloads/gemsite/poetics_site")
OUT_DIR.mkdir(parents=True, exist_ok=True)

NS = {'tei': 'http://www.tei-c.org/ns/1.0'}

# 1. Register an arbitrary number of localized versions dynamically
TEXT_REGISTRY = {
    "grc_kassel": {
        "path": "/Users/gcrane/github/canonical-greekLit/data/tlg0086/tlg034/tlg0086.tlg034.perseus-grc2.xml",
        "label": "Greek (Kassel, 1965)",
        "class": "greek-text",
        "is_greek": True
    },
    "grc_digi": {
        "path": "/Users/gcrane/github/canonical-greekLit/data/tlg0086/tlg034/tlg0086.tlg034.digicorpus-grc2.xml",
        "label": "Greek (Digital Corpus Variant)",
        "class": "greek-text",
        "is_greek": True
    },
    "eng_fyfe": {
        "path": "/Users/gcrane/github/canonical-greekLit/data/tlg0086/tlg034/tlg0086.tlg034.perseus-eng2.xml",
        "label": "English (W.H. Fyfe, 1927)",
        "class": "english-text",
        "is_greek": False
    },
    "eng_butcher": {
        "path": "/Users/gcrane/github/Poetics2.0/eng/tlg0086.tlg034.butcher1911.xml",
        "label": "English (S.H. Butcher, 1911)",
        "class": "english-text",
        "is_greek": False
    }
}

# ── Authentic Hopper Layout CSS (With Independent Flex Scroll) ──
PERSEUS_CSS = r"""
* { margin: 0; padding: 0; box-sizing: border-box; }
html, body {
    height: 100%;
    overflow: hidden; /* Prevents the main browser window from scrolling globally */
    font-family: "Palatino Linotype", "Book Antiqua", Palatino, Georgia, serif;
    font-size: 13px;
    color: #000;
    background: #fff;
}
a { color: #336699; text-decoration: none; }
a:hover { text-decoration: underline; }

/* Fixed Header Layout Block */
#header-container {
    flex-shrink: 0;
    background: #fff;
    z-index: 10;
}
#perseus-banner { background: #660000; color: #fff; padding: 8px 15px; display: flex; justify-content: space-between; align-items: center; border-bottom: 2px solid #330000; }
#perseus-banner h1 { font-size: 18px; font-weight: bold; }
#perseus-banner h1 a { color: #fff; }
#perseus-banner .doc-title { font-size: 12px; color: #ffccaa; font-weight: bold; }
#nav-bar { background: #ddddcc; border-bottom: 1px solid #999988; padding: 5px 15px; font-size: 11px; color: #555; }
#nav-bar a { margin-right: 15px; color: #003366; font-weight: bold; }

/* Hopper Visual Navigation Matrix Bar */
#browse-bar { background: #eeeeee; border-bottom: 1px solid #cccccc; padding: 8px 15px; font-size: 11px; }
.browse-row { margin-bottom: 6px; display: flex; align-items: center; }
.browse-row label { font-weight: bold; width: 80px; color: #555; flex-shrink: 0; }
.browse-items { display: flex; flex-wrap: wrap; gap: 3px; align-items: center; }

.nav-tile-wrapper { display: inline-flex; flex-direction: column; align-items: center; text-align: center; }
.nav-tile-link { display: block; padding: 2px 6px; background: #e0e0d0; color: #336699; border: 1px solid #bbbb99; min-width: 24px; text-align: center; font-weight: bold; }
.nav-tile-link:hover { background: #fff; text-decoration: none; border-color: #660000; }
.nav-tile-link.current { background: #660000; color: #fff !important; border-color: #330000; }
.size-bar { height: 3px; background: #990000; margin-top: 2px; border-radius: 1px; min-width: 4px; }

/* Main Screen View Splitter Framework */
#app-view-root {
    display: flex;
    flex-direction: column;
    height: 100vh;
}
#outer-wrapper { 
    display: flex; 
    flex: 1;
    min-height: 0; /* Critical for inner-scrollbar containment mechanics */
}

/* Sidebar Column Layout */
#sidebar-toc { 
    width: 240px; 
    background: #f5f5ee; 
    border-right: 1px solid #ccccbb; 
    padding: 12px; 
    font-size: 11px; 
    overflow-y: auto; 
    flex-shrink: 0; 
}
#sidebar-toc h3 { font-size: 12px; color: #660000; margin-bottom: 8px; border-bottom: 1px solid #ccccbb; padding-bottom: 3px; }
#sidebar-toc ul { list-style: none; padding-left: 0; }
#sidebar-toc li { margin-bottom: 5px; }
#sidebar-toc li.active-book { font-weight: bold; color: #660000; }

.toc-chapters { padding-left: 10px; margin-top: 3px; font-weight: normal; }
.toc-chapters li { margin-bottom: 3px; }
.toc-chapters a { color: #003366; }
.toc-chapters a.current { color: #660000; font-weight: bold; }
.toc-sections { padding-left: 12px; margin-top: 2px; display: flex; flex-wrap: wrap; gap: 4px; font-weight: normal; }
.toc-sections a { background: #e0e0d4; padding: 1px 4px; border-radius: 2px; color: #333; font-size: 10px; }
.toc-sections a:hover { background: #660000; color: #fff; text-decoration: none; }
.toc-sections a.current { background: #660000; color: #fff !important; font-weight: bold; }

/* Dual Mirror Reading Columns (Independent Container Scrolling) */
#main-container { 
    flex: 1; 
    display: flex; 
    min-width: 0;
    height: 100%;
}
#focus-text { 
    flex: 3; 
    min-width: 0; 
    padding: 15px;
    overflow-y: auto; /* Independent left pane scrollbar */
    height: 100%;
}
#cross-ref { 
    flex: 2; 
    min-width: 0; 
    border-left: 2px solid #eeeeee; 
    padding: 15px;
    overflow-y: auto; /* Independent right pane scrollbar */
    height: 100%;
    background: #fafafa;
}
.panel-header { 
    background: #660000; 
    color: #fff; 
    padding: 4px 8px; 
    font-size: 11px; 
    font-weight: bold; 
    margin-bottom: 10px;
    position: sticky;
    top: 0;
    z-index: 5;
}

/* Philological Selector Workspace Panel Component Layout */
#layout-configurator { background: #fdfae6; border: 1px dashed #d6c57d; padding: 8px; margin-bottom: 12px; font-size: 11px; border-radius: 3px; }
.config-group { margin-bottom: 4px; display: flex; align-items: center; }
.config-group strong { width: 110px; color: #555; }
.config-group a { margin-right: 12px; font-weight: normal; padding: 1px 4px; }
.config-group a.active { background: #660000; color: #fff; font-weight: bold; border-radius: 2px; text-decoration: none; pointer-events: none; }

.tier-indicator { background: #fffccb; border: 1px solid #e6db55; padding: 4px 8px; font-size: 11px; margin-bottom: 10px; font-weight: bold; color: #555; display: inline-block; }
.nav-arrows { display: flex; align-items: center; gap: 10px; font-size: 12px; font-weight: bold; margin: 8px 0; border-bottom: 1px dotted #ccc; padding-bottom: 6px; }
.nav-arrows a { font-size: 14px; color: #660000; }

.section-block { margin-bottom: 18px; position: relative; padding-left: 45px; line-height: 1.6; }
.section-block .sec-num { position: absolute; left: 0; top: 0; color: #990000; font-size: 11px; font-weight: bold; width: 35px; text-align: right; }
.section-block .sec-num a { color: #990000; text-decoration: underline; }
.greek-text { font-size: 16px; line-height: 1.7; font-family: "Gentium Plus", "Athena", Palatino, serif; }
.english-text { font-size: 13px; color: #111; }
.note { font-size: 11px; color: #666; font-style: italic; background: #f0f0f0; padding: 0 2px; }
#credits { background: #f9f9f9; border-top: 1px solid #eee; padding: 10px; margin-top: 30px; font-size: 11px; color: #555; }
"""

# ── Robust XML Parsing Engine ─────────────────────────────────
def extract_text_recursive(elem):
    parts = []
    if elem.text: parts.append(elem.text)
    for child in elem:
        tag = child.tag.replace('{http://www.tei-c.org/ns/1.0}', '')
        if tag == 'note':
            note_text = extract_text_recursive(child).strip()
            if note_text: parts.append(f'<span class="note">[{note_text}]</span>')
        elif tag in ('milestone', 'reg'): pass
        else: parts.append(extract_text_recursive(child))
        if child.tail: parts.append(child.tail)
    return ''.join(parts)

def parse_tei_hierarchy(path):
    if not os.path.exists(path):
        return None
    tree = ET.parse(path)
    root = tree.getroot()
    
    body = root.find('.//tei:body', NS)
    if body is None:
        body = root.find('.//body')
    if body is None:
        return None
        
    top_div = body.find('tei:div', NS) or body.find('div') or body

    data = OrderedDict()
    textparts = top_div.findall('tei:div', NS) or top_div.findall('div')
    
    if not textparts:
        textparts = top_div.findall('.//tei:div[@subtype="chapter"]', NS) or top_div.findall('.//div[@subtype="chapter"]')

    if textparts and (textparts[0].get('subtype') == 'chapter' or textparts[0].get('type') == 'chapter'):
        data['1'] = OrderedDict()
        for ch_div in textparts:
            ch_n = ch_div.get('n')
            if not ch_n: continue
            data['1'][ch_n] = OrderedDict()
            
            sections = ch_div.findall('tei:div[@subtype="section"]', NS) or ch_div.findall('div[@subtype="section"]')
            if not sections:
                sections = ch_div.findall('tei:div', NS) or ch_div.findall('div') or ch_div.findall('tei:p', NS) or ch_div.findall('p')
                
            for idx, sec_div in enumerate(sections):
                sec_n = sec_div.get('n') or str(idx + 1)
                paragraphs = sec_div.findall('tei:p', NS) or sec_div.findall('p')
                if paragraphs:
                    data['1'][ch_n][sec_n] = ' '.join(extract_text_recursive(p).strip() for p in paragraphs)
                else:
                    data['1'][ch_n][sec_n] = extract_text_recursive(sec_div).strip()
    else:
        for b_div in textparts:
            b_n = b_div.get('n') or '1'
            data[b_n] = OrderedDict()
            chapters = b_div.findall('tei:div[@subtype="chapter"]', NS) or b_div.findall('div[@subtype="chapter"]') or b_div.findall('tei:div', NS) or b_div.findall('div')
            for ch_div in chapters:
                ch_n = ch_div.get('n')
                if not ch_n: continue
                data[b_n][ch_n] = OrderedDict()
                
                sections = ch_div.findall('tei:div[@subtype="section"]', NS) or ch_div.findall('div[@subtype="section"]') or ch_div.findall('tei:p', NS) or ch_div.findall('p')
                for idx, sec_div in enumerate(sections):
                    sec_n = sec_div.get('n') or str(idx + 1)
                    paragraphs = sec_div.findall('tei:p', NS) or sec_div.findall('p')
                    if paragraphs:
                        data[b_n][ch_n][sec_n] = ' '.join(extract_text_recursive(p).strip() for p in paragraphs)
                    else:
                        data[b_n][ch_n][sec_n] = extract_text_recursive(sec_div).strip()
                        
    clean_data = OrderedDict()
    for bk, chaps in data.items():
        if chaps: clean_data[bk] = chaps
    return clean_data if clean_data else None

print("Ingesting registered textual configurations...")
MASTER_CORPUS = OrderedDict()
for v_id, config in TEXT_REGISTRY.items():
    parsed_data = parse_tei_hierarchy(config["path"])
    if parsed_data:
        MASTER_CORPUS[v_id] = parsed_data
        print(f" -> Loaded successfully: {config['label']}")

BASELINE_KEY = "grc_kassel"
baseline_structure = MASTER_CORPUS[BASELINE_KEY]

# ── Filename Generator Utilities ──────────────────────────────
def get_filename(book, chapter, focus, cross, section=None):
    if section:
        return f"focus_{focus}_cross_{cross}_b{book}_ch{chapter}_sec{section}.html"
    return f"focus_{focus}_cross_{cross}_b{book}_ch{chapter}.html"

# Calculate character footprint statistics for visual tiles
chapter_sizes = {}
for bn in baseline_structure:
    for cn in baseline_structure[bn]:
        chapter_sizes[(bn, cn)] = sum(len(txt) for txt in baseline_structure[bn][cn].values())
max_char_len = max(chapter_sizes.values()) if chapter_sizes else 1

def get_tile_width(book, chapter, max_px=30):
    size = chapter_sizes.get((book, chapter), 0)
    return max(4, int((size / max_char_len) * max_px))

# ── Component Building Blocks ───────────────────────────────
def render_sections_block(book_n, chap_n, text_version_id, focus_id, cross_id, is_isolated=False, isolated_sec_n=None):
    html = []
    version_data = MASTER_CORPUS.get(text_version_id, {})
    sections_dict = version_data.get(book_n, {}).get(chap_n, {})
    css_class = TEXT_REGISTRY[text_version_id]["class"]
    
    for sec_n, text in sections_dict.items():
        if is_isolated and sec_n != isolated_sec_n:
            continue
        label = sec_n if sec_n not in ('0', 'pr') else 'pr'
        sec_file_link = get_filename(book_n, chap_n, focus_id, cross_id, sec_n)
        
        html.append(
            f'<div class="section-block {css_class}">'
            f'<span class="sec-num"><a href="{sec_file_link}" title="Isolate Section">{label}</a></span>'
            f'<p>{text if text.strip() else "<i>[Text part omitted in this edition]</i>"}</p>'
            f'</div>'
        )
    return '\n'.join(html)

def render_layout_configurator(book_n, chap_n, active_focus, active_cross, current_sec=None):
    focus_links = []
    cross_links = []
    for v_id, config in TEXT_REGISTRY.items():
        f_cls = ' class="active"' if v_id == active_focus else ''
        f_dest = get_filename(book_n, chap_n, v_id, active_cross, current_sec)
        focus_links.append(f'<a href="{f_dest}"{f_cls}>{config["label"]}</a>')
        
        c_cls = ' class="active"' if v_id == active_cross else ''
        c_dest = get_filename(book_n, chap_n, active_focus, v_id, current_sec)
        cross_links.append(f'<a href="{c_dest}"{c_cls}>{config["label"]}</a>')
        
    return f"""<div id="layout-configurator">
        <div class="config-group"><strong>Focus Panel:</strong> {' '.join(focus_links)}</div>
        <div class="config-group" style="margin-top:5px;"><strong>Cross-Ref Panel:</strong> {' '.join(cross_links)}</div>
    </div>"""

def render_top_navigation_matrix(book_n, chap_n, active_focus, active_cross, routing_tier='chapter'):
    book_links = []
    for bn in baseline_structure:
        first_ch = list(baseline_structure[bn].keys())[0]
        cls = ' class="current"' if bn == book_n else ''
        dest = get_filename(bn, first_ch, active_focus, active_cross) if routing_tier == 'chapter' else get_filename(bn, first_ch, active_focus, active_cross, list(baseline_structure[bn][first_ch].keys())[0])
        book_links.append(f'<a href="{dest}"{cls} style="font-weight:bold; padding:2px 6px; margin-right:4px;">Book {bn}</a>')
    
    chap_links = []
    for cn in baseline_structure[book_n]:
        cls = ' current' if cn == chap_n else ''
        dest = get_filename(book_n, cn, active_focus, active_cross) if routing_tier == 'chapter' else get_filename(book_n, cn, active_focus, active_cross, list(baseline_structure[book_n][cn].keys())[0])
        pixel_width = get_tile_width(book_n, cn)
        
        chap_links.append(
            f'<div class="nav-tile-wrapper">'
            f'  <a href="{dest}" class="nav-tile-link{cls}">{cn}</a>'
            f'  <div class="size-bar" style="width:{pixel_width}px;"></div>'
            f'</div>'
        )
        
    return f"""<div id="browse-bar">
    <div class="browse-row"><label>Books:</label><div class="browse-items">{' '.join(book_links)}</div></div>
    <div class="browse-row" style="margin-top:8px;"><label>Chapters:</label><div class="browse-items" style="gap:5px 3px;">{' '.join(chap_links)}</div></div>
</div>"""

def render_sidebar_toc(current_book, current_chap, current_sec_n, active_focus, active_cross, routing_tier='chapter'):
    html = ["<div id=\"sidebar-toc\">", "<h3>Table of Contents</h3>", "<ul>"]
    for bn in baseline_structure:
        if bn == current_book:
            html.append(f'<li class="active-book">Poetics Work Hierarchy <ul class="toc-chapters">')
            for cn in baseline_structure[bn]:
                is_active_ch = (cn == current_chap)
                cls = ' class="current"' if is_active_ch else ''
                dest = get_filename(bn, cn, active_focus, active_cross) if routing_tier == 'chapter' else get_filename(bn, cn, active_focus, active_cross, list(baseline_structure[bn][cn].keys())[0])
                html.append(f'<li><a href="{dest}"{cls}>Chapter {cn}</a>')
                
                if is_active_ch:
                    html.append('<ul class="toc-sections">')
                    for sn in baseline_structure[bn][cn]:
                        s_cls = ' class="current"' if (routing_tier == 'section' and sn == current_sec_n) else ''
                        s_dest = get_filename(bn, cn, active_focus, active_cross, sn)
                        html.append(f'<li><a href="{s_dest}"{s_cls}>{sn}</a></li>')
                    html.append('</ul>')
                
                html.append('</li>')
            html.append('</ul></li>')
        else:
            first_ch = list(baseline_structure[bn].keys())[0]
            dest = get_filename(bn, first_ch, active_focus, active_cross)
            html.append(f'<li><a href="{dest}">Book {bn}</a></li>')
    html.append("</ul></div>")
    return "\n".join(html)

# ── Sequential Path Assembly Unpacking ────────────────────────
chapter_list = []
section_list = []
for bn in baseline_structure:
    for cn in baseline_structure[bn]:
        chapter_list.append((bn, cn))
        for sn in baseline_structure[bn][cn]:
            section_list.append((bn, cn, sn))

print("Compiling global alignment matrix layout variants...")
start_time = time.time()
total_compiled = 0

for focus_id in TEXT_REGISTRY:
    for cross_id in TEXT_REGISTRY:
        
        # ==========================================
        # PHASE 1: GENERATE CHAPTER-LEVEL SEPARATE FILES
        # ==========================================
        for ch_idx, (book_n, chap_n) in enumerate(chapter_list):
            prev_html, next_html = '&#x25C0;', '&#x25B6;'
            if ch_idx > 0:
                pb, pc = chapter_list[ch_idx - 1]
                prev_html = f'<a href="{get_filename(pb, pc, focus_id, cross_id)}">&#x25C0; Prev Chapter ({pc})</a>'
            if ch_idx < len(chapter_list) - 1:
                nb, nc = chapter_list[ch_idx + 1]
                next_html = f'<a href="{get_filename(nb, nc, focus_id, cross_id)}">Next Chapter ({nc}) &#x25B6;</a>'
                
            nav_html = f'<div class="nav-arrows">{prev_html} <span style="margin:0 auto;color:#666;">Chapter Context Tier</span> {next_html}</div>'

            focus_html = render_sections_block(book_n, chap_n, focus_id, focus_id, cross_id)
            cross_html = render_sections_block(book_n, chap_n, cross_id, focus_id, cross_id)

            config_panel = render_layout_configurator(book_n, chap_n, focus_id, cross_id)
            browse_bar = render_top_navigation_matrix(book_n, chap_n, focus_id, cross_id, routing_tier='chapter')
            sidebar_toc = render_sidebar_toc(book_n, chap_n, None, focus_id, cross_id, routing_tier='chapter')
            
            page_content = f"""<!DOCTYPE html>
<html>
<head><meta charset="UTF-8"><title>Aristotle Poetics Ch. {chap_n}</title><style>{PERSEUS_CSS}</style></head>
<body>
<div id="app-view-root">
  <div id="header-container">
    <div id="perseus-banner"><h1><a href="{get_filename('1', '1', 'grc_kassel', 'eng_fyfe')}">Perseus Digital Library</a></h1><span class="doc-title">Aristotle, Poetics Workspace</span></div>
    <div id="nav-bar"><a href="{get_filename('1', '1', 'grc_kassel', 'eng_fyfe')}">Home Workspace</a></div>
    {browse_bar}
  </div>
  <div id="outer-wrapper">
    {sidebar_toc}
    <div id="main-container">
      <div id="focus-text">
        {config_panel}
        <div class="tier-indicator">Chapter View Mode (Click left numbers to isolate section view)</div>
        {nav_html}
        <h2 style="font-size:14px;color:#660000;margin-bottom:10px;">Aristotle, Poetics: Chapter {chap_n}</h2>
        {focus_html}
        {nav_html}
      </div>
      <div id="cross-ref"><div class="panel-header">{TEXT_REGISTRY[cross_id]["label"]}</div>{cross_html}</div>
    </div>
  </div>
</div>
</body>
</html>"""
            (OUT_DIR / get_filename(book_n, chap_n, focus_id, cross_id)).write_text(page_content, encoding='utf-8')
            total_compiled += 1

        # ==========================================
        # PHASE 2: GENERATE SECTION-LEVEL SEPARATE FILES
        # ==========================================
        for sec_idx, (book_n, chap_n, sec_n) in enumerate(section_list):
            prev_html, next_html = '&#x25C0;', '&#x25B6;'
            if sec_idx > 0:
                pb, pc, ps = section_list[sec_idx - 1]
                prev_html = f'<a href="{get_filename(pb, pc, focus_id, cross_id, ps)}">&#x25C0; Prev Section ({pc}.{ps})</a>'
            if sec_idx < len(section_list) - 1:
                nb, nc, ns = section_list[sec_idx + 1]
                next_html = f'<a href="{get_filename(nb, nc, focus_id, cross_id, ns)}">Next Section ({nc}.{ns}) &#x25B6;</a>'
                
            nav_html = f'<div class="nav-arrows">{prev_html} <span style="margin:0 auto;color:#666;">Isolated Section Tier</span> {next_html}</div>'

            focus_html = render_sections_block(book_n, chap_n, focus_id, focus_id, cross_id, is_isolated=True, isolated_sec_n=sec_n)
            cross_html = render_sections_block(book_n, chap_n, cross_id, focus_id, cross_id, is_isolated=True, isolated_sec_n=sec_n)

            config_panel = render_layout_configurator(book_n, chap_n, focus_id, cross_id, sec_n)
            browse_bar = render_top_navigation_matrix(book_n, chap_n, focus_id, cross_id, routing_tier='section')
            sidebar_toc = render_sidebar_toc(book_n, chap_n, sec_n, focus_id, cross_id, routing_tier='section')
            
            page_content = f"""<!DOCTYPE html>
<html>
<head><meta charset="UTF-8"><title>Aristotle Poetics Sec. {chap_n}.{sec_n}</title><style>{PERSEUS_CSS}</style></head>
<body>
<div id="app-view-root">
  <div id="header-container">
    <div id="perseus-banner"><h1><a href="{get_filename('1', '1', 'grc_kassel', 'eng_fyfe')}">Perseus Digital Library</a></h1><span class="doc-title">Aristotle, Poetics Workspace</span></div>
    <div id="nav-bar"><a href="{get_filename('1', '1', 'grc_kassel', 'eng_fyfe')}">Home Workspace</a> | <a href="{get_filename(book_n, chap_n, focus_id, cross_id)}" style="color:#660000;">&uarr; Up to Full Chapter View</a></div>
    {browse_bar}
  </div>
  <div id="outer-wrapper">
    {sidebar_toc}
    <div id="main-container">
      <div id="focus-text">
        {config_panel}
        <div class="tier-indicator" style="background:#e3f2fd; border-color:#90caf9;">Isolated Section View Mode</div>
        {nav_html}
        <h2 style="font-size:14px;color:#660000;margin-bottom:10px;">Aristotle, Poetics: Chapter {chap_n}, Section {sec_n}</h2>
        {focus_html}
        {nav_html}
      </div>
      <div id="cross-ref"><div class="panel-header">{TEXT_REGISTRY[cross_id]["label"]}</div>{cross_html}</div>
    </div>
  </div>
</div>
</body>
</html>"""
            (OUT_DIR / get_filename(book_n, chap_n, focus_id, cross_id, sec_n)).write_text(page_content, encoding='utf-8')
            total_compiled += 1

print(f"Success! Generated {total_compiled} standalone matrix view files with isolated viewport scroll containers.")
```
