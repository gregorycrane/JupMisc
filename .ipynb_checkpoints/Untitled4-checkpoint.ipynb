```python
import re

for i in range(1,448):
    if(i<10):
        curnum = '00' + str(i)
    elif(i<100):
        curnum = '0' + str(i)
    else:
        curnum =   str(i)

    fname= 'bsb11724902_ocr_pages/page_' + curnum + '.txt'
    outfname = re.sub('page_','opage_',fname)
    outf = open(outfname,'w',, encoding="utf-8")
    f = open(fname)
    for l in f:
        l = re.sub('\n+$','',l)
        l = re.sub('<span class="ocr_line"[^>]+>','\n',l)
        l = re.sub('<[^>]+>','',l)
        print(l,file=outf)
    f.close()
    outf.close()
    print(fname)
```
