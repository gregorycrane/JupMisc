```python
from lxml import etree
from sentence_transformers import SentenceTransformer, util
import numpy as np

# Load TEI
tree = etree.parse("/Users/gcrane/github/GRC_misc/heike.sadler1918.xml")
root = tree.getroot()

# Extract tokens (very basic; can adjust for your TEI structure)
tokens = root.xpath("//text()")  # or use //w for tokenized texts

# Load embedding model
model = SentenceTransformer('all-mpnet-base-v2')

# Emotion concept descriptions
emotion_prompts = {
    "sadness": "a feeling of sorrow, grief, or deep unhappiness",
    "anger": "a feeling of rage, hostility, or fury",
    "joy": "a feeling of happiness, delight, or pleasure",
    "fear": "a feeling of dread, anxiety, or terror",
    "desire": "a feeling of longing or wanting something strongly"
}

emotion_embs = {
    e: model.encode(desc, convert_to_tensor=True)
    for e, desc in emotion_prompts.items()
}

results = []

for token in tokens:
    tok_emb = model.encode(token, convert_to_tensor=True)
    # Compute similarity to all emotions
    sims = {e: float(util.cos_sim(tok_emb, emb)) for e, emb in emotion_embs.items()}
    # Choose best
    best_emotion = max(sims, key=sims.get)
    if sims[best_emotion] > 0.45:   # empirical threshold
        results.append((token, best_emotion, sims[best_emotion]))
```

```python


def find_emotion_tokens(tei_file, threshold=0.45):
    """
    Parse a TEI XML file, extract text tokens, and identify emotion-bearing words
    using Sentence Transformers semantic similarity.

    Parameters
    ----------
    tei_file : str
        Path to the TEI XML file.
    threshold : float
        Cosine similarity threshold for accepting an emotion match.

    Returns
    -------
    list of tuples
        (token, emotion_label, similarity_score)
    """
    
    # Load TEI
    tree = etree.parse(tei_file)
    root = tree.getroot()

    # Extract tokens (adjust as needed for your TEI structure)
    tokens = root.xpath("//text()")

    # Load embedding model (best accuracy)
    model = SentenceTransformer('all-mpnet-base-v2')

    # Emotion concept descriptions
    emotion_prompts = {
        "sadness": "a feeling of sorrow, grief, or deep unhappiness",
        "anger": "a feeling of rage, hostility, or fury",
        "joy": "a feeling of happiness, delight, or pleasure",
        "fear": "a feeling of dread, anxiety, or terror",
        "desire": "a feeling of longing or wanting something strongly"
    }

    # Precompute embeddings for each emotion concept
    emotion_embs = {
        e: model.encode(desc, convert_to_tensor=True)
        for e, desc in emotion_prompts.items()
    }

    results = []

    # Evaluate each token
    for token in tokens:
        tok_emb = model.encode(token, convert_to_tensor=True)

        # Compute similarity to each emotion category
        sims = {e: float(util.cos_sim(tok_emb, emb)) for e, emb in emotion_embs.items()}
        best_emotion = max(sims, key=sims.get)

        if sims[best_emotion] > threshold:
            results.append((token, best_emotion, sims[best_emotion]))

    return results
```

```python
import torch
from transformers import BertTokenizer, BertModel
from lxml import etree
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

def find_emotions_contextual_bert(tei_file, threshold=0.25):
    """
    Use contextual BERT token embeddings to identify emotion-bearing tokens in TEI XML.

    Parameters
    ----------
    tei_file : str
        Path to TEI XML file.
    threshold : float
        Minimal cosine similarity score to count a token as emotion-related.

    Returns
    -------
    list of dict:
        [{ "token": ..., "emotion": ..., "score": ..., "sentence": ... }, ...]
    """

    # Load BERT (contextual encoder)
    tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")
    model = BertModel.from_pretrained("bert-base-uncased")
    model.eval()

    # Parse TEI XML
    tree = etree.parse(tei_file)
    root = tree.getroot()

    # Extract sentences (you can refine this later)
    sentences = root.xpath("//text()")

    # Emotion category prompts
    emotion_prompts = {
        "sadness": "a feeling of sorrow grief lament unhappiness melancholy despair",
        "anger":   "anger rage hostility wrath fury resentment",
        "joy":     "joy happiness delight pleasure gladness",
        "fear":    "fear dread anxiety terror apprehension",
        "desire":  "desire longing yearning wanting craving"
    }

    # Compute embedding for each emotion prompt
    with torch.no_grad():
        emotion_vecs = {}
        for emo, text in emotion_prompts.items():
            inputs = tokenizer(text, return_tensors="pt")
            outputs = model(**inputs)
            # Mean-pool last hidden states
            emb = outputs.last_hidden_state.mean(dim=1).numpy()
            emotion_vecs[emo] = emb

    results = []

    # Process each sentence
    for sent in sentences:
        sent = sent.strip()
        if not sent:
            continue

        # Tokenize sentence into WordPiece tokens
        inputs = tokenizer(sent, return_tensors="pt", truncation=True)
        with torch.no_grad():
            outputs = model(**inputs)

        # Contextual token embeddings
        token_embeddings = outputs.last_hidden_state.squeeze(0).numpy()
        tokens = tokenizer.convert_ids_to_tokens(inputs["input_ids"].squeeze(0))

        # Score each token
        for tok, vec in zip(tokens, token_embeddings):

            # Skip CLS / SEP
            if tok in ["[CLS]", "[SEP]"]:
                continue

            # Compute similarity to each emotion category
            sims = {
                emo: float(cosine_similarity(vec.reshape(1, -1), emo_vec)[0][0])
                for emo, emo_vec in emotion_vecs.items()
            }

            best_emo = max(sims, key=sims.get)
            best_score = sims[best_emo]

            if best_score >= threshold:
                results.append({
                    "token": tok,
                    "emotion": best_emo,
                    "score": best_score,
                    "sentence": sent
                })

    return results
```
