```python
import requests
import time
import os

# Base URL template
base_url = "https://api.digitale-sammlungen.de/ocr/bsb11263667/{}"

# Output folder
output_folder = "dindorfsophlex_ocr_pages"
os.makedirs(output_folder, exist_ok=True)

# Page range
start_page = 1
end_page = 552

for page in range(start_page, end_page + 1):
    url = base_url.format(page)
    print(f"Downloading page {page} from {url}...")

    try:
        response = requests.get(url)
        response.raise_for_status()

        output_path = os.path.join(output_folder, f"{page:03}.txt")
        with open(output_path, "w", encoding="utf-8") as f:
            f.write(response.text)

        print(f"Saved to {output_path}")
    except requests.HTTPError as e:
        print(f"Failed to download page {page}: {e}")

    time.sleep(1)
```

```python
import requests
import time
import os
from bs4 import BeautifulSoup
import unicodedata

base_url = "https://api.digitale-sammlungen.de/ocr/bsb10993766/{}"
output_folder = "dindsophreview"
os.makedirs(output_folder, exist_ok=True)

start_page = 1
end_page = 24

for page in range(start_page, end_page + 1):
    url = base_url.format(page)
    print(f"Downloading page {page} from {url}...")

    try:
        response = requests.get(url)
        response.raise_for_status()
        response.encoding = 'utf-8'

        soup = BeautifulSoup( unicodedata.normalize("NFC", response.text), "html.parser")

        # Extract each OCR line and preserve line breaks
        lines = []
        for line in soup.find_all("span", class_="ocr_line"):
            line_text = line.get_text(separator=" ", strip=True)
            if line_text:
                lines.append(line_text)

        output_path = os.path.join(output_folder, f"page_{page:03}.txt")
        with open(output_path, "w", encoding="utf-8") as f:
            f.write("\n".join(lines))  # <-- insert line breaks explicitly!

        print(f"Saved page {page} with line breaks to {output_path}")

    except requests.HTTPError as e:
        print(f"Failed to download page {page}: {e}")

    time.sleep(1)
```

```python
import sys
from bs4 import BeautifulSoup

# --- Configuration ---
# This X-coordinate is used to determine the split point between the two columns.
# Based on the HOCR data, a value around 950 pixels effectively separates the columns.
COLUMN_SEPARATOR_X = 950

def fix_encoding(text: str) -> str:
    """
    Corrects garbled UTF-8 text that was incorrectly read as a single-byte encoding.
    
    Args:
        text: The input string with potential encoding errors (e.g., 'Î±Î¹ÏƒÏ‡Ïá¿¶Ï‚').

    Returns:
        The corrected string (e.g., 'αἰσχρῶς'). Returns original text if decoding fails.
    """
    try:
        # This sequence of encoding and decoding reverses the common error
        # of misinterpreting UTF-8 bytes as a single-byte encoding like latin-1.
        return text.encode('latin1').decode('utf-8')
    except (UnicodeEncodeError, UnicodeDecodeError):
        # If the text is already correct, the operation may fail. Return the original.
        return text

def process_hocr_file(file_path: str):
    """
    Parses an HOCR file, corrects Greek character encoding, separates the two
    columns based on word coordinates, and prints the result to the console.

    Args:
        file_path: The path to the HOCR HTML file.
    """
    try:
        with open(file_path, 'r', encoding='utf-8') as f:
            soup = BeautifulSoup(f, 'html.parser')
    except FileNotFoundError:
        print(f"Error: The file '{file_path}' was not found.", file=sys.stderr)
        sys.exit(1)

    # Find all line elements in the HOCR file
    lines = soup.find_all('span', class_='ocr_line')

    for line in lines:
        left_col_words = []
        right_col_words = []

        # Find all word elements within the current line
        words = line.find_all('span', class_='ocrx_word')

        for word in words:
            # The 'title' attribute contains metadata, including the bounding box (bbox).
            title_attr = word.get('title', '')
            if 'bbox' not in title_attr:
                continue

            try:
                # Extract the x0 coordinate from the bbox string (e.g., "bbox 189 297 315 333")
                bbox_parts = title_attr.split(';')
                bbox_str = [part for part in bbox_parts if 'bbox' in part][0]
                coords = bbox_str.strip().split(' ')
                x0 = int(coords[1])
                
                # Correct the text encoding
                word_text = fix_encoding(word.get_text())

                # Assign the word to the correct column based on its horizontal position
                if x0 < COLUMN_SEPARATOR_X:
                    left_col_words.append(word_text)
                else:
                    right_col_words.append(word_text)
            
            except (ValueError, IndexError):
                # Skip words with malformed bbox data
                continue
        
        # Join the words in each column list into a single string
        left_text = " ".join(left_col_words)
        right_text = " ".join(right_col_words)

        # Print the formatted line, handling cases where one or both columns are empty
        if left_text and right_text:
            print(f"{left_text} [COLUMN 2 START] {right_text}")
        elif left_text:
            print(left_text)
        elif right_text:
            print(f"[COLUMN 2 START] {right_text}")

hocr_file_path = '/Users/gcrane/github/JupMisc/dindorfsophlex_ocr_pages/212.txt'
process_hocr_file(hocr_file_path)
```

```python
import sys
from bs4 import BeautifulSoup

# --- Configuration ---
# X-coordinate to separate columns.
COLUMN_SEPARATOR_X = 950

# --- Layout Analysis Configuration ---
# The minimum number of words a column must have to be considered valid.
MIN_WORDS_PER_COLUMN = 10
# The allowable ratio between the widths of the two columns (e.g., 0.7 means
# the left column can be down to 70% of the right column's width).
WIDTH_RATIO_MIN = 0.7
WIDTH_RATIO_MAX = 1.4

def fix_encoding(text: str) -> str:
    """
    Corrects garbled UTF-8 text that was incorrectly read as a single-byte encoding.
    """
    try:
        return text.encode('latin1').decode('utf-8')
    except (UnicodeEncodeError, UnicodeDecodeError):
        return text

def get_bbox_coords(element) -> tuple[int, int, int, int] | None:
    """
    Parses a 'title' attribute to extract bbox coordinates (x0, y0, x1, y1).
    
    Args:
        element: A BeautifulSoup element.
        
    Returns:
        A tuple of four integers, or None if parsing fails.
    """
    title_attr = element.get('title', '')
    if 'bbox' not in title_attr:
        return None
    try:
        # Find the part of the title string that contains 'bbox ...'
        bbox_part = [part for part in title_attr.split(';') if 'bbox' in part][0]
        # The format is "bbox x0 y0 x1 y1"
        coords = bbox_part.strip().split(' ')
        return tuple(map(int, coords[1:]))
    except (ValueError, IndexError):
        return None

def analyze_page_layout(soup: BeautifulSoup) -> bool:
    """
    Checks if the page has a valid two-column layout.

    A valid layout has words in both columns, and the column widths are comparable.
    
    Args:
        soup: The BeautifulSoup object for the entire page.
        
    Returns:
        True if the page appears to be two-column, False otherwise.
    """
    all_words = soup.find_all('span', class_='ocrx_word')
    if not all_words:
        return False

    left_x_coords, right_x_coords = [], []

    for word in all_words:
        coords = get_bbox_coords(word)
        if not coords:
            continue
        x0, _, x1, _ = coords

        if x0 < COLUMN_SEPARATOR_X:
            left_x_coords.append((x0, x1))
        else:
            right_x_coords.append((x0, x1))

    # 1. Check if both columns have a sufficient number of words.
    if len(left_x_coords) < MIN_WORDS_PER_COLUMN or len(right_x_coords) < MIN_WORDS_PER_COLUMN:
        return False

    # 2. Check if the column widths are roughly the same.
    width_left = max(x[1] for x in left_x_coords) - min(x[0] for x in left_x_coords)
    width_right = max(x[1] for x in right_x_coords) - min(x[0] for x in right_x_coords)

    if width_left <= 0 or width_right <= 0:
        return False

    ratio = width_left / width_right
    return WIDTH_RATIO_MIN < ratio < WIDTH_RATIO_MAX

def process_hocr_file(file_path: str,outf):
    """
    Parses an HOCR file, validates layout, corrects encoding, sorts lines by
    Y-coordinate, and prints the re-ordered two-column text.
    
    Args:
        file_path: The path to the HOCR HTML file.
    """
    try:
        with open(file_path, 'r', encoding='utf-8') as f:
            soup = BeautifulSoup(f, 'html.parser')
    except FileNotFoundError:
        print(f"Error: The file '{file_path}' was not found.", file=sys.stderr)
        sys.exit(1)

    if not analyze_page_layout(soup):
        print("--- Single-column layout detected. Printing content as is. ---", file=sys.stderr)
        lines = soup.find_all('span', class_='ocr_line')
        for line in lines:
            words = [fix_encoding(word.get_text()) for word in line.find_all('span', class_='ocrx_word')]
            print(" ".join(words),file=outf)
        return

    print("--- Two-column layout detected. Processing... ---", file=sys.stderr)
    left_column_content = []
    right_column_content = []

    lines = soup.find_all('span', class_='ocr_line')
    for line in lines:
        line_coords = get_bbox_coords(line)
        # Use the line's y-coordinate as the primary sorting key.
        # If a line has no coordinate, try to get it from the first word.
        if not line_coords:
            first_word = line.find('span', class_='ocrx_word')
            if first_word:
                line_coords = get_bbox_coords(first_word)
            else:
                continue # Skip empty lines
        
        y0 = line_coords[1]

        left_col_words = []
        right_col_words = []
        words = line.find_all('span', class_='ocrx_word')

        for word in words:
            word_coords = get_bbox_coords(word)
            if not word_coords:
                continue
            
            word_x0 = word_coords[0]
            word_text = fix_encoding(word.get_text())

            if word_x0 < COLUMN_SEPARATOR_X:
                left_col_words.append(word_text)
            else:
                right_col_words.append(word_text)

        left_text = " ".join(left_col_words)
        right_text = " ".join(right_col_words)

        if left_text:
            left_column_content.append((y0, left_text))
        if right_text:
            right_column_content.append((y0, right_text))

    # Sort each column's lines independently based on their vertical position.
    left_column_content.sort(key=lambda item: item[0])
    right_column_content.sort(key=lambda item: item[0])

    # Print the merged and sorted content.
    num_lines = max(len(left_column_content), len(right_column_content))
    for i in range(num_lines):
        left_part = left_column_content[i][1] if i < len(left_column_content) else ""
        right_part = right_column_content[i][1] if i < len(right_column_content) else ""

        if left_part and right_part:
            print(f"{left_part} [Column 2.{i+1} START] {right_part}",file=outf)
        elif right_part:
            print(f"[Column 2.{i+1} START] {right_part}",file=outf)
        elif left_part:
            print(left_part,file=outf)

outf = open('dindorf-sophocles-lexicon1.txt','w')
for i in range(1,533):
    print('\n\n<pb n="'+str(i)+'"/>',file=outf)
    padded_str = f"{i:03d}"
    hocr_file_path = '/Users/gcrane/github/JupMisc/dindorfsophlex_ocr_pages/'+padded_str+'.txt'
    process_hocr_file(hocr_file_path,outf)
```
