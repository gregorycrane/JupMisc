```python
import re
from lxml import etree
import html
import os

# --- Configuration ---
# Directory containing your XML files
XML_DIR = '../GRC_misc/'
# Directory where you want to save the HTML output
OUTPUT_DIR = './html_output_hermann/'
# Ensure the output directory exists
os.makedirs(OUTPUT_DIR, exist_ok=True)

# List of play identifiers (used to build filenames)
# Add other play IDs if you have corresponding text and app crit files
PLAY_IDS = ['ag'] # Currently only processing Agamemnon based on provided files

# XML namespaces (though not strictly needed for these simpler files)
ns = {'tei': 'http://www.tei-c.org/ns/1.0'} # Kept for potential future use

print(f"Processing plays: {', '.join(PLAY_IDS)}")
print(f"HTML output will be saved in '{OUTPUT_DIR}'.")
```

```python
# --- Helper Function ---
def get_node_html_content(node):
    """Recursively get text content, handling specific tags like note, gap, stage."""
    parts = []
    if node.text:
        parts.append(html.escape(node.text))
    for child in node:
        if child.tag == 'note' and child.get('type') == 'editorial':
            parts.append(f'<span class="editorial-note">{html.escape("".join(child.itertext()))}</span>')
        elif child.tag == 'gap':
            parts.append('<span class="gap">[...]</span>') # Represent gap
        elif child.tag == 'stage':
             parts.append(f'<span class="stage-direction">({html.escape("".join(child.itertext()))})</span>')
        elif child.tag == 'lb': # Handle line breaks within app crit notes if needed
             parts.append('<br>')
        elif child.tag == 'lem':
             parts.append(f'<span class="app-lemma">{html.escape("".join(child.itertext()))}</span>')
        elif child.tag == 'quote':
             parts.append(f'<span class="app-quote">"{html.escape("".join(child.itertext()))}"</span>')
        else:
            # Recursively process children or get text
             parts.append(get_node_html_content(child)) # Recursive call for nested tags

        if child.tail:
            parts.append(html.escape(child.tail))
    return "".join(parts)
```

```python
# --- Core Processing Function ---
def process_play(text_xml_filepath, app_crit_xml_filepath, html_output_filepath):
    """
    Parses Aeschylus Hermann TEI XML files (text and app crit),
    extracts relevant data, and generates a two-column HTML file.
    """
    print(f"\nProcessing '{os.path.basename(text_xml_filepath)}' and '{os.path.basename(app_crit_xml_filepath)}'...")
    try:
        # --- 1. Parsing ---
        parser = etree.XMLParser(remove_blank_text=True)
        text_root = etree.parse(text_xml_filepath, parser)
        app_root = etree.parse(app_crit_xml_filepath, parser)
        print("   - XML files parsed.")

        # --- 2. Data Extraction ---
        greek_col = {}
        app_crit_col = {}

        # Extract Greek Text
        # Using xpath to find all <l> elements with an 'n' attribute anywhere
        for line in text_root.xpath('.//l[@n]'):
            line_num_str = line.get('n')
            if line_num_str and line_num_str.isdigit(): # Ensure line number is valid
                # Use helper to reconstruct line text considering notes, gaps, stages
                line_content_html = get_node_html_content(line).strip()

                # Find speaker if it's the first line of a speech
                speaker_html = ''
                # Check if the line is the first child <l> within its parent <sp>
                parent_sp = line.find('..')
                if parent_sp is not None and parent_sp.tag == 'sp':
                     first_l = parent_sp.find('l')
                     # Check if this line is the first <l> element found
                     if first_l is line:
                         speaker_tag = parent_sp.find('speaker')
                         if speaker_tag is not None and speaker_tag.text:
                              speaker_html = f'<span class="speaker">{html.escape(speaker_tag.text.strip())}</span> '

                html_content = f'<div class="line" id="g-{line_num_str}"><a class="line-num" data-line="{line_num_str}">{line_num_str}</a> <span class="text">{speaker_html}{line_content_html}</span></div>'
                greek_col[line_num_str] = html_content

        # Extract App Crit
        # Using xpath to find all <p> elements with an 'n' attribute anywhere
        for p_note in app_root.xpath('.//p[@n]'):
            line_num_str = p_note.get('n')
             # Handle potential ranges like "10-11" - apply note to all lines in range
            line_nums = []
            if line_num_str:
                if '-' in line_num_str:
                    try:
                        start, end = map(int, line_num_str.split('-'))
                        line_nums = [str(i) for i in range(start, end + 1)]
                    except ValueError:
                         if line_num_str.isdigit(): # If range parsing fails but it's a digit
                              line_nums = [line_num_str]
                         else:
                              print(f"     - Warning: Skipping invalid app crit line reference '{line_num_str}'")
                              continue # Skip this entry if line number is invalid
                elif line_num_str.isdigit():
                     line_nums = [line_num_str]
                else:
                    print(f"     - Warning: Skipping invalid app crit line reference '{line_num_str}'")
                    continue # Skip invalid non-numeric, non-range entries


            if line_nums:
                # Use helper function to get potentially formatted text
                app_text = get_node_html_content(p_note).strip()
                if not app_text: # Skip empty notes
                     continue

                for num_str in line_nums:
                     html_content = f'<div class="line" id="app-{num_str}"><a class="line-num" data-line="{num_str}">{num_str}</a> <span class="text">{app_text}</span></div>'
                     app_crit_col.setdefault(num_str, []).append(html_content)


        print(f"   - Data extracted: {len(greek_col)} Greek lines, {sum(len(v) for v in app_crit_col.values())} app crit entries.")

        # --- 3. HTML Generation ---
        max_line_num = 0
        all_keys = list(greek_col.keys()) + list(app_crit_col.keys())
        if all_keys:
            valid_keys = [int(k) for k in all_keys if k.isdigit()]
            if valid_keys: max_line_num = max(valid_keys)

        # Generate Greek HTML (include placeholders for missing lines)
        greek_html_lines = []
        for i in range(1, max_line_num + 1):
            line_key = str(i)
            greek_html_lines.append(greek_col.get(line_key, f'<div class="line empty" id="g-{line_key}"><a class="line-num" data-line="{line_key}">{line_key}</a> <span class="text"></span></div>'))
        greek_html = "\n".join(greek_html_lines)


        # Generate App Crit HTML (only include lines with data)
        app_crit_lines = []
        # Sort keys numerically before iterating to maintain order
        sorted_app_keys = sorted([k for k in app_crit_col.keys() if k.isdigit()], key=int)
        for line_key in sorted_app_keys:
            if line_key in app_crit_col:
                app_crit_lines.extend(app_crit_col[line_key])
        app_crit_html = "\n".join(app_crit_lines)


        # Metadata for Header (Partly hardcoded for Hermann edition)
        title_tag = text_root.find('.//teiHeader/titleStmt/title', ns)
        main_title = title_tag.text if title_tag is not None else os.path.basename(text_xml_filepath) # Fallback title
        author = "Aeschylus" # Assume Aeschylus
        editor = "Gottfried Hermann"
        pub_date = "1852"


        # --- HTML Template (2 Columns) ---
        # NOTE: Escaped curly braces {{ }} in the <script> block for f-string formatting
        html_template = f"""
        <!DOCTYPE html>
        <html lang="en">
        <head>
            <meta charset="UTF-8">
            <meta name="viewport" content="width=device-width, initial-scale=1.0">
            <title>{main_title} (Hermann Edition)</title>
            <style>
                body {{
                    font-family: 'Georgia', serif; display: flex; flex-direction: column;
                    height: 100vh; margin: 0; background-color: #fdfdfd;
                }}
                header {{
                    padding: 10px 20px; border-bottom: 2px solid #ddd;
                    background-color: #fff; text-align: center; flex-shrink: 0;
                }}
                h1 {{ margin: 0; font-size: 1.8em; color: #333; }}
                h2 {{ margin: 5px 0 0; font-size: 1.1em; color: #555; font-weight: normal;}}
                .container {{
                    display: grid; grid-template-columns: repeat(2, 1fr); /* Two columns */
                    gap: 15px; /* Slightly increased gap */
                    flex-grow: 1; padding: 15px; overflow: hidden; /* Ensure container handles overflow */
                }}
                .column {{
                    background-color: #ffffff; border: 1px solid #e0e0e0; border-radius: 4px;
                    display: flex; flex-direction: column; overflow: hidden; /* Prevent content spill */
                }}
                .column h3 {{
                    text-align: center; margin: 0; padding: 12px;
                    border-bottom: 1px solid #e0e0e0; background-color: #f0f0f0; /* Slightly darker header */
                    color: #444; font-size: 1em; font-weight: bold;
                    flex-shrink: 0;
                }}
                .content {{
                     padding: 10px 15px;
                     overflow-y: auto; /* Enable vertical scroll WITHIN column */
                     height: 100%; /* Allow content div to fill column space */
                     line-height: 1.6; /* Improve readability */
                     flex-grow: 1;
                }}
                .line {{
                    display: flex; align-items: baseline; padding: 2px 5px; /* Reduced vertical padding */
                    border-radius: 3px;
                     margin-bottom: 2px; /* Small space between lines */
                     /* Greek column lines have a min-height */
                     min-height: 1.5em;
                }}
                /* Remove min-height for App Crit where lines might be skipped */
                #app-crit-content .line {{
                     min-height: auto;
                     align-items: flex-start; /* Align numbers top for multi-line app crit */
                }}
                 /* Style specifically for empty lines in the Greek column */
                #greek-content .line.empty .text {{
                     color: #ccc; /* Make empty placeholders less prominent */
                }}

                .line.highlight {{ background-color: #e7f5ff; }}
                .line-num {{
                    flex-shrink: 0; width: 40px; font-size: 0.75em; /* Smaller line numbers */
                    color: #999; /* Lighter numbers */
                    cursor: pointer; text-align: right; margin-right: 12px; /* Increased margin */
                    font-family: monospace;
                    padding-top: 0.2em; /* Align better with first line of text */
                }}
                .line-num:hover {{ color: #007bff; text-decoration: underline; }}
                .text {{ flex-grow: 1; }} /* Allow text to take remaining space */
                .speaker {{ font-weight: bold; margin-right: 8px; color: #800000; }}
                /* App Crit specific styles */
                .app-lemma {{ font-style: italic; color: #0056b3; background-color: #f0f8ff; padding: 0 3px; border-radius: 2px;}}
                .app-quote {{ font-family: monospace; color: #444; }}
                .editorial-note {{ color: #006400; font-style: italic; }} /* Style for editorial notes */
                .gap {{ color: #999; font-style: italic; }} /* Style for gaps */
                 .stage-direction {{ color: #777; font-style: italic; font-size: 0.9em; }}
                 /* Ensure app crit text wraps */
                 #app-crit-content .text {{
                      word-break: break-word;
                      white-space: normal; /* Override potential nowrap */
                 }}

            </style>
        </head>
        <body>
            <header>
                <h1>{main_title}</h1>
                <h2>{author} | Edited by: {editor} ({pub_date})</h2>
            </header>
            <div class="container">
                <div class="column">
                    <h3>Greek Text</h3>
                    <div class="content" id="greek-content">{greek_html}</div>
                </div>
                <div class="column">
                    <h3>Apparatus Criticus</h3>
                    <div class="content" id="app-crit-content">{app_crit_html}</div>
                </div>
            </div>
            <script>
                document.addEventListener('DOMContentLoaded', function() {{
                    const container = document.querySelector('.container');
                    const contents = document.querySelectorAll('.content'); // Get all content divs
                    let lastClickedLine = null;
                    let syncScroll = false; // Flag to prevent scroll loops

                    container.addEventListener('click', function(event) {{
                        if (event.target.classList.contains('line-num')) {{
                            event.preventDefault(); // Prevent default anchor behavior
                            const lineNumber = event.target.dataset.line;

                            // Remove previous highlights
                            if(lastClickedLine) {{
                                document.querySelectorAll(`.line.highlight`).forEach(el => el.classList.remove('highlight'));
                            }}

                            const prefixes = ['g', 'app']; // Only two columns
                            let firstElementFound = null; // To store the first element to scroll to

                            prefixes.forEach(prefix => {{
                                const elementId = `${{prefix}}-${{lineNumber}}`;
                                const element = document.getElementById(elementId);
                                if (element) {{
                                    element.classList.add('highlight');
                                    // Smooth scroll the first relevant element into view
                                    if (!firstElementFound) {{
                                         firstElementFound = element;
                                         // Scroll the column's content div, not the element itself directly
                                          element.parentElement.scrollTo({{
                                               top: element.offsetTop - element.parentElement.offsetTop - 50, // Adjust offset as needed
                                               behavior: 'smooth'
                                           }});
                                    }}
                                }}
                            }});
                            lastClickedLine = lineNumber;
                        }}
                    }});

                 // Basic scroll synchronization (optional, can be complex)
                 // This is a simplified version and might not be perfectly smooth
                 contents.forEach(content => {{
                      content.addEventListener('scroll', () => {{
                           if (syncScroll) return; // Prevent loop
                           syncScroll = true;
                           // Calculate scroll percentage only if scrollHeight is greater than clientHeight
                           const scrollableHeight = content.scrollHeight - content.clientHeight;
                           const scrollPercentage = scrollableHeight > 0 ? content.scrollTop / scrollableHeight : 0;
                           
                           contents.forEach(otherContent => {{
                                if (otherContent !== content) {{
                                     const otherScrollableHeight = otherContent.scrollHeight - otherContent.clientHeight;
                                     if (otherScrollableHeight > 0) {{
                                          otherContent.scrollTop = scrollPercentage * otherScrollableHeight;
                                     }}
                                }}
                           }});
                           // Use timeout to reset the flag after scroll events settle
                           setTimeout(() => {{ syncScroll = false; }}, 50); // Adjust timeout if needed
                      }});
                 }});

                }});
            </script>
        </body>
        </html>
        """

        # --- 4. Write to file ---
        with open(html_output_filepath, 'w', encoding='utf-8') as f:
            f.write(html_template)
        print(f"   - Successfully created '{os.path.basename(html_output_filepath)}'.")

    except FileNotFoundError as e:
        print(f"   - Error: File not found. {e}")
    except etree.XMLSyntaxError as e:
        print(f"   - Error: XML syntax error. {e}")
    except Exception as e:
        print(f"   - Error: An unexpected error occurred. {e}")
        import traceback
        traceback.print_exc() # Print full traceback for unexpected errors
```

```python
# --- Main Execution Loop ---
for play_id in PLAY_IDS:
    # Construct file paths based on Hermann convention
    text_file = f'aesch.{play_id}.hermann1852-2.xml' # Assuming '-2' suffix for text
    app_crit_file = f'aesch.{play_id}.appcrit-hermann1852.xml'
    output_file = f'aeschylus_{play_id}_hermann.html'

    text_xml_path = os.path.join(XML_DIR, text_file)
    app_crit_xml_path = os.path.join(XML_DIR, app_crit_file)
    output_html_path = os.path.join(OUTPUT_DIR, output_file)

    # Check if input files exist before processing
    if not os.path.exists(text_xml_path):
        print(f"\nSkipping '{play_id}': Text file not found at '{text_xml_path}'")
        continue
    if not os.path.exists(app_crit_xml_path):
        print(f"\nSkipping '{play_id}': App crit file not found at '{app_crit_xml_path}'")
        continue

    # Process the current play
    process_play(text_xml_path, app_crit_xml_path, output_html_path)

print("\nProcessing complete for available Hermann plays.")
```
