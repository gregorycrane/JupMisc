This notebook analyzes the XML files from [gregorycrane/LSJLogeion](https://github.com/gregorycrane/lsjlogeion), the Logeion edition of the LSJ Greek-English Lexicon maintained at the University of Chicago.

````python
import os, re, glob, json, unicodedata
from lxml import etree
from collections import Counter, defaultdict
import matplotlib.pyplot as plt
import matplotlib
import numpy as np
import pandas as pd
import seaborn as sns

matplotlib.rcParams['figure.figsize'] = (14, 7)
matplotlib.rcParams['font.size'] = 12
try:
    plt.style.use('seaborn-v0_8-whitegrid')
except OSError:
    try:
        plt.style.use('seaborn-whitegrid')
    except OSError:
        pass  # use default style

REPO_DIR = '/Users/gcrane/github/LSJLogeion'  # adjust path as needed
XML_FILES = sorted(glob.glob(os.path.join(REPO_DIR, 'greatscott*.xml')))
print(f'Found {len(XML_FILES)} XML files')
````

## 1. Parse All XML Files


````python
def parse_lsj_file(filepath):
    """Parse one LSJ XML file and return structured data."""
    parser = etree.XMLParser(recover=True, huge_tree=True)
    tree = etree.parse(filepath, parser)
    root = tree.getroot()
    entries = root.findall('.//div2')
    
    result = []
    for entry in entries:
        key = entry.get('key', '')
        head = entry.find('.//head')
        headword = ''.join(head.itertext()).strip() if head is not None else key
        full_text = ''.join(entry.itertext())
        
        senses = entry.findall('.//sense')
        bibls = entry.findall('.//bibl')
        
        authors = []
        for a in entry.findall('.//author'):
            t = ''.join(a.itertext()).strip()
            if t: authors.append(t)
        
        works = []
        for b in bibls:
            a_el = b.find('.//author')
            t_el = b.find('.//title')
            a_t = ''.join(a_el.itertext()).strip() if a_el is not None else ''
            w_t = ''.join(t_el.itertext()).strip() if t_el is not None else ''
            if a_t and w_t:
                works.append(f'{a_t}, {w_t}')
            elif a_t:
                works.append(a_t)
        
        greek_chars = sum(1 for ch in full_text if 0x0370 <= ord(ch) <= 0x03FF or 0x1F00 <= ord(ch) <= 0x1FFF)
        greek_words_in_foreign = []
        for f in entry.findall('.//foreign'):
            if f.get('lang','') in ('grc','greek'):
                t = ''.join(f.itertext()).strip()
                if t: greek_words_in_foreign.extend(t.split())
        
        quotes = entry.findall('.//quote')
        etyms = entry.findall('.//etym')
        pos_els = entry.findall('.//pos')
        pos_list = [''.join(p.itertext()).strip() for p in pos_els]
        
        dialect_els = entry.findall('.//gram[@type="dialect"]')
        dialects = [''.join(d.itertext()).strip() for d in dialect_els]
        
        sense_levels = [s.get('level','') for s in senses]
        
        result.append({
            'key': key, 'headword': headword,
            'text_len': len(full_text),
            'n_senses': len(senses), 'sense_levels': sense_levels,
            'n_bibls': len(bibls),
            'authors': authors, 'works': works,
            'greek_chars': greek_chars,
            'greek_words': len(greek_words_in_foreign),
            'n_quotes': len(quotes),
            'n_etyms': len(etyms),
            'pos': pos_list, 'dialects': dialects,
        })
    return result

# Parse all files
all_entries = []
for i, f in enumerate(XML_FILES):
    all_entries.extend(parse_lsj_file(f))
    if (i+1) % 10 == 0:
        print(f'  Parsed {i+1}/{len(XML_FILES)} files, {len(all_entries):,} entries so far')
print(f'\nTotal entries parsed: {len(all_entries):,}')

df = pd.DataFrame(all_entries)
````

## 2. Summary Statistics


````python
print('='*60)
print('LSJ LOGEION — SUMMARY STATISTICS')
print('='*60)
print(f'(1) Dictionary entries:        {len(df):,}')
total_bibls = df["n_bibls"].sum()
print(f'(2) Citations to sources:      {total_bibls:,}')
total_greek = df["greek_words"].sum()
total_greek_chars = df["greek_chars"].sum()
print(f'(4) Greek words (in <foreign>):{total_greek:,}')
print(f'    Greek characters (total):   {total_greek_chars:,}')
total_senses = df["n_senses"].sum()
print(f'(5) Word senses:               {total_senses:,}')
print(f'    Unique authors cited:       {len(set(a for al in df["authors"] for a in al)):,}')
print(f'    Entries with ≥1 sense:      {(df["n_senses"]>0).sum():,}')
print(f'    Entries with ≥1 citation:   {(df["n_bibls"]>0).sum():,}')
print(f'    Total quoted passages:      {df["n_quotes"].sum():,}')
print(f'    Etymology elements:         {df["n_etyms"].sum():,}')
print(f'    Avg senses/entry:           {df["n_senses"].mean():.2f}')
print(f'    Avg citations/entry:        {df["n_bibls"].mean():.2f}')
print(f'    Median entry length:        {df["text_len"].median():.0f} chars')
````

## 3. Most Commonly Cited Authors


````python
author_counts = Counter()
for al in df['authors']:
    author_counts.update(al)

top30 = author_counts.most_common(30)
labels, vals = zip(*top30)

fig, ax = plt.subplots(figsize=(14, 8))
bars = ax.barh(range(len(labels)), vals, color=plt.cm.viridis(np.linspace(0.2, 0.9, len(labels))))
ax.set_yticks(range(len(labels)))
ax.set_yticklabels(labels)
ax.invert_yaxis()
ax.set_xlabel('Number of Citations')
ax.set_title('Top 30 Most Cited Authors in LSJ', fontsize=16, fontweight='bold')
for bar, v in zip(bars, vals):
    ax.text(bar.get_width() + 20, bar.get_y() + bar.get_height()/2, f'{v:,}', va='center', fontsize=9)
plt.tight_layout()
plt.show()
````

## 4. Most Commonly Cited Works (Author + Title)


````python
work_counts = Counter()
for wl in df['works']:
    work_counts.update(wl)

top25 = work_counts.most_common(25)
labels, vals = zip(*top25)

fig, ax = plt.subplots(figsize=(14, 8))
bars = ax.barh(range(len(labels)), vals, color=plt.cm.magma(np.linspace(0.2, 0.85, len(labels))))
ax.set_yticks(range(len(labels)))
ax.set_yticklabels(labels)
ax.invert_yaxis()
ax.set_xlabel('Number of Citations')
ax.set_title('Top 25 Most Cited Works in LSJ', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
````

## 5. Entry Length Distribution


````python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

ax1.hist(df['text_len'], bins=100, color='#2ecc71', edgecolor='white', alpha=0.8)
ax1.set_xlabel('Entry Length (characters)')
ax1.set_ylabel('Count')
ax1.set_title('Entry Length Distribution (all entries)')
ax1.set_xlim(0, 2000)

ax2.hist(df['text_len'], bins=np.logspace(1, 4, 50), color='#3498db', edgecolor='white', alpha=0.8)
ax2.set_xscale('log')
ax2.set_xlabel('Entry Length (characters, log scale)')
ax2.set_ylabel('Count')
ax2.set_title('Entry Length Distribution (log scale)')

plt.tight_layout()
plt.show()

# Longest entries
print('Top 30 longest entries:')
for _, row in df.nlargest(30, 'text_len').iterrows():
    print(f'  {row["headword"]:30s}  {row["text_len"]:6,} chars  {row["n_senses"]:3d} senses  {row["n_bibls"]:4d} citations')
````

## 6. Senses per Entry


````python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

sense_dist = df['n_senses'].value_counts().sort_index()
ax1.bar(sense_dist.index[:15], sense_dist.values[:15], color='#e74c3c', edgecolor='white')
ax1.set_xlabel('Number of Senses')
ax1.set_ylabel('Number of Entries')
ax1.set_title('Senses per Entry')

# Polysemy vs entry length
poly = df[df['n_senses'] > 0]
ax2.scatter(poly['n_senses'], poly['text_len'], alpha=0.3, s=10, c='#9b59b6')
ax2.set_xlabel('Number of Senses')
ax2.set_ylabel('Entry Length (chars)')
ax2.set_title('Polysemy vs Entry Length')
ax2.set_yscale('log')

plt.tight_layout()
plt.show()

print(f'Entries with 0 senses:  {(df["n_senses"]==0).sum():,} ({100*(df["n_senses"]==0).mean():.1f}%)')
print(f'Entries with 1 sense:   {(df["n_senses"]==1).sum():,} ({100*(df["n_senses"]==1).mean():.1f}%)')
print(f'Entries with 2-5:       {((df["n_senses"]>=2)&(df["n_senses"]<=5)).sum():,}')
print(f'Entries with 6+:        {(df["n_senses"]>=6).sum():,}')
print(f'Most polysemous entries:')
for _, row in df.nlargest(10, 'n_senses').iterrows():
    print(f'  {row["headword"]:25s} {row["n_senses"]:3d} senses')
````

## 7. Dialect Tags


````python
dialect_counts = Counter()
for dl in df['dialects']:
    dialect_counts.update(dl)

if dialect_counts:
    top_d = dialect_counts.most_common(15)
    labels, vals = zip(*top_d)
    fig, ax = plt.subplots(figsize=(10, 6))
    ax.bar(labels, vals, color='#1abc9c', edgecolor='white')
    ax.set_ylabel('Occurrences')
    ax.set_title('Dialect Labels in LSJ', fontsize=14, fontweight='bold')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
````

## 8. Additional Lexicographic Statistics

These metrics are commonly used in computational analysis of machine-readable dictionaries.


````python
# Hapax citanda — entries with exactly 1 citation
hapax = (df['n_bibls'] == 1).sum()
print(f'Hapax citanda (entries with exactly 1 citation): {hapax:,} ({100*hapax/len(df):.1f}%)')

# Cross-reference density
# (entries mentioning other headwords — approximate via short entries with no senses)
xref_like = ((df['n_senses'] == 0) & (df['text_len'] < 60)).sum()
print(f'Short cross-ref entries (no senses, <60 chars): {xref_like:,} ({100*xref_like/len(df):.1f}%)')

# Etymology coverage
has_etym = (df['n_etyms'] > 0).sum()
print(f'Entries with etymology elements: {has_etym:,} ({100*has_etym/len(df):.1f}%)')

# Quoted Greek passages
has_quotes = (df['n_quotes'] > 0).sum()
print(f'Entries with quoted passages: {has_quotes:,} ({100*has_quotes/len(df):.1f}%)')

# Citation density (citations per 100 chars of entry text)
df['cit_density'] = df['n_bibls'] / (df['text_len'] / 100)
print(f'\nMean citation density: {df["cit_density"].mean():.2f} citations per 100 chars')

# POS distribution
pos_counts = Counter()
for pl in df['pos']:
    pos_counts.update(pl)
print(f'\nExplicit POS tags found: {sum(pos_counts.values()):,}')
for p, c in pos_counts.most_common(10):
    print(f'  {p:20s} {c:5d}')

# Sense hierarchy depth
all_levels = []
for sl in df['sense_levels']:
    all_levels.extend(sl)
level_counts = Counter(all_levels)
print(f'\nSense hierarchy levels:')
for l, c in sorted(level_counts.items()):
    print(f'  Level {l:>3s}: {c:,}')

# Zipf's law check — author citation frequency
author_counts = Counter()
for al in df['authors']:
    author_counts.update(al)

fig, ax = plt.subplots(figsize=(10, 6))
ranks = np.arange(1, len(author_counts)+1)
freqs = sorted(author_counts.values(), reverse=True)
ax.loglog(ranks, freqs, 'o-', markersize=3, alpha=0.7, color='#e67e22')
ax.set_xlabel('Author Rank (log)')
ax.set_ylabel('Citation Frequency (log)')
ax.set_title("Zipf's Law in LSJ Author Citations", fontsize=14, fontweight='bold')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

## 9. Citation Coverage Heatmap — Author × Century


````python
# Map common abbreviations to approximate centuries
author_century = {
    'Hom.': -8, 'Hes.': -7, 'Pi.': -5, 'A.': -5, 'S.': -5, 'E.': -5,
    'Ar.': -5, 'Hdt.': -5, 'Th.': -5, 'X.': -4, 'Pl.': -4, 'Arist.': -4,
    'D.': -4, 'Isoc.': -4, 'Hp.': -5, 'Thphr.': -3, 'Call.': -3,
    'A.R.': -3, 'Plb.': -2, 'D.S.': -1, 'Str.': -1, 'Ph.': 1,
    'Plu.': 1, 'Gal.': 2, 'Luc.': 2, 'Ath.': 3, 'Eust.': 12,
    'Nonn.': 5, 'Hsch.': 5, 'Suid.': 10, 'Poll.': 2,
}

century_counts = Counter()
for al in df['authors']:
    for a in al:
        if a in author_century:
            century_counts[author_century[a]] += 1

if century_counts:
    centuries = sorted(century_counts.keys())
    counts = [century_counts[c] for c in centuries]
    labels = [f'{abs(c)} {"BC" if c<0 else "AD"}' for c in centuries]
    
    fig, ax = plt.subplots(figsize=(14, 5))
    ax.bar(labels, counts, color=plt.cm.RdYlGn(np.linspace(0.2, 0.9, len(labels))))
    ax.set_ylabel('Citation Count')
    ax.set_title('LSJ Citations by Approximate Century of Author', fontsize=14, fontweight='bold')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
````
