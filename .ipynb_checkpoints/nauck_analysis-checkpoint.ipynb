# Nauck's *Tragicorum Graecorum Fragmenta* — Statistical Analysis

This notebook parses the TEI-XML edition of Augustus Nauck's *Tragicorum Graecorum Fragmenta* (2nd ed., 1889) and extracts statistics on authors, works, fragment counts, word counts, and source citations.

```python
import xml.etree.ElementTree as ET
import re
from collections import Counter, defaultdict
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['figure.dpi'] = 120
matplotlib.rcParams['figure.figsize'] = (12, 6)

tree = ET.parse('/Users/gcrane/github/GRC_misc/nauck.tragfrag-2.xml')
root = tree.getroot()

# Extract the (malformed) namespace — this edition has spaces in the namespace URI
NS = None
for elem in root.iter():
    if '}' in elem.tag:
        NS = elem.tag.split('}')[0] + '}'
        break

def t(name):
    """Return a fully-qualified tag name."""
    return f'{NS}{name}'

print(f'XML namespace: {NS}')
print('Tree parsed successfully.')
```

## 1. Parse the chapter structure

Each `<div type="textpart" subtype="chapter">` is a section. The big three tragedians (Aeschylus, Sophocles, Euripides) each span three consecutive chapters: named works, *incertarum fabularum fragmenta*, and *fragmenta dubia et spuria*. From chapter 10 onward the minor tragedians each have their own chapter. Chapter 62 is *Adespota* (fragments of unknown authorship).

```python
# Collect all chapter divs
chapters = []
for elem in root.iter():
    if (elem.tag == t('div') and
        elem.get('type') == 'textpart' and
        elem.get('subtype') == 'chapter'):
        n = elem.get('n')
        head_elem = None
        for child in elem:
            if child.tag == t('head'):
                head_elem = child
                break
        # Some chapters have no <head>; the author name is in the first <p>
        first_p = None
        if head_elem is None:
            for child in elem:
                if child.tag == t('p') and child.text and child.text.strip():
                    first_p = child.text.strip()
                    break
        head_text = head_elem.text if head_elem is not None else first_p
        chapters.append({'n': n, 'head': head_text, 'elem': elem})

print(f'Total chapter divs: {len(chapters)}')
for ch in chapters:
    print(f"  ch.{ch['n']:>3s}: {ch['head']}")
```

## 2. Map chapters to authors

We assign each chapter to an author. Chapters 2–3 belong to Aeschylus, 5–6 to Sophocles, 8–9 to Euripides. Chapter 10 is a header section for the minor tragedians (containing Choerilus). Chapter 62 is Adespota.

```python
# Map chapter number -> author name
author_map = {}

# The big three have incerta/dubia chapters after their main chapter
for n in ['1', '2', '3']:
    author_map[n] = 'AESCHYLVS'
for n in ['4', '5', '6']:
    author_map[n] = 'SOPHOCLES'
for n in ['7', '8', '9']:
    author_map[n] = 'EVRIPIDES'

# Chapter 10 has the header "TRAGICI MINORES" but contains Choerilus
author_map['10'] = 'CHOERILVS'
author_map['62'] = 'ADESPOTA'

# For all other chapters, extract author from first <head> or <p>
for ch in chapters:
    n = ch['n']
    if n in author_map:
        continue
    elem = ch['elem']
    author_name = None
    for child in elem:
        if child.tag == t('head') and child.text:
            txt = child.text.strip()
            if txt and txt != 'TRAGICI MINORES':
                author_name = txt
                break
        elif child.tag == t('p') and child.text:
            txt = child.text.strip()
            if txt and not txt[0].isdigit() and len(txt) < 50 and ':' not in txt[:20]:
                author_name = txt
                break
    if author_name is None:
        author_name = ch['head'] or f'UNKNOWN (ch.{n})'
    author_map[n] = author_name

print('Chapter -> Author mapping:')
for ch in chapters:
    print(f"  ch.{ch['n']:>3s}: {author_map[ch['n']]}")
```

## 3. Extract fragments, works, and verse text

Within each chapter, the structure follows a repeating pattern:
- **Work titles** appear as `<head>` or short all-caps `<p>` elements.
- **Fragment numbers** appear as `<p>` with a bare number.
- **Verse text** appears in `<lg>` (line-group) elements containing `<l>` (line) elements.
- **Source citations** appear in `<p>` elements *after* each `<lg>`, identifying who quoted the fragment.

The typical sequence is: `<p>frag_num</p>` → `<lg>verse</lg>` → `<p>source citation</p>`.

```python
def is_work_title(text):
    """Heuristic: work titles are short, mostly-uppercase Greek or Latin names."""
    if not text:
        return False
    text = text.strip()
    if len(text) > 60 or len(text) < 2:
        return False
    if text[0].isdigit():
        return False
    if ':' in text:
        return False
    alpha_chars = [c for c in text if c.isalpha()]
    if not alpha_chars:
        return False
    upper_ratio = sum(1 for c in alpha_chars if c.isupper()) / len(alpha_chars)
    return upper_ratio > 0.8


def is_fragment_number(text):
    """Check if text is a bare fragment number like '1', '2a', '132'."""
    if not text:
        return False
    return bool(re.match(r'^\d+[a-z]?$', text.strip()))


# Common source abbreviations in Nauck's apparatus
KNOWN_SOURCES = {
    'Stob': 'Stobaeus', 'Ath': 'Athenaeus', 'Athen': 'Athenaeus',
    'Hesych': 'Hesychius', 'Schol': 'Scholia', 'Plut': 'Plutarch',
    'Sext': 'Sextus Empiricus', 'Sextus': 'Sextus Empiricus',
    'Clem': 'Clement of Alexandria', 'Clemens': 'Clement of Alexandria',
    'Galen': 'Galen', 'Eust': 'Eustathius',
    'Etym': 'Etymologicum Magnum', 'Harpocr': 'Harpocration',
    'Steph': 'Stephanus Byzantius', 'Phot': 'Photius',
    'Suid': 'Suda', 'Diog': 'Diogenes Laertius',
    'Aristot': 'Aristotle', 'Arist': 'Aristotle',
    'Cic': 'Cicero', 'Plat': 'Plato', 'Poll': 'Pollux',
    'Paus': 'Pausanias', 'Pausan': 'Pausanias',
    'Orion': 'Orion', 'Macrob': 'Macrobius', 'Tzetzes': 'Tzetzes',
    'Strabo': 'Strabo', 'Anecd': 'Anecdota',
    'Dio': 'Dio Chrysostom', 'Lucian': 'Lucian',
    'Porphyr': 'Porphyry', 'Theophil': 'Theophilus',
    'Sueton': 'Suetonius', 'Dicaearch': 'Dicaearchus',
    'Demetr': 'Demetrius', 'Diod': 'Diodorus Siculus',
    'Plin': 'Pliny', 'Senec': 'Seneca',
    'Philodem': 'Philodemus', 'Priscian': 'Priscian',
    'Apollod': 'Apollodorus', 'Chrysipp': 'Chrysippus',
}


def extract_source_name(text):
    """Extract the citing source from a citation paragraph."""
    if not text:
        return None
    text = text.strip()
    if len(text) < 5:
        return None

    m = re.match(r'^([A-Za-z]+)', text)
    if not m:
        return None
    first_word = m.group(1)

    skip = {'cf', 'de', 'ex', 'et', 'in', 'ad', 'ab', 'cum', 'quae', 'ubi',
            'item', 'est', 'sed', 'vel', 'sic', 'vide', 'haec', 'hoc',
            'non', 'qui', 'nisi', 'idem', 'hunc', 'huius', 'quod',
            'apud', 'ante', 'post', 'supra', 'infra', 'inter', 'fuit',
            'Huic', 'Obiter', 'Vita', 'Argum', 'Incertae', 'Huius',
            'quot', 'versus', 'verba', 'ita', 'at', 'ut', 'quam',
            'pro', 'per', 'sub', 'tam', 'aut', 'nec', 'hic', 'his',
            'eadem', 'idem', 'tamen', 'autem', 'quidem',
            'praeterea', 'fortasse', 'videtur'}
    if first_word in skip or first_word.lower() in skip:
        return None
    if len(first_word) < 2:
        return None

    for abbrev, full_name in KNOWN_SOURCES.items():
        if first_word.startswith(abbrev) or first_word == abbrev:
            return full_name

    if first_word[0].isupper() and len(first_word) >= 3:
        return first_word

    return None


def get_verse_words(lg_elem):
    """Extract all words from <l> elements within an <lg>."""
    words = []
    for l_elem in lg_elem.iter(t('l')):
        text = ''.join(l_elem.itertext()).strip()
        if text:
            tokens = re.findall(r'[\w]+', text, re.UNICODE)
            words.extend(tokens)
    return words


# Main parsing loop
all_fragments = []

for ch in chapters:
    author = author_map[ch['n']]
    elem = ch['elem']
    current_work = None
    current_frag_num = None
    is_first_head = True

    children = list(elem)
    i = 0
    while i < len(children):
        child = children[i]
        child_tag = child.tag.split('}')[-1] if '}' in child.tag else child.tag
        child_text = (child.text or '').strip()

        if child_tag == 'head':
            if is_first_head:
                is_first_head = False
                full_text = ''.join(child.itertext()).strip()
                lines = full_text.split('\n')
                if len(lines) > 1 and is_work_title(lines[-1].strip()):
                    current_work = lines[-1].strip()
            else:
                full_text = ''.join(child.itertext()).strip()
                if is_work_title(full_text):
                    current_work = full_text
            i += 1
            continue

        if child_tag == 'p':
            if is_fragment_number(child_text):
                current_frag_num = child_text.strip()
            elif is_work_title(child_text):
                current_work = child_text
            i += 1
            continue

        if child_tag == 'lg':
            words = get_verse_words(child)
            verse_lines = list(child.iter(t('l')))

            # Collect source citations from surrounding <p> elements
            frag_sources = []

            # Check <p> elements BEFORE this <lg> (between frag_num and here)
            for j in range(max(0, i - 4), i):
                prev = children[j]
                prev_tag = prev.tag.split('}')[-1] if '}' in prev.tag else prev.tag
                if prev_tag == 'p':
                    prev_text = (prev.text or '').strip()
                    if (not is_fragment_number(prev_text) and
                        not is_work_title(prev_text)):
                        src = extract_source_name(prev_text)
                        if src:
                            frag_sources.append(src)

            # Check the <p> element AFTER this <lg>
            for j in range(i + 1, min(len(children), i + 3)):
                nxt = children[j]
                nxt_tag = nxt.tag.split('}')[-1] if '}' in nxt.tag else nxt.tag
                if nxt_tag == 'p':
                    nxt_text = (nxt.text or '').strip()
                    if (not is_fragment_number(nxt_text) and
                        not is_work_title(nxt_text)):
                        src = extract_source_name(nxt_text)
                        if src:
                            frag_sources.append(src)
                    break
                elif nxt_tag == 'lg':
                    break

            all_fragments.append({
                'author': author,
                'chapter': ch['n'],
                'work': current_work,
                'frag_num': current_frag_num or '?',
                'words': words,
                'word_count': len(words),
                'line_count': len(verse_lines),
                'sources': list(set(frag_sources)),
            })
            i += 1
            continue

        i += 1

print(f'Total fragments extracted: {len(all_fragments):,}')
print(f'Total verse lines (<l> elements): {sum(f["line_count"] for f in all_fragments):,}')
print(f'Total words of verse text: {sum(f["word_count"] for f in all_fragments):,}')
```

## 4. How many authors?

```python
authors = sorted(set(f['author'] for f in all_fragments))
named_authors = [a for a in authors if a != 'ADESPOTA']
print(f'Total distinct author sections (including Adespota): {len(authors)}')
print(f'Named authors: {len(named_authors)}')
print()
for a in authors:
    print(f'  {a}')
```

## 5. How many different works?

```python
works_by_author = defaultdict(set)
for f in all_fragments:
    if f['work']:
        works_by_author[f['author']].add(f['work'])

# Filter out meta-titles like "INCERTARVM FABVLARVM FRAGMENTA"
meta_titles = {'INCERTARVM FABVLARVM FRAGMENTA',
               'INCERTARVM FABNLARVM FRAGMENTA',
               'INCERTAE FABVLAE FRAGMENTVM',
               'INCERTAE FABULAE FRAGMENTUM'}

real_works = defaultdict(set)
for author, works in works_by_author.items():
    for w in works:
        if w not in meta_titles:
            real_works[author].add(w)

total_works = sum(len(ws) for ws in real_works.values())
print(f'Total distinct play/work titles (excl. section headings): {total_works}')
print()

for author, works in sorted(real_works.items(), key=lambda x: -len(x[1])):
    print(f'{author}: {len(works)} works')
    for w in sorted(works):
        print(f'    {w}')
    print()
```

## 6. Words of surviving text per author

```python
words_by_author = defaultdict(int)
frags_by_author = defaultdict(int)
lines_by_author = defaultdict(int)

for f in all_fragments:
    words_by_author[f['author']] += f['word_count']
    frags_by_author[f['author']] += 1
    lines_by_author[f['author']] += f['line_count']

print(f'{"Author":<30s} {"Fragments":>10s} {"Lines":>8s} {"Words":>8s}')
print('-' * 60)
for author in sorted(words_by_author, key=lambda a: -words_by_author[a]):
    print(f'{author:<30s} {frags_by_author[author]:>10d} '
          f'{lines_by_author[author]:>8d} {words_by_author[author]:>8d}')
```

```python
# Bar chart: words per author (top 15)
sorted_authors = sorted(words_by_author, key=lambda a: -words_by_author[a])
top_n = 15
top_authors = sorted_authors[:top_n]

fig, ax = plt.subplots(figsize=(14, 6))
bars = ax.barh(range(len(top_authors)),
               [words_by_author[a] for a in top_authors],
               color='#2c5f8a', edgecolor='white')
ax.set_yticks(range(len(top_authors)))
ax.set_yticklabels(top_authors, fontsize=10)
ax.invert_yaxis()
ax.set_xlabel('Words of surviving verse text')
ax.set_title('Words of Surviving Verse Text per Author (Top 15)')
for bar, author in zip(bars, top_authors):
    ax.text(bar.get_width() + 50, bar.get_y() + bar.get_height() / 2,
            f'{words_by_author[author]:,}', va='center', fontsize=9)
plt.tight_layout()
plt.show()
```

## 7. Words of surviving text per work (top 30)

```python
words_by_work = defaultdict(int)
frags_by_work = defaultdict(int)
lines_by_work = defaultdict(int)

for f in all_fragments:
    key = (f['author'], f['work'] or '[no title]')
    words_by_work[key] += f['word_count']
    frags_by_work[key] += 1
    lines_by_work[key] += f['line_count']

sorted_works = sorted(words_by_work, key=lambda k: -words_by_work[k])

print(f'{"Author":<25s} {"Work":<35s} {"Frags":>6s} {"Lines":>6s} {"Words":>6s}')
print('-' * 82)
for (author, work) in sorted_works[:30]:
    print(f'{author:<25s} {work[:33]:<35s} {frags_by_work[(author, work)]:>6d} '
          f'{lines_by_work[(author, work)]:>6d} {words_by_work[(author, work)]:>6d}')
```

```python
# Bar chart: top 20 works by word count
top_works = sorted_works[:20]
labels = [f'{a}: {w[:25]}' for a, w in top_works]
values = [words_by_work[k] for k in top_works]

fig, ax = plt.subplots(figsize=(14, 7))
colors = ['#c44e52' if a == 'EVRIPIDES' else
          '#4c72b0' if a == 'AESCHYLVS' else
          '#55a868' if a == 'SOPHOCLES' else
          '#ccb974' if a == 'ADESPOTA' else
          '#8172b2' for a, w in top_works]
ax.barh(range(len(labels)), values, color=colors, edgecolor='white')
ax.set_yticks(range(len(labels)))
ax.set_yticklabels(labels, fontsize=9)
ax.invert_yaxis()
ax.set_xlabel('Words of surviving verse text')
ax.set_title('Top 20 Works by Surviving Verse Text')

from matplotlib.patches import Patch
legend_elements = [
    Patch(facecolor='#4c72b0', label='Aeschylus'),
    Patch(facecolor='#55a868', label='Sophocles'),
    Patch(facecolor='#c44e52', label='Euripides'),
    Patch(facecolor='#ccb974', label='Adespota'),
    Patch(facecolor='#8172b2', label='Others'),
]
ax.legend(handles=legend_elements, loc='lower right')
plt.tight_layout()
plt.show()
```

## 8. Sources of quotation

These are the ancient authors and texts from which the tragic fragments are quoted — Stobaeus, Athenaeus, Hesychius, various scholia, etc.

```python
all_sources = []
for f in all_fragments:
    all_sources.extend(f['sources'])

source_counts = Counter(all_sources)

print(f'Total source citations extracted: {len(all_sources):,}')
print(f'Distinct citing sources identified: {len(source_counts)}')
print()
print(f'{"Source":<35s} {"Citations":>10s}')
print('-' * 47)
for src, count in source_counts.most_common(40):
    print(f'{src:<35s} {count:>10d}')
```

```python
# Bar chart of top 20 sources
top_sources = source_counts.most_common(20)
src_names = [s for s, c in top_sources]
src_vals = [c for s, c in top_sources]

fig, ax = plt.subplots(figsize=(12, 6))
ax.barh(range(len(src_names)), src_vals, color='#dd8452', edgecolor='white')
ax.set_yticks(range(len(src_names)))
ax.set_yticklabels(src_names, fontsize=10)
ax.invert_yaxis()
ax.set_xlabel('Number of fragment citations')
ax.set_title('Top 20 Sources Citing Tragic Fragments')
for i, v in enumerate(src_vals):
    ax.text(v + 2, i, str(v), va='center', fontsize=9)
plt.tight_layout()
plt.show()
```

## 9. Fragment size distribution

```python
frag_sizes = [f['word_count'] for f in all_fragments if f['word_count'] > 0]

print('=== Fragment size distribution (words per fragment) ===')
print(f'  Total fragments with verse text: {len(frag_sizes):,}')
print(f'  Mean words per fragment:         {sum(frag_sizes) / len(frag_sizes):.1f}')
sorted_sizes = sorted(frag_sizes)
print(f'  Median words per fragment:       {sorted_sizes[len(sorted_sizes) // 2]}')
print(f'  Max words in a single fragment:  {max(frag_sizes)}')
print(f'  Single-word fragments:           {sum(1 for s in frag_sizes if s == 1)}')
print(f'  2\u20135 words:                       {sum(1 for s in frag_sizes if 2 <= s <= 5)}')
print(f'  6\u201320 words:                      {sum(1 for s in frag_sizes if 6 <= s <= 20)}')
print(f'  21\u201350 words:                     {sum(1 for s in frag_sizes if 21 <= s <= 50)}')
print(f'  51\u2013100 words:                    {sum(1 for s in frag_sizes if 51 <= s <= 100)}')
print(f'  >100 words:                      {sum(1 for s in frag_sizes if s > 100)}')
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(frag_sizes, bins=50, color='#4c72b0', edgecolor='white', range=(0, 150))
axes[0].set_xlabel('Words per fragment')
axes[0].set_ylabel('Number of fragments')
axes[0].set_title('Fragment Size Distribution (0\u2013150 words)')
axes[0].axvline(sum(frag_sizes) / len(frag_sizes), color='red',
                linestyle='--',
                label=f'Mean ({sum(frag_sizes) / len(frag_sizes):.1f})')
axes[0].legend()

axes[1].hist(frag_sizes, bins=80, color='#4c72b0', edgecolor='white')
axes[1].set_xlabel('Words per fragment')
axes[1].set_ylabel('Number of fragments (log scale)')
axes[1].set_title('Fragment Size Distribution (full range)')
axes[1].set_yscale('log')

plt.tight_layout()
plt.show()
```

## 10. Big Three vs. Minor Tragedians vs. Adespota

```python
big_three = {'AESCHYLVS', 'SOPHOCLES', 'EVRIPIDES'}

categories = {'Big Three (Aesch/Soph/Eur)': [],
              'Minor Tragedians': [],
              'Adespota': []}
for f in all_fragments:
    if f['author'] in big_three:
        categories['Big Three (Aesch/Soph/Eur)'].append(f)
    elif f['author'] == 'ADESPOTA':
        categories['Adespota'].append(f)
    else:
        categories['Minor Tragedians'].append(f)

print(f'{"Category":<30s} {"Authors":>8s} {"Works":>8s} '
      f'{"Frags":>8s} {"Lines":>8s} {"Words":>8s}')
print('-' * 76)
for cat, frags in categories.items():
    n_authors = len(set(f['author'] for f in frags))
    n_works = len(set((f['author'], f['work']) for f in frags if f['work']))
    n_frags = len(frags)
    n_lines = sum(f['line_count'] for f in frags)
    n_words = sum(f['word_count'] for f in frags)
    print(f'{cat:<30s} {n_authors:>8d} {n_works:>8d} '
          f'{n_frags:>8d} {n_lines:>8d} {n_words:>8d}')

# Pie chart
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
cat_names = list(categories.keys())
cat_frags = [len(categories[c]) for c in cat_names]
cat_words = [sum(f['word_count'] for f in categories[c]) for c in cat_names]
colors = ['#4c72b0', '#8172b2', '#ccb974']

axes[0].pie(cat_frags, labels=cat_names, autopct='%1.1f%%', colors=colors,
            startangle=90)
axes[0].set_title('Share of Fragments')

axes[1].pie(cat_words, labels=cat_names, autopct='%1.1f%%', colors=colors,
            startangle=90)
axes[1].set_title('Share of Surviving Words')

plt.tight_layout()
plt.show()
```

## 11. Average fragment length by author

```python
print(f'{"Author":<30s} {"Frags":>6s} {"Avg Words":>10s} {"Avg Lines":>10s}')
print('-' * 60)
author_stats = []
for author in sorted(set(f['author'] for f in all_fragments)):
    afrags = [f for f in all_fragments
              if f['author'] == author and f['word_count'] > 0]
    if len(afrags) >= 5:
        avg_words = sum(f['word_count'] for f in afrags) / len(afrags)
        avg_lines = sum(f['line_count'] for f in afrags) / len(afrags)
        author_stats.append((author, len(afrags), avg_words, avg_lines))

for author, n, avg_w, avg_l in sorted(author_stats, key=lambda x: -x[2]):
    print(f'{author:<30s} {n:>6d} {avg_w:>10.1f} {avg_l:>10.1f}')
```

## 12. Diversity of sources per author

```python
sources_per_author = defaultdict(Counter)
for f in all_fragments:
    for src in f['sources']:
        sources_per_author[f['author']][src] += 1

print(f'{"Author":<30s} {"Distinct":>8s}  Top 3 Sources')
print('-' * 90)
for author in sorted(sources_per_author,
                     key=lambda a: -len(sources_per_author[a]))[:20]:
    sc = sources_per_author[author]
    top3 = ', '.join(f'{s} ({c})' for s, c in sc.most_common(3))
    print(f'{author:<30s} {len(sc):>8d}  {top3}')
```

```python
sorted_by_words = sorted(all_fragments, key=lambda f: -f['word_count'])
print(f'{"Author":<25s} {"Work":<30s} {"Fr.":>5s} {"Lines":>6s} {"Words":>6s}')
print('-' * 76)
for f in sorted_by_words[:20]:
    work = (f['work'] or '[untitled]')[:28]
    print(f'{f["author"]:<25s} {work:<30s} {f["frag_num"]:>5s} '
          f'{f["line_count"]:>6d} {f["word_count"]:>6d}')
```

## 14. Summary

```python
named_auth = [a for a in set(f['author'] for f in all_fragments)
              if a != 'ADESPOTA']
all_works_set = set(
    (f['author'], f['work']) for f in all_fragments
    if f['work'] and f['work'] not in meta_titles
)
total_words = sum(f['word_count'] for f in all_fragments)
total_lines = sum(f['line_count'] for f in all_fragments)

print('=' * 60)
print('SUMMARY STATISTICS')
print('=' * 60)
print(f'Named authors:                {len(named_auth)}')
print(f'Distinct play titles:         {len(all_works_set)}')
print(f'Total fragments (with text):  {len(all_fragments):,}')
print(f'Total verse lines:            {total_lines:,}')
print(f'Total words of verse text:    {total_words:,}')
print(f'Distinct citing sources:      {len(source_counts)}')
print(f'Total source citations:       {sum(source_counts.values()):,}')
print(f'Mean words per fragment:      '
      f'{total_words / len(all_fragments):.1f}')
frag_wc = sorted(f['word_count'] for f in all_fragments)
print(f'Median words per fragment:    '
      f'{frag_wc[len(frag_wc) // 2]}')
print()
eur_pct = words_by_author["EVRIPIDES"] / total_words * 100
soph_pct = words_by_author["SOPHOCLES"] / total_words * 100
aesch_pct = words_by_author["AESCHYLVS"] / total_words * 100
print(f'Euripides accounts for {eur_pct:.0f}% of all surviving words,')
print(f'Sophocles {soph_pct:.0f}%, and Aeschylus {aesch_pct:.0f}%.')
print()
print('Most fragments are very short \u2014 the median is just a few words.')
print('Single-word fragments (glosses) and short phrases dominate,')
print('while substantial multi-line passages are rare.')
```
