# Perseus CTS Pipeline — Scaife Viewer Integration

**Refactored from `perseus-urn-lib-v8.ipynb`** to implement the scalable, sustainable strategy for managing content in the [Scaife Viewer](https://scaife.perseus.org/) ecosystem.

## Architecture: Three Output Backends

| Backend | Output | Deployment | Use Case |
|---------|--------|------------|----------|
| **ATLAS JSONL** | `.jsonl` files per edition | `new-atlas` Django server | Full Scaife Viewer with ATLAS v2 |
| **CTS Catalog** | `__cts__.xml` metadata | Nautilus/MyCapytain resolver | CTS API for Scaife 1.0 |
| **Static Site** | `index.html` + JS chunks | **GitHub Pages (serverless)** | Zero-server deployment |

## Pipeline Flow

```
TEI XML  →  Parse Layer  →  Canonical Model  →  Output Backends
              (4 modes)       (Python dicts       ├─ ATLAS JSONL
               │                keyed by URN)     ├─ CTS __cts__.xml
               ├─ hierarchical                    └─ Static Site (GitHub Pages)
               ├─ milestones
               ├─ poetry_cards
               └─ auto-detect
```

Run cells sequentially. Cell 1 sets up configuration, Cell 2 has the parse/model layer, Cell 3 has the output backends, Cell 4 runs the pipeline.

Set your local paths, choose which works to process, and select output backends.

````python
import os
import json
import re
from pathlib import Path
from collections import OrderedDict
from dataclasses import dataclass, field
from typing import Dict, List, Optional, Tuple
import xml.etree.ElementTree as ET
from abc import ABC, abstractmethod

# ═══════════════════════════════════════════════════════════════
# PATHS — Update these to match your local checkout
# ═══════════════════════════════════════════════════════════════
BASE_DIR = Path("/Users/gcrane/github")
CANONICAL_GREEKLIST = BASE_DIR / "canonical-greekLit" / "data"
POETICS_DIR = BASE_DIR / "Poetics2.0"

# Output directory (will contain atlas/, cts/, and site/ subdirs)
OUTPUT_DIR = Path("/Users/gcrane/github/persverscomp")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# Which backends to run: any combination of "atlas", "cts", "site"
ACTIVE_BACKENDS = ["atlas", "cts", "site"]

# TEI namespace
TEI_NS = "{http://www.tei-c.org/ns/1.0}"

print(f"Base directory:  {BASE_DIR}")
print(f"Output target:   {OUTPUT_DIR}")
print(f"Active backends: {ACTIVE_BACKENDS}")
````

## Cell 2 — Data Model & TEI Parsing Layer

Core data classes and all four TEI parsing modes (hierarchical, milestones, poetry cards, auto-detect).

````python
# ═══════════════════════════════════════════════════════════════
# DATA CLASSES
# ═══════════════════════════════════════════════════════════════

@dataclass
class EditionConfig:
    """Configuration for a single edition/translation of a work."""
    path: str
    label: str
    css_class: str
    lang: str = ""
    parse_mode: str = "auto"
    description: str = ""

    def __post_init__(self):
        if not self.lang:
            class_map = {
                "greek-text": "grc", "english-text": "eng",
                "french-text": "fre", "german-text": "ger",
                "italian-text": "ita", "latin-text": "lat",
            }
            self.lang = class_map.get(self.css_class, "und")


@dataclass
class WorkConfig:
    """Configuration for a single work (all its editions)."""
    textgroup: str
    work: str
    label: str = ""
    editions: Dict[str, EditionConfig] = field(default_factory=dict)

    @property
    def urn_work(self) -> str:
        return f"urn:cts:greekLit:{self.textgroup}.{self.work}"

    @property
    def work_key(self) -> str:
        return f"{self.textgroup}.{self.work}"


@dataclass
class ParsedEdition:
    """Parsed content for one edition of a work."""
    edition_id: str
    config: EditionConfig
    urn: str
    data: OrderedDict  # data[book][chapter][section] = html

    @property
    def segment_count(self) -> int:
        return sum(len(s) for c in self.data.values() for s in c.values())


@dataclass
class ParsedWork:
    """All parsed editions for a single work, with structural metadata."""
    config: WorkConfig
    editions: Dict[str, ParsedEdition]
    structure: OrderedDict
    is_flat: bool
    is_poetry: bool
    chapter_sequence: list


# ═══════════════════════════════════════════════════════════════
# TEI PARSING FUNCTIONS
# ═══════════════════════════════════════════════════════════════

def find_text_root(root):
    """Locate the primary text body in a TEI document."""
    for div in root.findall(f".//{TEI_NS}div") + root.findall(".//div"):
        if div.get("type") == "translation":
            return div
    return (
        root.find(f".//{TEI_NS}body")
        or root.find(".//body")
        or root.find(".//*body")
    )


def extract_text_recursive(elem, strip_paragraphs=False):
    """Recursively extract text from TEI → HTML."""
    parts = []
    tag = elem.tag.split("}")[-1]

    if tag == "l":
        line_num = (elem.get("n") or "").strip()
        parts.append(f'<div class="verse-line" data-line="{line_num}">')
    elif tag == "speaker":
        parts.append('<strong class="speaker-attr">')
    elif tag == "stage":
        parts.append('<div class="stage-direction">')
    elif tag == "hi":
        rend = elem.get("rend", "italic")
        parts.append(f'<span class="render-{rend}">')
    elif tag == "quote":
        q_type = elem.get("type", "blockquote")
        parts.append(f'<blockquote class="quote-block type-{q_type}">')
    elif tag == "p" and not strip_paragraphs:
        parts.append("<p>")

    if elem.text:
        parts.append(elem.text)

    for child in elem:
        child_tag = child.tag.split("}")[-1]
        if child_tag == "note":
            note_text = extract_text_recursive(child, strip_paragraphs).strip()
            if note_text:
                parts.append(f'<span class="note">[{note_text}]</span>')
        elif child_tag == "lb":
            parts.append("<br/>")
        else:
            parts.append(extract_text_recursive(child, strip_paragraphs))
        if child.tail:
            parts.append(child.tail)

    if tag == "l": parts.append("</div>")
    elif tag == "speaker": parts.append(": </strong>")
    elif tag == "stage": parts.append("</div>")
    elif tag == "hi": parts.append("</span>")
    elif tag == "quote": parts.append("</blockquote>")
    elif tag == "p" and not strip_paragraphs: parts.append("</p>")
    return "".join(parts)


def parse_hierarchical_tei(path):
    """Parse TEI with nested div[@type] hierarchy."""
    if not os.path.exists(path): return None
    tree = ET.parse(path)
    text_entry = find_text_root(tree.getroot())
    if text_entry is None: return None
    data = OrderedDict()

    def walk_divisions(node, current_path):
        tag = node.tag.split("}")[-1]
        subtype = node.get("subtype") or node.get("type")
        n_val = node.get("n")
        if tag == "div" and subtype in ("book","chapter","section","part","textpart") and n_val:
            resolved = "chapter" if subtype in ("part","textpart") and n_val.isdigit() else subtype
            new_path = current_path + [(resolved, str(n_val).strip())]
        else:
            new_path = current_path

        paragraphs = node.findall(f"{TEI_NS}p") or node.findall("p")
        if paragraphs and new_path:
            key_map = {t: v for t, v in new_path}
            bk = key_map.get("book", "1")
            ch = key_map.get("chapter", key_map.get("part", "1"))
            sec = key_map.get("section", n_val or "1")
            if bk not in data: data[bk] = OrderedDict()
            if ch not in data[bk]: data[bk][ch] = OrderedDict()
            combined = " ".join(extract_text_recursive(p).strip() for p in paragraphs)
            if combined: data[bk][ch][sec] = combined
            return
        for child in node:
            walk_divisions(child, new_path)

    walk_divisions(text_entry, [])
    return data


def parse_milestone_aligned_tei(path):
    """Parse TEI using milestone markers for structure."""
    if not os.path.exists(path): return None
    tree = ET.parse(path)
    text_entry = find_text_root(tree.getroot())
    if text_entry is None: return None
    data = OrderedDict()
    current_ch, current_sec, current_text = "1", "1", []

    def _flush():
        nonlocal current_ch, current_sec, current_text
        if current_text:
            bk = "1"
            data.setdefault(bk, OrderedDict()).setdefault(current_ch, OrderedDict())
            html_blob = " ".join(t.strip() for t in current_text if t.strip())
            if html_blob:
                if current_sec in data[bk][current_ch]:
                    data[bk][current_ch][current_sec] += " " + html_blob
                else:
                    data[bk][current_ch][current_sec] = html_blob
        current_text = []

    def _walk(elem):
        nonlocal current_ch, current_sec, current_text
        tag = elem.tag.split("}")[-1]
        if tag == "milestone":
            unit = elem.get("unit") or ""
            val = (elem.get("n") or "").strip()
            if val:
                if unit == "bekker" and "." in val:
                    parts = val.split(".", 1)
                    _flush(); current_ch = parts[0]; current_sec = parts[1]
                else:
                    num_match = re.match(r"^\d+", val)
                    if num_match:
                        if unit in ("chapter", "chapter-start", ""):
                            _flush(); current_ch = str(int(num_match.group(0))); current_sec = "1"
                        elif unit in ("section", "section-start"):
                            _flush(); current_sec = str(int(num_match.group(0)))
            if elem.tail and elem.tail.strip(): current_text.append(elem.tail)
            return
        if tag == "note":
            note_html = extract_text_recursive(elem, strip_paragraphs=True).strip()
            if note_html: current_text.append(f'<span class="note">[{note_html}]</span>')
            if elem.tail and elem.tail.strip(): current_text.append(elem.tail)
            return
        if tag in ("hi", "foreign", "emph"):
            rend = elem.get("rend", "italic")
            current_text.append(f'<span class="render-{rend}">')
            if elem.text: current_text.append(elem.text)
            for child in elem: _walk(child)
            current_text.append("</span>")
            if elem.tail and elem.tail.strip(): current_text.append(elem.tail)
            return
        if tag in ("p", "div"):
            if elem.text and elem.text.strip(): current_text.append(elem.text)
            for child in elem: _walk(child)
            if elem.tail and elem.tail.strip(): current_text.append(elem.tail)
            return
        if elem.text and elem.text.strip(): current_text.append(elem.text)
        for child in elem: _walk(child)
        if elem.tail and elem.tail.strip(): current_text.append(elem.tail)

    _walk(text_entry)
    _flush()
    return data


def build_poetry_canonical_intervals(editions):
    """Build card intervals from the Greek edition's milestone markers."""
    grc_cfg = editions.get("perseus-grc2") or list(editions.values())[0]
    tree = ET.parse(grc_cfg.path)
    text_entry = find_text_root(tree.getroot())
    landmarks = []
    for elem in text_entry.iter():
        if elem.tag.endswith("milestone") and elem.get("unit") == "card":
            landmarks.append(("card", elem.get("n").strip()))
        elif elem.tag.endswith("l"):
            ln = (elem.get("n") or "").strip()
            if ln: landmarks.append(("line", ln))
    intervals = []
    for idx, item in enumerate(landmarks):
        if item[0] == "card":
            card_n = item[1]
            first_l, last_l = None, None
            for ahead in landmarks[idx+1:]:
                if ahead[0] == "card": break
                if ahead[0] == "line":
                    if first_l is None: first_l = ahead[1]
                    last_l = ahead[1]
            if not first_l: first_l = card_n
            if not last_l: last_l = first_l
            intervals.append({"card_n": card_n, "label": f"{first_l}-{last_l}"})
    return intervals


def parse_poetry_cards_tei(path, master_intervals):
    """Parse TEI structured by drama/poetry card milestones."""
    if not os.path.exists(path): return None
    tree = ET.parse(path)
    text_entry = find_text_root(tree.getroot())
    data = OrderedDict()
    data["1"] = OrderedDict()
    current_label = master_intervals[0]["label"] if master_intervals else "1"
    current_text = []

    def _flush():
        nonlocal current_label, current_text
        if current_text and current_label:
            html = " ".join(t.strip() for t in current_text if t.strip())
            if html:
                if current_label not in data["1"]: data["1"][current_label] = OrderedDict()
                data["1"][current_label]["1"] = html
        current_text = []

    def _walk(elem):
        nonlocal current_label, current_text
        tag = elem.tag.split("}")[-1]
        if tag == "milestone" and elem.get("unit") == "card":
            val = (elem.get("n") or "").strip()
            match = [r for r in master_intervals if r["card_n"] == val]
            if match: _flush(); current_label = match[0]["label"]
        elif tag in ("l", "stage"):
            current_text.append(extract_text_recursive(elem, strip_paragraphs=True))
            return
        elif tag == "note":
            note_html = extract_text_recursive(elem, strip_paragraphs=True).strip()
            if note_html: current_text.append(f'<span class="note">[{note_html}]</span>')
            if elem.tail and elem.tail.strip(): current_text.append(elem.tail)
            return
        if elem.text and elem.text.strip(): current_text.append(elem.text)
        for child in elem: _walk(child)
        if elem.tail and elem.tail.strip(): current_text.append(elem.tail)

    _walk(text_entry)
    _flush()
    return data


def parse_edition(cfg, master_intervals=None):
    """Dispatch to the correct parser based on parse_mode."""
    if cfg.parse_mode == "poetry_cards":
        return parse_poetry_cards_tei(cfg.path, master_intervals or [])
    elif cfg.parse_mode == "milestones":
        return parse_milestone_aligned_tei(cfg.path)
    else:
        return parse_hierarchical_tei(cfg.path)


def generate_canonical_id(work_key, label_string):
    """Generate a stable canonical ID from work key + edition label."""
    prefix = work_key.replace(".", "_")
    match = re.search(r"\((.*?)\)", label_string)
    if match:
        content = match.group(1)
        parts = content.split(",")
        editor = parts[0].strip()
        year = parts[1].strip() if len(parts) > 1 else ""
        editor_last = editor.split()[-1].lower().replace(".", "")
        year_clean = re.sub(r"\D", "", year)
        return f"{prefix}_{editor_last}_{year_clean}"
    fallback = re.sub(r"\W+", "_", label_string).lower()
    return f"{prefix}_{fallback}"

print("✓ Data model and parsing layer loaded.")
````

## Cell 3 — Output Backends

Three output backends that share the same canonical data model:

1. **AtlasJsonlBackend** — JSONL files for [Scaife ATLAS v2](https://github.com/scaife-viewer/new-atlas) ingestion
2. **CtsMetadataBackend** — `__cts__.xml` catalog files for the CTS resolver
3. **StaticSiteBackend** — Complete serverless HTML+JS site for **GitHub Pages** deployment

````python
# ═══════════════════════════════════════════════════════════════
# BACKEND BASE CLASS
# ═══════════════════════════════════════════════════════════════

class OutputBackend(ABC):
    @abstractmethod
    def emit_work(self, work: ParsedWork) -> None: pass
    @abstractmethod
    def finalize(self, works: List[ParsedWork]) -> None: pass


# ═══════════════════════════════════════════════════════════════
# BACKEND 1: ATLAS JSONL (for Scaife new-atlas)
# ═══════════════════════════════════════════════════════════════

class AtlasJsonlBackend(OutputBackend):
    """
    Produce JSONL for Scaife ATLAS v2 ingestion.
    
    After running, copy files to new-atlas/test-data/texts/
    and run: ./startup.sh -t
    """
    def __init__(self, output_dir):
        self.output_dir = output_dir / "atlas" / "texts"
        self.output_dir.mkdir(parents=True, exist_ok=True)
        self.manifest = []

    def emit_work(self, work):
        for ed_id, edition in work.editions.items():
            lines = []
            for bk, chapters in edition.data.items():
                for ch, sections in chapters.items():
                    for sec, html_content in sections.items():
                        if work.is_flat:
                            ref = f"{ch}.{sec}" if sec != "1" or len(sections) > 1 else ch
                        else:
                            ref = f"{bk}.{ch}.{sec}" if sec != "1" or len(sections) > 1 else f"{bk}.{ch}"
                        passage_urn = f"{edition.urn}:{ref}"
                        plain = re.sub(r"<[^>]+>", "", html_content)
                        plain = re.sub(r"\s+", " ", plain).strip()
                        record = {
                            "urn": passage_urn,
                            "text_content": plain,
                            "html_content": html_content,
                            "ref": ref,
                            "label": edition.config.label,
                            "lang": edition.config.lang,
                        }
                        lines.append(json.dumps(record, ensure_ascii=False))
            filename = f"{work.config.work_key}.{ed_id}.jsonl"
            (self.output_dir / filename).write_text("\n".join(lines) + "\n", encoding="utf-8")
            self.manifest.append({"urn": edition.urn, "label": edition.config.label, "lang": edition.config.lang, "file": filename, "segments": len(lines)})
            print(f"    → ATLAS: {filename} ({len(lines)} nodes)")

    def finalize(self, works):
        (self.output_dir / "manifest.json").write_text(json.dumps(self.manifest, indent=2, ensure_ascii=False), encoding="utf-8")
        print(f"\n  [ATLAS] Manifest written to {self.output_dir / 'manifest.json'}")
        print(f"  [ATLAS] To ingest: cp {self.output_dir}/*.jsonl <new-atlas>/test-data/texts/ && ./startup.sh -t")


# ═══════════════════════════════════════════════════════════════
# BACKEND 2: CTS __cts__.xml Catalog
# ═══════════════════════════════════════════════════════════════

class CtsMetadataBackend(OutputBackend):
    """Generate __cts__.xml for CTS/CapiTainS resolver (Nautilus/MyCapytain)."""
    CTS_NS = "http://chs.harvard.edu/xmlns/cts"

    def __init__(self, output_dir):
        self.output_dir = output_dir / "cts"
        self.output_dir.mkdir(parents=True, exist_ok=True)
        self.seen_tg = set()

    def emit_work(self, work):
        tg, wk = work.config.textgroup, work.config.work
        if tg not in self.seen_tg:
            tg_dir = self.output_dir / tg
            tg_dir.mkdir(parents=True, exist_ok=True)
            xml = f'<?xml version="1.0" encoding="UTF-8"?>\n<TextGroup xmlns="{self.CTS_NS}" urn="urn:cts:greekLit:{tg}">\n  <groupname xml:lang="eng">{tg}</groupname>\n</TextGroup>'
            (tg_dir / "__cts__.xml").write_text(xml, encoding="utf-8")
            self.seen_tg.add(tg)

        wk_dir = self.output_dir / tg / wk
        wk_dir.mkdir(parents=True, exist_ok=True)
        entries = []
        for ed_id, ed in work.editions.items():
            tag = "edition" if ed.config.lang == "grc" else "translation"
            entries.append(f'  <{tag} urn="{ed.urn}" workUrn="{work.config.urn_work}" xml:lang="{ed.config.lang}">\n    <label xml:lang="eng">{ed.config.label}</label>\n    <description xml:lang="eng">{ed.config.label}</description>\n  </{tag}>')
        xml = f'<?xml version="1.0" encoding="UTF-8"?>\n<work xmlns="{self.CTS_NS}" groupUrn="urn:cts:greekLit:{tg}" urn="{work.config.urn_work}" xml:lang="grc">\n  <title xml:lang="eng">{work.config.label or work.config.work_key}</title>\n' + "\n".join(entries) + "\n</work>"
        (wk_dir / "__cts__.xml").write_text(xml, encoding="utf-8")
        print(f"    → CTS: {tg}/{wk}/__cts__.xml ({len(work.editions)} editions)")

    def finalize(self, works):
        print(f"\n  [CTS] Catalog files written to {self.output_dir}")


# ═══════════════════════════════════════════════════════════════
# BACKEND 3: Serverless Static Site (GitHub Pages)
# ═══════════════════════════════════════════════════════════════

class StaticSiteBackend(OutputBackend):
    """
    Generate a complete static site deployable to GitHub Pages.
    
    Produces:
      site/index.html          — self-contained viewer app
      site/corpus/…/chunks/    — JS chunk files loaded on demand
      site/.nojekyll           — tells GitHub Pages not to process with Jekyll
      site/README.md           — repo documentation
    
    This is the serverless strategy: no Django, no ATLAS, no CTS API.
    Just static files served from any CDN or GitHub Pages.
    """
    def __init__(self, output_dir):
        self.site_dir = output_dir / "site"
        self.site_dir.mkdir(parents=True, exist_ok=True)
        self.global_structures = {}
        self.global_registries = {}

    def emit_work(self, work):
        wc = work.config
        # Build text registry
        for ed_id, edition in work.editions.items():
            canonical_id = generate_canonical_id(wc.work_key, edition.config.label)
            self.global_registries[canonical_id] = {
                "urn": edition.urn, "label": edition.config.label,
                "class": edition.config.css_class, "textgroup": wc.textgroup,
                "work": wc.work, "short_id": ed_id,
            }
        # Store structure
        if work.is_flat:
            flat_key = list(work.structure.keys())[0]
            self.global_structures[wc.work_key] = work.structure[flat_key]
        else:
            self.global_structures[wc.work_key] = work.structure

        # Write JS chunk files
        baseline_ed = list(work.editions.values())[0]
        for c_idx, coord in enumerate(work.chapter_sequence):
            bk_id, ch_id = coord["book"], coord["chapter"]
            lookup_bk = bk_id if bk_id else list(baseline_ed.data.keys())[0]
            baseline_secs = list(baseline_ed.data[lookup_bk][ch_id].keys())
            sections_payload = OrderedDict()
            for sec in baseline_secs:
                sections_payload[sec] = {}
                for ed_id in wc.editions:
                    ed = work.editions.get(ed_id)
                    if ed:
                        ch_data = ed.data.get(lookup_bk, {}).get(ch_id, {})
                        sections_payload[sec][ed_id] = ch_data.get(sec, "<i>[Text range missing]</i>")

            if bk_id:
                passage_urn = f"urn:cts:greekLit:{wc.work_key}:{bk_id}.{ch_id}"
                prev_urn = f"urn:cts:greekLit:{wc.work_key}:{work.chapter_sequence[c_idx-1]['book']}.{work.chapter_sequence[c_idx-1]['chapter']}" if c_idx > 0 else None
                next_urn = f"urn:cts:greekLit:{wc.work_key}:{work.chapter_sequence[c_idx+1]['book']}.{work.chapter_sequence[c_idx+1]['chapter']}" if c_idx < len(work.chapter_sequence)-1 else None
                chunk_fn = f"chunk_b{bk_id}_ch{ch_id}.js"
            else:
                passage_urn = f"urn:cts:greekLit:{wc.work_key}:{ch_id}"
                prev_urn = f"urn:cts:greekLit:{wc.work_key}:{work.chapter_sequence[c_idx-1]['chapter']}" if c_idx > 0 else None
                next_urn = f"urn:cts:greekLit:{wc.work_key}:{work.chapter_sequence[c_idx+1]['chapter']}" if c_idx < len(work.chapter_sequence)-1 else None
                chunk_fn = f"chunk_ch{ch_id}.js"

            payload = {"urn": passage_urn, "textgroup": wc.textgroup, "work": wc.work,
                       "book": bk_id, "chapter": ch_id, "sections": sections_payload,
                       "navigation": {"prev": prev_urn, "next": next_urn}}
            chunk_dir = self.site_dir / "corpus" / wc.textgroup / wc.work / "chunks"
            chunk_dir.mkdir(parents=True, exist_ok=True)
            js = f'registerWorkspaceChunk("{passage_urn}", {json.dumps(payload, indent=2, ensure_ascii=False)});'
            (chunk_dir / chunk_fn).write_text(js, encoding="utf-8")

        print(f"    → Site: {len(work.chapter_sequence)} JS chunks")

    def finalize(self, works):
        struct_json = json.dumps(self.global_structures)
        registry_json = json.dumps(self.global_registries)

        # Build work labels for the UI
        work_labels = {}
        for w in works:
            work_labels[w.config.work_key] = w.config.label or w.config.work_key

        self._write_index_html(struct_json, registry_json, json.dumps(work_labels))
        (self.site_dir / ".nojekyll").write_text("", encoding="utf-8")
        self._write_readme()
        print(f"\n  [Site] Static site written to {self.site_dir}")
        print(f"  [Site] Deploy to GitHub Pages: push the site/ directory to your repo")
        print(f"  [Site] Or serve locally: cd {self.site_dir} && python -m http.server 8080")

    def _write_readme(self):
        readme = """# Perseus Parallel Workspace — Serverless Static Site

## Deployment to GitHub Pages

1. Push the contents of this `site/` directory to your GitHub repository
2. Enable GitHub Pages in Settings → Pages → Source: Deploy from branch
3. Set the branch and folder (e.g., `main` / `docs` or `gh-pages` / root)
4. The site will be available at `https://<user>.github.io/<repo>/`

## Local Development

```bash
cd site/
python -m http.server 8080
# Open http://localhost:8080
```

## Architecture

This is a fully serverless deployment. All text data is pre-compiled into
JavaScript chunk files under `corpus/`. The browser loads chunks on demand
via `<script>` tag injection — no API server, no database, no build step.

The `index.html` file contains the complete viewer application with both
the Parallel Grid and Classic Asymmetric reading modes.
"""
        (self.site_dir / "README.md").write_text(readme, encoding="utf-8")

    def _write_index_html(self, struct_json, registry_json, work_labels_json):
        # Read the HTML template from the companion file
        html = self._build_html_template()
        html = html.replace("STRUCT_REPLACE", struct_json)
        html = html.replace("REGISTRY_REPLACE", registry_json)
        html = html.replace("WORK_LABELS_REPLACE", work_labels_json)
        (self.site_dir / "index.html").write_text(html, encoding="utf-8")

    def _build_html_template(self):
        """
        Generate the complete self-contained HTML application.
        This is the viewer from the original notebook, refactored to use
        dynamic work labels instead of hardcoded text group names.
        """
        return '''<!DOCTYPE html>
<html>
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Perseus Parallel Workspace — Scaife-Compatible Serverless Viewer</title>
    <link href="https://fonts.googleapis.com/css2?family=Gentium+Plus&display=swap" rel="stylesheet">
    <style>
        * { margin: 0; padding: 0; box-sizing: border-box; }
        html, body { height: 100%; width: 100%; font-family: "Palatino Linotype","Book Antiqua",Palatino,Georgia,serif; font-size: 14px; color: #111; background: #fdfdfa; }
        a { color: #880000; text-decoration: none; font-weight: bold; }
        a:hover { text-decoration: underline; }

        #splash-view-root { display: flex; flex-direction: column; width: 100%; min-height: 100vh; }
        #splash-wrapper { max-width: 880px; margin: 0 auto; padding: 50px 20px; display: flex; flex-direction: column; gap: 20px; width: 100%; }
        .cta-container { text-align: center; margin: 10px 0 25px 0; border-bottom: 1px solid #e2e2d5; padding-bottom: 25px; }
        .launch-workspace-btn { display: inline-block; background: #660000; color: #fff !important; padding: 14px 45px; font-size: 16px; font-weight: bold; border-radius: 4px; }
        .launch-workspace-btn:hover { background: #440000; text-decoration: none; }
        #manifesto-render-target h1 { display: none; }
        #manifesto-render-target h2 { font-size: 18px; color: #333; margin-top: 28px; margin-bottom: 12px; font-variant: small-caps; border-bottom: 1px dashed #e2e2d5; padding-bottom: 5px; }
        #manifesto-render-target p { line-height: 1.6; margin-bottom: 14px; color: #333; text-align: justify; }
        #manifesto-render-target blockquote { border-left: 3px solid #660000; padding: 10px 15px; margin: 20px 0; font-style: italic; color: #444; }

        #app-view-root { display: none; flex-direction: column; height: 100vh; width: 100vw; overflow: hidden; background: #fff; }
        #header-container { flex-shrink: 0; background: #fff; border-bottom: 1px solid #ccc; z-index: 10; }
        #perseus-banner { background: #660000; color: #fff; padding: 5px 15px; display: flex; justify-content: space-between; align-items: center; }
        .banner-logo-title-group { display: flex; align-items: center; gap: 12px; }
        .banner-logo-title-group h1 a { color: #fff; font-size: 16px; font-weight: bold; font-family: Georgia, serif; }
        .mode-toggle-panel { display: flex; gap: 8px; }
        .toggle-btn { background: #440000; color: #ffccaa; border: 1px solid #993333; font-size: 11px; padding: 3px 10px; border-radius: 3px; cursor: pointer; font-weight: bold; }
        .toggle-btn.active-mode { background: #fff; color: #660000; border-color: #fff; }
        .toggle-btn.col4-toggle { background: #333; color: #aaa; border-color: #555; font-size: 10px; padding: 2px 8px; margin-left: 4px; }
        .toggle-btn.col4-toggle.active-mode { background: #fff; color: #333; border-color: #fff; }
        #nav-bar { background: #ddddcc; border-bottom: 1px solid #999988; padding: 4px 15px; font-size: 11px; font-weight: bold; }
        #browse-bar { background: #eee; border-bottom: 1px solid #ccc; padding: 6px 15px; font-size: 11px; display: flex; flex-direction: column; gap: 5px; }
        .browse-row { display: flex; align-items: flex-start; }
        .browse-row.hidden-row { display: none !important; }
        .browse-row label { font-weight: bold; width: 80px; color: #444; flex-shrink: 0; padding-top: 1px; }
        .browse-items { display: flex; flex-wrap: wrap; gap: 4px; }
        .browse-items a { padding: 1px 6px; background: #e0e0d0; color: #336699; border: 1px solid #bbbb99; font-weight: bold; border-radius: 2px; cursor: pointer; }
        .browse-items a.current { background: #660000; color: #fff; border-color: #330000; }
        .browse-items a.sec-pill { background: #fff; color: #555; border: 1px solid #ccc; font-weight: normal; font-size: 10px; padding: 1px 5px; }
        .browse-items a.sec-pill.active-pill { background: #660000; color: #fff; border-color: #330000; font-weight: bold; }

        #outer-wrapper { display: flex; width: 100%; flex: 1; min-height: 0; overflow: hidden; }
        #main-container { display: flex; width: 100%; height: 100%; min-width: 0; }
        .reading-column { display: flex; flex-direction: column; border-right: 1px solid #eee; background: #fff; min-width: 0; height: 100%; }
        .reading-column:last-child { border-right: none; }
        .cross-panel { background: #fafafa; }
        .column-content-scroll { flex: 1; overflow-y: auto; padding: 15px; }
        .panel-header { background: #660000; color: #fff; padding: 4px 8px; font-size: 11px; font-weight: bold; position: sticky; top: 0; z-index: 5; display: flex; justify-content: space-between; align-items: center; flex-shrink: 0; }
        .edition-select-dropdown { background: #fff; color: #333; font-size: 10px; padding: 1px 4px; border: 1px solid #ccc; border-radius: 2px; }
        .make-focus-btn { font-size: 10px; padding: 1px 6px; margin-left: 6px; background: rgba(255,255,255,0.2); color: #fff; border: 1px solid rgba(255,255,255,0.4); border-radius: 3px; cursor: pointer; }

        #classic-left-rail { display: none; width: auto; min-width: 80px; max-width: 320px; flex-shrink: 0; background: #f4f4f0; border-right: 1px solid #dcdcc8; padding: 10px 12px; overflow-y: auto; }
        #classic-left-rail h3 { font-size: 12px; font-variant: small-caps; color: #555; margin-bottom: 8px; border-bottom: 1px solid #ccccbb; padding-bottom: 3px; }
        .rail-tree-item { display: block; font-size: 12px; padding: 3px 6px; color: #336699; font-weight: bold; border-radius: 2px; margin-bottom: 2px; cursor: pointer; white-space: nowrap; }
        .rail-tree-item:hover { background: #e8e8df; }
        .rail-tree-item.active-rail-node { background: #660000; color: #fff !important; }

        #outer-wrapper.mode-parallel #main-container { flex-direction: row; }
        #outer-wrapper.mode-parallel .reading-column { flex: 1; height: 100%; }
        #outer-wrapper.mode-parallel #classic-left-rail { display: none; }
        #outer-wrapper.mode-classic #classic-left-rail { display: block; }
        #outer-wrapper.mode-classic #main-container { display: grid; grid-template-columns: 3fr 2fr; grid-template-rows: 1fr 1fr; height: 100%; width: 100%; }
        #outer-wrapper.mode-classic #col_f { grid-column: 1; grid-row: 1 / span 2; border-right: 2px solid #ccc; }
        #outer-wrapper.mode-classic #col_c1 { grid-column: 2; grid-row: 1; border-bottom: 1px solid #ccc; }
        #outer-wrapper.mode-classic #col_c2 { grid-column: 2; grid-row: 2; border-top: 1px solid #ccc; }
        #col_c3 { display: none; }
        .four-columns #col_c3 { display: flex !important; }
        .four-columns.mode-parallel .reading-column { flex: 1 !important; }

        .section-row { border-bottom: 1px solid #f0f0f0; padding: 10px 0; scroll-margin-top: 5px; }
        .section-row.hidden-section { display: none !important; }
        .greek-text { font-size: 15px; line-height: 1.7; font-family: "Gentium Plus","Athena",serif; color: #000; }
        .english-text { font-size: 13px; line-height: 1.6; color: #222; }
        .french-text { font-size: 13px; line-height: 1.6; color: #2b2b2b; }
        .german-text { font-size: 13px; line-height: 1.6; color: #222; }
        .italian-text { font-size: 13px; line-height: 1.6; color: #222; }
        .latin-text { font-size: 13px; line-height: 1.6; color: #111; font-style: italic; }
        .poetry-grid-layout { display: grid !important; grid-template-columns: 45px 1fr; row-gap: 4px; }
        .poetry-grid-layout .line-num-cell { text-align: right; padding-right: 12px; font-size: 11px; color: #999; font-weight: bold; font-family: monospace; }
        .poetry-grid-layout .line-text-cell { line-height: 1.7; }
        .poetry-grid-layout .stage-direction { grid-column: 1 / span 2; font-style: italic; color: #445566; background: #fdfdfd; border-left: 3px solid #ccc; padding: 8px 12px; margin: 10px 0; }
        .prose-inline-layout { display: block; line-height: 1.6; }
        .prose-inline-layout .prose-marker { font-weight: bold; color: #336699; margin-right: 8px; display: inline-block; }
        .prose-inline-layout .prose-body-inline { display: inline; }
        .prose-body-inline p, .prose-body-inline div { display: inline !important; margin: 0 !important; padding: 0 !important; }
        .speaker-attr { color: #445566; font-variant: small-caps; display: block; margin-top: 8px; margin-bottom: 4px; }
        .poetry-grid-layout .speaker-attr { grid-column: 1 / span 2; }
        .render-italic { font-style: italic; }
        .render-bold { font-weight: bold; }
        .note { font-size: 11px; color: #555; background: #fdfbf7; border-left: 2px solid #880000; padding: 2px 6px; margin: 4px 0 4px 1em; display: block; }
        .poetry-grid-layout .note { grid-column: 1 / span 2; }
        .quote-block { margin: 10px 0 10px 2.5em; padding-left: 12px; border-left: 1px dashed #bbb; }
        .viewport-footer-controls { display: flex; justify-content: space-between; align-items: center; gap: 8px; margin-top: 20px; padding-top: 10px; border-top: 1px solid #ddd; }
        .poetry-grid-layout .viewport-footer-controls { grid-column: 1 / span 2; display: flex !important; }
        .action-btn { display: inline-block; background: #660000; color: #fff !important; padding: 4px 10px; font-size: 11px; font-weight: bold; border-radius: 2px; border: none; cursor: pointer; }
        .action-btn.secondary { background: #666; }
        .action-btn:hover { background: #330000; }
        .action-btn.secondary:hover { background: #444; }
        .resize-handle { flex-shrink: 0; background: #e0e0e0; position: relative; z-index: 20; }
        .resize-handle:hover, .resize-handle.dragging { background: #660000; }
        .resize-handle-h { width: 5px; cursor: col-resize; height: 100%; }
    </style>
</head>
<body>
  <div id="splash-view-root">
    <div id="splash-wrapper">
      <div class="cta-container">
        <a href="#" class="launch-workspace-btn" onclick="launchWorkspace(event)">Launch Parallel Workspace &rarr;</a>
      </div>
      <div id="manifesto-render-target">
        <h2>Perseus Parallel Workspace &mdash; Serverless Static Deployment</h2>
        <p>This is a fully serverless reading environment for pre-modern text collections, compatible with the <a href="https://scaife.perseus.org/">Scaife Viewer</a> ecosystem. All text data is pre-compiled into static JavaScript chunks. No API server, no database &mdash; just HTML files served from GitHub Pages or any CDN.</p>
        <h2>What can you do here?</h2>
        <p>Compare multiple editions and translations side by side in the <strong>Parallel Grid</strong> view, or use the <strong>Classic Asymmetric</strong> layout inspired by Perseus 4. Navigate by book, chapter, and section. Switch editions in any column independently.</p>
      </div>
    </div>
  </div>

  <div id="app-view-root">
    <div id="header-container">
      <div id="perseus-banner">
        <div class="banner-logo-title-group">
          <h1><a href="#" onclick="window.location.reload()">Perseus Workspace</a></h1>
        </div>
        <div class="mode-toggle-panel">
          <button id="btn-mode-parallel" class="toggle-btn active-mode" onclick="setWorkspaceLayoutMode('parallel')">Parallel Grid</button>
          <button id="btn-mode-classic" class="toggle-btn" onclick="setWorkspaceLayoutMode('classic')">Classic Asymmetric</button>
          <button id="btn-toggle-col4" class="toggle-btn col4-toggle" onclick="toggleFourthColumn()">+Col 4</button>
        </div>
      </div>
      <div id="nav-bar"><span id="frame-context-label">Loading...</span></div>
      <div id="browse-bar">
        <div class="browse-row" id="row-work-selector"><label>Works:</label><div class="browse-items" id="work-items-container"></div></div>
        <div class="browse-row" id="row-book-container"><label>Books:</label><div class="browse-items" id="book-items-container"></div></div>
        <div class="browse-row" id="row-chapter-selector"><label id="chapter-row-label">Chapters:</label><div class="browse-items" id="chapter-items-container"></div></div>
        <div class="browse-row" id="row-section-selector"><label>Sections:</label><div class="browse-items" id="section-items-container"></div></div>
      </div>
    </div>
    <div id="outer-wrapper" class="mode-parallel">
      <div id="classic-left-rail"><h3 id="rail-title">Navigation</h3><div id="rail-nodes-target"></div></div>
      <div id="main-container">
        <div class="reading-column" id="col_f"><div class="panel-header"><span>Focus</span><select class="edition-select-dropdown" id="select_f" onchange="updateColumnContent('f', this.value)"></select></div><div class="column-content-scroll" id="content_f"></div></div>
        <div class="resize-handle resize-handle-h" id="resize-f-c1"></div>
        <div class="reading-column cross-panel" id="col_c1"><div class="panel-header"><span>Comparison 1 <button class="make-focus-btn" onclick="swapToFocus('c1')">⇦ Focus</button></span><select class="edition-select-dropdown" id="select_c1" onchange="updateColumnContent('c1', this.value)"></select></div><div class="column-content-scroll" id="content_c1"></div></div>
        <div class="resize-handle resize-handle-h" id="resize-c1-c2"></div>
        <div class="reading-column cross-panel" id="col_c2"><div class="panel-header"><span>Comparison 2 <button class="make-focus-btn" onclick="swapToFocus('c2')">⇦ Focus</button></span><select class="edition-select-dropdown" id="select_c2" onchange="updateColumnContent('c2', this.value)"></select></div><div class="column-content-scroll" id="content_c2"></div></div>
        <div class="resize-handle resize-handle-h" id="resize-c2-c3"></div>
        <div class="reading-column cross-panel" id="col_c3"><div class="panel-header"><span>Comparison 3 <button class="make-focus-btn" onclick="swapToFocus('c3')">⇦ Focus</button></span><select class="edition-select-dropdown" id="select_c3" onchange="updateColumnContent('c3', this.value)"></select></div><div class="column-content-scroll" id="content_c3"></div></div>
      </div>
    </div>
  </div>

  <script>
    const GLOBAL_STRUCTURES = STRUCT_REPLACE;
    const TEXT_REGISTRY = REGISTRY_REPLACE;
    const WORK_LABELS = WORK_LABELS_REPLACE;
    const WORKSPACE_DATA_REGISTRY = new Map();
    let activeWorkKey = "", activeUrnContext = "", activeSectionFilter = null, currentActiveMode = "parallel";
    let columnEditions = { f: "", c1: "", c2: "", c3: "" };
    let fourthColumnActive = false;

    function launchWorkspace(e) { if(e) e.preventDefault(); document.getElementById("splash-view-root").style.display="none"; document.getElementById("app-view-root").style.display="flex"; initializeRoutingFromURL(); }
    window.registerWorkspaceChunk = function(urn, data) { WORKSPACE_DATA_REGISTRY.set(urn, data); triggerRenderLifecycle(urn); };

    document.addEventListener("DOMContentLoaded", function() {
        const params = new URLSearchParams(window.location.search);
        if (params.has("w")) { document.getElementById("splash-view-root").style.display="none"; document.getElementById("app-view-root").style.display="flex"; initializeRoutingFromURL(); }
    });

    function isFlatStructure(wKey) { return Array.isArray(GLOBAL_STRUCTURES[wKey]); }
    function isPoetryWork(wKey) { return wKey.startsWith("tlg0011."); }

    function populateDropdownsForWork(tgId, wkId) {
        const prefix = tgId + "_" + wkId + "_";
        const valid = Object.keys(TEXT_REGISTRY).filter(v => v.startsWith(prefix));
        if (!valid.length) return;
        if (!columnEditions.f || !valid.includes(columnEditions.f)) columnEditions.f = valid[0];
        if (!columnEditions.c1 || !valid.includes(columnEditions.c1)) columnEditions.c1 = valid[1] || valid[0];
        if (!columnEditions.c2 || !valid.includes(columnEditions.c2)) columnEditions.c2 = valid[2] || valid[0];
        if (!columnEditions.c3 || !valid.includes(columnEditions.c3)) columnEditions.c3 = valid[3] || valid[0];
        ["f","c1","c2","c3"].forEach(p => {
            const sel = document.getElementById("select_"+p); if(!sel) return; sel.innerHTML="";
            valid.forEach(v => { const o=document.createElement("option"); o.value=v; o.innerText=TEXT_REGISTRY[v].label; if(columnEditions[p]===v) o.selected=true; sel.appendChild(o); });
        });
    }

    function initializeRoutingFromURL() {
        const params = new URLSearchParams(window.location.search);
        const raw = params.get("w") || "";
        let b="", ch=""; activeSectionFilter=null;
        columnEditions.c1=params.get("right")||""; columnEditions.c2=params.get("right2")||""; columnEditions.c3=params.get("right3")||"";
        if (params.get("cols")==="4") setTimeout(()=>toggleFourthColumn(),0);
        if (raw.includes(":")) {
            const segs=raw.split(":"); let wp=segs[0];
            if (TEXT_REGISTRY[wp]) { columnEditions.f=wp; const pts=wp.split("_"); wp=pts[0]+"."+pts[1]; } else { columnEditions.f=""; }
            activeWorkKey=wp; const ps=segs[1].split(".");
            if (isFlatStructure(activeWorkKey)) { b=""; ch=ps[0]||"1"; if(ps[1]) activeSectionFilter=ps[1]; }
            else { b=ps[0]; ch=ps[1]||"1"; if(ps[2]) activeSectionFilter=ps[2]; }
        } else {
            activeWorkKey=raw||Object.keys(GLOBAL_STRUCTURES)[0];
            if(!GLOBAL_STRUCTURES[activeWorkKey]) activeWorkKey=Object.keys(GLOBAL_STRUCTURES)[0];
            if (isFlatStructure(activeWorkKey)) { b=""; ch=GLOBAL_STRUCTURES[activeWorkKey][0]||"1"; }
            else { b=Object.keys(GLOBAL_STRUCTURES[activeWorkKey])[0]; ch=GLOBAL_STRUCTURES[activeWorkKey][b][0]||"1"; }
            columnEditions.f="";
        }
        const urn = b ? "urn:cts:greekLit:"+activeWorkKey+":"+b+"."+ch : "urn:cts:greekLit:"+activeWorkKey+":"+ch;
        resolveAndInjectUrn(urn);
    }

    function resolveAndInjectUrn(urn) {
        if (WORKSPACE_DATA_REGISTRY.has(urn)) { triggerRenderLifecycle(urn); return; }
        document.getElementById("frame-context-label").innerText = "Loading "+urn+"...";
        const parts=urn.split(":"), wc=parts[3].split("."), pc=parts[4].split(".");
        populateDropdownsForWork(wc[0],wc[1]);
        const script=document.createElement("script");
        const base=window.location.pathname.substring(0,window.location.pathname.lastIndexOf("/")+1);
        if (pc.length>1 && !isFlatStructure(wc[0]+"."+wc[1])) script.src=base+"corpus/"+wc[0]+"/"+wc[1]+"/chunks/chunk_b"+pc[0]+"_ch"+pc[1]+".js";
        else script.src=base+"corpus/"+wc[0]+"/"+wc[1]+"/chunks/chunk_ch"+pc[0]+".js";
        script.onerror=()=>{document.getElementById("frame-context-label").innerText="Failed: "+urn;};
        document.head.appendChild(script);
    }

    function triggerRenderLifecycle(urn) {
        activeUrnContext=urn; const payload=WORKSPACE_DATA_REGISTRY.get(urn); if(!payload) return;
        activeWorkKey=payload.textgroup+"."+payload.work;
        populateDropdownsForWork(payload.textgroup,payload.work);
        updateURLState(payload.book,payload.chapter);
        renderNavigationControls(payload);
        renderActiveContentLayers(payload);
    }

    function updateURLState(book,chapter) {
        const params=new URLSearchParams(); let pv=book?book+"."+chapter:chapter;
        if(activeSectionFilter) pv+="."+activeSectionFilter;
        params.set("w",activeWorkKey+"."+columnEditions.f+":"+pv);
        if(columnEditions.c1) params.set("right",columnEditions.c1);
        if(columnEditions.c2) params.set("right2",columnEditions.c2);
        if(columnEditions.c3) params.set("right3",columnEditions.c3);
        if(fourthColumnActive) params.set("cols","4");
        window.history.replaceState(null,"",window.location.pathname+"?"+params.toString());
    }

    window.selectSectionDirectly = function(secId) { activeSectionFilter=secId; if(WORKSPACE_DATA_REGISTRY.has(activeUrnContext)) triggerRenderLifecycle(activeUrnContext); };
    window.clearSectionFilter = function(e) { if(e) e.preventDefault(); activeSectionFilter=null; if(WORKSPACE_DATA_REGISTRY.has(activeUrnContext)) triggerRenderLifecycle(activeUrnContext); };
    window.updateColumnContent = function(prefix,value) { columnEditions[prefix]=value; if(activeUrnContext&&WORKSPACE_DATA_REGISTRY.has(activeUrnContext)){const p=WORKSPACE_DATA_REGISTRY.get(activeUrnContext);updateURLState(p.book,p.chapter);renderActiveContentLayers(p);} };
    window.swapToFocus = function(src) { const f=columnEditions.f,s=columnEditions[src]; columnEditions.f=s; columnEditions[src]=f; const sf=document.getElementById("select_f"),ss=document.getElementById("select_"+src); if(sf)sf.value=s; if(ss)ss.value=f; if(activeUrnContext&&WORKSPACE_DATA_REGISTRY.has(activeUrnContext)){const p=WORKSPACE_DATA_REGISTRY.get(activeUrnContext);updateURLState(p.book,p.chapter);renderActiveContentLayers(p);} };

    window.setWorkspaceLayoutMode = function(mode) {
        currentActiveMode=mode; const frame=document.getElementById("outer-wrapper");
        const bp=document.getElementById("btn-mode-parallel"), bc=document.getElementById("btn-mode-classic");
        if(mode==="classic"){frame.className="mode-classic"+(fourthColumnActive?" four-columns":"");bc.classList.add("active-mode");bp.classList.remove("active-mode");}
        else{frame.className="mode-parallel"+(fourthColumnActive?" four-columns":"");bp.classList.add("active-mode");bc.classList.remove("active-mode");}
        if(WORKSPACE_DATA_REGISTRY.has(activeUrnContext)) renderNavigationControls(WORKSPACE_DATA_REGISTRY.get(activeUrnContext));
    };

    window.toggleFourthColumn = function() {
        fourthColumnActive=!fourthColumnActive; const btn=document.getElementById("btn-toggle-col4"), ow=document.getElementById("outer-wrapper");
        if(fourthColumnActive){ow.classList.add("four-columns");btn.classList.add("active-mode");btn.innerText="−Col 4";}
        else{ow.classList.remove("four-columns");btn.classList.remove("active-mode");btn.innerText="+Col 4";}
        if(fourthColumnActive&&activeUrnContext&&WORKSPACE_DATA_REGISTRY.has(activeUrnContext)) renderActiveContentLayers(WORKSPACE_DATA_REGISTRY.get(activeUrnContext));
    };

    function renderNavigationControls(payload) {
        document.getElementById("frame-context-label").innerText = "Context: "+payload.urn;
        const poetry=isPoetryWork(activeWorkKey);
        const rowWork=document.getElementById("row-work-selector"),rowBook=document.getElementById("row-book-container"),rowCh=document.getElementById("row-chapter-selector"),rowSec=document.getElementById("row-section-selector");
        if(currentActiveMode==="classic"){if(rowWork)rowWork.classList.add("hidden-row");if(rowBook)rowBook.classList.add("hidden-row");if(rowCh)rowCh.classList.add("hidden-row");if(rowSec){if(poetry)rowSec.classList.add("hidden-row");else rowSec.classList.remove("hidden-row");}}
        else{if(rowWork)rowWork.classList.remove("hidden-row");if(rowCh)rowCh.classList.remove("hidden-row");if(rowSec){if(poetry)rowSec.classList.add("hidden-row");else rowSec.classList.remove("hidden-row");}if(rowBook){if(isFlatStructure(activeWorkKey))rowBook.classList.add("hidden-row");else rowBook.classList.remove("hidden-row");}}

        const wc=document.getElementById("work-items-container");
        if(wc){wc.innerHTML="";Object.keys(GLOBAL_STRUCTURES).forEach(wk=>{const a=document.createElement("a");a.innerText=WORK_LABELS[wk]||wk;if(wk===activeWorkKey)a.className="current";a.onclick=()=>{activeSectionFilter=null;activeWorkKey=wk;if(isFlatStructure(wk))resolveAndInjectUrn("urn:cts:greekLit:"+wk+":"+GLOBAL_STRUCTURES[wk][0]);else{const tb=Object.keys(GLOBAL_STRUCTURES[wk])[0];resolveAndInjectUrn("urn:cts:greekLit:"+wk+":"+tb+"."+GLOBAL_STRUCTURES[wk][tb][0]);}};wc.appendChild(a);});}

        if(currentActiveMode!=="classic"&&!isFlatStructure(activeWorkKey)){const bc=document.getElementById("book-items-container");if(bc){bc.innerHTML="";Object.keys(GLOBAL_STRUCTURES[activeWorkKey]).forEach(bk=>{const a=document.createElement("a");a.innerText=bk;if(bk===payload.book)a.className="current";a.onclick=()=>{activeSectionFilter=null;resolveAndInjectUrn("urn:cts:greekLit:"+activeWorkKey+":"+bk+"."+GLOBAL_STRUCTURES[activeWorkKey][bk][0]);};bc.appendChild(a);});}}

        const clbl=document.getElementById("chapter-row-label"); if(clbl)clbl.innerText=poetry?"Lines:":"Chapters:";
        const cc=document.getElementById("chapter-items-container");
        if(cc){cc.innerHTML="";const cl=isFlatStructure(activeWorkKey)?GLOBAL_STRUCTURES[activeWorkKey]:GLOBAL_STRUCTURES[activeWorkKey][payload.book];
        cl.forEach(ch=>{const a=document.createElement("a");a.innerText=ch;if(ch===payload.chapter)a.className="current";a.onclick=()=>{activeSectionFilter=null;const u=payload.book?"urn:cts:greekLit:"+activeWorkKey+":"+payload.book+"."+ch:"urn:cts:greekLit:"+activeWorkKey+":"+ch;resolveAndInjectUrn(u);};cc.appendChild(a);});}

        if(!poetry){const sc=document.getElementById("section-items-container");if(sc){sc.innerHTML="";Object.keys(payload.sections).forEach(sec=>{const a=document.createElement("a");a.innerText=sec;a.className="sec-pill";if(sec===activeSectionFilter)a.classList.add("active-pill");a.onclick=()=>selectSectionDirectly(sec);sc.appendChild(a);});}}

        const rn=document.getElementById("rail-nodes-target"),rt=document.getElementById("rail-title");
        if(rn){rn.innerHTML="";if(isFlatStructure(activeWorkKey)){rt.innerText=poetry?"Lines":"Chapters";GLOBAL_STRUCTURES[activeWorkKey].forEach(ch=>{const n=document.createElement("div");n.className="rail-tree-item "+(!payload.book&&payload.chapter===ch?"active-rail-node":"");n.innerText=(poetry?"Lines ":"Ch ")+ch;n.onclick=()=>{activeSectionFilter=null;resolveAndInjectUrn("urn:cts:greekLit:"+activeWorkKey+":"+ch);};rn.appendChild(n);});}
        else{rt.innerText="Book "+payload.book;GLOBAL_STRUCTURES[activeWorkKey][payload.book].forEach(ch=>{const n=document.createElement("div");n.className="rail-tree-item "+(payload.chapter===ch?"active-rail-node":"");n.innerText="Ch "+ch;n.onclick=()=>{activeSectionFilter=null;resolveAndInjectUrn("urn:cts:greekLit:"+activeWorkKey+":"+payload.book+"."+ch);};rn.appendChild(n);});}}
    }

    function renderActiveContentLayers(payload) {
        ["f","c1","c2","c3"].forEach(prefix=>{
            const tc=document.getElementById("content_"+prefix); if(!tc) return; tc.innerHTML="";
            const vId=columnEditions[prefix], meta=TEXT_REGISTRY[vId], css=meta?meta.class:"english-text", sid=meta?meta.short_id:vId, poetry=isPoetryWork(activeWorkKey);
            Object.keys(payload.sections).forEach(sec=>{
                const hidden=activeSectionFilter!==null&&activeSectionFilter!==sec;
                const row=document.createElement("div"); row.className="section-row s-idx-"+sec+(hidden?" hidden-section":"");
                const txt=payload.sections[sec][sid]||"<i>[Missing]</i>"; const label=poetry?sec:"["+sec+"]";
                if(poetry){
                    const w=document.createElement("div"); w.className=css+" poetry-grid-layout";
                    const tmp=document.createElement("div"); tmp.innerHTML=txt;
                    while(tmp.firstChild){const c=tmp.firstChild;if(c.nodeType===1&&c.classList.contains("verse-line")){const ln=c.getAttribute("data-line")||"";const nc=document.createElement("span");nc.className="line-num-cell";nc.innerText=ln;const lc=document.createElement("div");lc.className="line-text-cell";lc.innerHTML=c.innerHTML;w.appendChild(nc);w.appendChild(lc);tmp.removeChild(c);}else{if(c.nodeType===1){c.style.gridColumn="1 / span 2";}w.appendChild(c);}}
                    row.appendChild(w);
                } else { row.innerHTML='<div class="prose-inline-layout '+css+'"><span class="prose-marker"><a href="javascript:void(0)" onclick="selectSectionDirectly(\''+sec+'\')">'+label+'</a></span><div class="prose-body-inline">'+txt+'</div></div>'; }
                tc.appendChild(row);
            });
            const footer=document.createElement("div"); footer.className="viewport-footer-controls"; let prev="",next="";
            if(activeSectionFilter){const sa=Object.keys(payload.sections),ci=sa.indexOf(activeSectionFilter);if(ci>0)prev='<a onclick="selectSectionDirectly(\''+sa[ci-1]+'\')" class="action-btn">&larr; ['+sa[ci-1]+']</a>';else if(payload.navigation.prev)prev='<a class="action-btn" onclick="loadAdjacentUrn(\''+payload.navigation.prev+'\',true)">&larr; Prev</a>';if(ci<sa.length-1)next='<a onclick="selectSectionDirectly(\''+sa[ci+1]+'\')" class="action-btn">['+sa[ci+1]+'] &rarr;</a>';else if(payload.navigation.next)next='<a class="action-btn" onclick="loadAdjacentUrn(\''+payload.navigation.next+'\',false)">Next &rarr;</a>';}
            else{if(payload.navigation.prev)prev='<a class="action-btn" onclick="loadAdjacentUrn(\''+payload.navigation.prev+'\')">&larr; Prev</a>';if(payload.navigation.next)next='<a class="action-btn" onclick="loadAdjacentUrn(\''+payload.navigation.next+'\')">Next &rarr;</a>';}
            footer.innerHTML='<div style="display:flex;gap:6px">'+prev+' <a href="javascript:void(0)" class="action-btn secondary" onclick="clearSectionFilter(event)">Full View</a></div>'+next;
            const pg=tc.querySelector(".poetry-grid-layout"); if(pg&&poetry)pg.appendChild(footer); else tc.appendChild(footer);
        });
    }

    window.loadAdjacentUrn = function(urn, selectLast) {
        activeSectionFilter=null;
        if(selectLast&&WORKSPACE_DATA_REGISTRY.has(urn)){const d=WORKSPACE_DATA_REGISTRY.get(urn),s=Object.keys(d.sections);activeSectionFilter=s.length?s[s.length-1]:null;triggerRenderLifecycle(urn);}
        else resolveAndInjectUrn(urn);
    };

    // Resize handles
    document.addEventListener("DOMContentLoaded",function(){
        document.querySelectorAll(".resize-handle").forEach(h=>{
            h.addEventListener("mousedown",function(e){
                e.preventDefault(); const handle=e.currentTarget, container=document.getElementById("main-container");
                handle.classList.add("dragging"); document.body.classList.add("resizing-cols");
                const cols=Array.from(container.querySelectorAll(".reading-column")), hi=Array.from(container.children).indexOf(handle);
                const prev=container.children[hi-1],next=container.children[hi+1]; if(!prev||!next)return;
                const startX=e.clientX, pw=prev.getBoundingClientRect().width, nw=next.getBoundingClientRect().width;
                cols.forEach(c=>{const w=c.getBoundingClientRect().width;c.style.flex="none";c.style.width=w+"px";});
                function onMove(ev){const dx=ev.clientX-startX;const np=Math.max(100,pw+dx),nn=Math.max(100,nw-dx);prev.style.width=np+"px";next.style.width=nn+"px";}
                function onEnd(){handle.classList.remove("dragging");document.body.classList.remove("resizing-cols");document.removeEventListener("mousemove",onMove);document.removeEventListener("mouseup",onEnd);}
                document.addEventListener("mousemove",onMove);document.addEventListener("mouseup",onEnd);
            });
        });
    });
  </script>
</body>
</html>'''

print("✓ Output backends loaded (ATLAS JSONL, CTS Catalog, Static Site).")
````

## Cell 5 — Run Pipeline

Execute the full pipeline: parse all works, emit to all active backends.

````python
# ═══════════════════════════════════════════════════════════════
# PIPELINE ORCHESTRATOR
# ═══════════════════════════════════════════════════════════════

def parse_work(wc):
    """Parse all editions of a work into a ParsedWork."""
    print(f"\nIngesting: {wc.work_key} ({wc.label})")
    is_poetry = any(cfg.parse_mode == "poetry_cards" for cfg in wc.editions.values())
    master_intervals = build_poetry_canonical_intervals(wc.editions) if is_poetry else None

    parsed_editions = {}
    for ed_id, cfg in wc.editions.items():
        urn = f"urn:cts:greekLit:{wc.work_key}.{ed_id}"
        data = parse_edition(cfg, master_intervals)
        if data and sum(len(s) for c in data.values() for s in c.values()) > 0:
            seg = sum(len(s) for c in data.values() for s in c.values())
            parsed_editions[ed_id] = ParsedEdition(edition_id=ed_id, config=cfg, urn=urn, data=data)
            print(f"  ✓ {ed_id}: {seg} segments")
        else:
            print(f"  ✗ {ed_id}: parse failed")

    if not parsed_editions:
        return None

    baseline = list(parsed_editions.values())[0]
    has_books = len(baseline.data.keys()) > 1
    structure = OrderedDict()
    chapter_seq = []

    if has_books:
        for bk, ch_dict in baseline.data.items():
            structure[bk] = list(ch_dict.keys())
            for ck in ch_dict.keys(): chapter_seq.append({"book": bk, "chapter": ck})
    else:
        single_bk = list(baseline.data.keys())[0]
        structure["_flat_chapters"] = list(baseline.data[single_bk].keys())
        for ck in baseline.data[single_bk].keys(): chapter_seq.append({"book": None, "chapter": ck})

    return ParsedWork(config=wc, editions=parsed_editions, structure=structure,
                      is_flat=not has_books, is_poetry=is_poetry, chapter_sequence=chapter_seq)


# ── Initialize backends ──
backends = []
if "atlas" in ACTIVE_BACKENDS:
    backends.append(AtlasJsonlBackend(OUTPUT_DIR))
if "cts" in ACTIVE_BACKENDS:
    backends.append(CtsMetadataBackend(OUTPUT_DIR))
if "site" in ACTIVE_BACKENDS:
    backends.append(StaticSiteBackend(OUTPUT_DIR))

print("=" * 65)
print("Perseus CTS Pipeline — Scaife-Compatible Ingestion")
print("=" * 65)

parsed_works = []
for wc in WORK_REGISTRY:
    pw = parse_work(wc)
    if pw:
        parsed_works.append(pw)
        for backend in backends:
            backend.emit_work(pw)

for backend in backends:
    backend.finalize(parsed_works)

print("\n" + "=" * 65)
print(f"[SUCCESS] Pipeline completed: {len(parsed_works)} works processed")
print(f"Output directory: {OUTPUT_DIR}")
print("=" * 65)
````
