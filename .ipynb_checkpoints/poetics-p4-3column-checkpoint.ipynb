```python
import os
import re
import time
from pathlib import Path
from collections import OrderedDict
import xml.etree.ElementTree as ET

# ── Local Configuration for Aristotle's Poetics ────────────────
OUT_DIR = Path("/Users/gcrane/Downloads/gemsite/poetics_site2")
OUT_DIR.mkdir(parents=True, exist_ok=True)

NS = {'tei': 'http://www.tei-c.org/ns/1.0'}

# 1. Register an arbitrary number of localized versions dynamically
TEXT_REGISTRY = {
    "grc_kassel": {
        "path": "/Users/gcrane/github/canonical-greekLit/data/tlg0086/tlg034/tlg0086.tlg034.perseus-grc2.xml",
        "label": "Greek (Kassel, 1965)",
        "class": "greek-text",
        "is_greek": True
    },
    "grc_digi": {
        "path": "/Users/gcrane/github/canonical-greekLit/data/tlg0086/tlg034/tlg0086.tlg034.digicorpus-grc2.xml",
        "label": "Greek (Digital Corpus Variant)",
        "class": "greek-text",
        "is_greek": True
    },
    "eng_fyfe": {
        "path": "/Users/gcrane/github/canonical-greekLit/data/tlg0086/tlg034/tlg0086.tlg034.perseus-eng2.xml",
        "label": "English (W.H. Fyfe, 1927)",
        "class": "english-text",
        "is_greek": False
    },
    "eng_butcher": {
        "path": "/Users/gcrane/github/Poetics2.0/eng/tlg0086.tlg034.butcher1911.xml",
        "label": "English (S.H. Butcher, 1911)",
        "class": "english-text",
        "is_greek": False
    }
}

# ── Hopper Layout CSS (Global Configuration Block & State Persistence) ──
PERSEUS_CSS = r"""
* { margin: 0; padding: 0; box-sizing: border-box; }
html, body {
    height: 100%;
    overflow: hidden; 
    font-family: "Palatino Linotype", "Book Antiqua", Palatino, Georgia, serif;
    font-size: 13px;
    color: #000;
    background: #fff;
}
a { color: #336699; text-decoration: none; }
a:hover { text-decoration: underline; }

/* Fixed Header Layout Block */
#header-container {
    flex-shrink: 0;
    background: #fff;
    z-index: 10;
}
#perseus-banner { background: #660000; color: #fff; padding: 8px 15px; display: flex; justify-content: space-between; align-items: center; border-bottom: 2px solid #330000; }
#perseus-banner h1 { font-size: 18px; font-weight: bold; }
#perseus-banner h1 a { color: #fff; }
#perseus-banner .doc-title { font-size: 12px; color: #ffccaa; font-weight: bold; }

/* Navigation Ribbon */
#nav-bar { 
    background: #ddddcc; 
    border-bottom: 1px solid #999988; 
    padding: 5px 15px; 
    font-size: 11px; 
    color: #555;
    display: flex;
    align-items: center;
}
#nav-bar a { margin-right: 15px; color: #003366; font-weight: bold; }

/* Pure CSS Hash State Persistence Infrastructure */
.toc-toggle-btn {
    background: #660000;
    color: #fff !important;
    padding: 2px 8px;
    border-radius: 3px;
    font-weight: bold;
    margin-right: 20px;
}
.show-toc-label { display: none; }
.hide-toc-label { display: inline; }

/* Trigger State Shifts when URL hash contains '#notoc' */
#outer-wrapper:target #sidebar-toc {
    width: 0px;
    padding: 0px;
    border-right: none;
    overflow: hidden;
}
#outer-wrapper:target ~ #header-container #nav-bar .show-toc-label { display: inline; }
#outer-wrapper:target ~ #header-container #nav-bar .hide-toc-label { display: none; }

/* Hopper Visual Navigation Matrix Bar */
#browse-bar { background: #eeeeee; border-bottom: 1px solid #cccccc; padding: 8px 15px; font-size: 11px; }
.browse-row { margin-bottom: 4px; display: flex; align-items: center; }
.browse-row label { font-weight: bold; width: 100px; color: #555; flex-shrink: 0; }
.browse-items { display: flex; flex-wrap: wrap; gap: 3px; align-items: center; }

.nav-tile-wrapper { display: inline-flex; flex-direction: column; align-items: center; text-align: center; }
.nav-tile-link { display: block; padding: 2px 6px; background: #e0e0d0; color: #336699; border: 1px solid #bbbb99; min-width: 24px; text-align: center; font-weight: bold; }
.nav-tile-link:hover { background: #fff; text-decoration: none; border-color: #660000; }
.nav-tile-link.current { background: #660000; color: #fff !important; border-color: #330000; }
.size-bar { height: 3px; background: #990000; margin-top: 2px; border-radius: 1px; min-width: 4px; }

/* Global Configuration Workspace Panel (Now spanning above all columns) */
#layout-configurator { 
    background: #fdfae6; 
    border-bottom: 1px dashed #d6c57d; 
    padding: 8px 15px; 
    font-size: 11px; 
}
.config-group { margin-bottom: 4px; display: flex; align-items: center; }
.config-group:last-child { margin-bottom: 0; }
.config-group strong { width: 120px; color: #555; flex-shrink: 0; }
.config-group-items { display: flex; flex-wrap: wrap; gap: 2px; }
.config-group a { margin-right: 10px; font-weight: normal; padding: 1px 4px; }
.config-group a.active { background: #660000; color: #fff; font-weight: bold; border-radius: 2px; text-decoration: none; pointer-events: none; }

/* Main Screen Layout Container splits */
#app-view-root { display: flex; flex-direction: column-reverse; height: 100vh; } /* column-reverse permits target selector bubbles to match layout-configurator backwards in html tree */
#outer-wrapper { display: flex; flex: 1; min-height: 0; }

/* Sidebar Layout */
#sidebar-toc { 
    width: 240px; 
    background: #f5f5ee; 
    border-right: 1px solid #ccccbb; 
    padding: 12px; 
    font-size: 11px; 
    overflow-y: auto; 
    flex-shrink: 0;
    transition: width 0.15s ease, padding 0.15s ease;
}
#sidebar-toc h3 { font-size: 12px; color: #660000; margin-bottom: 8px; border-bottom: 1px solid #ccccbb; padding-bottom: 3px; }
#sidebar-toc ul { list-style: none; padding-left: 0; }
#sidebar-toc li { margin-bottom: 5px; }
#sidebar-toc li.active-book { font-weight: bold; color: #660000; }

.toc-chapters { padding-left: 10px; margin-top: 3px; font-weight: normal; }
.toc-chapters li { margin-bottom: 3px; }
.toc-chapters a { color: #003366; }
.toc-chapters a.current { color: #660000; font-weight: bold; }
.toc-sections { padding-left: 12px; margin-top: 2px; display: flex; flex-wrap: wrap; gap: 4px; font-weight: normal; }
.toc-sections a { background: #e0e0d4; padding: 1px 4px; border-radius: 2px; color: #333; font-size: 10px; }
.toc-sections a:hover { background: #660000; color: #fff; text-decoration: none; }
.toc-sections a.current { background: #660000; color: #fff !important; font-weight: bold; }

/* Alignment Matrix Columns */
#main-container { 
    flex: 1; 
    display: flex; 
    min-width: 0;
    height: 100%;
}
.reading-column {
    flex: 1;
    min-width: 0;
    padding: 15px;
    overflow-y: auto;
    height: 100%;
}
#focus-text {
    background: #fff;
}
.cross-panel { 
    border-left: 2px solid #eeeeee; 
    background: #fafafa;
}
.panel-header { 
    background: #660000; 
    color: #fff; 
    padding: 4px 8px; 
    font-size: 11px; 
    font-weight: bold; 
    margin-bottom: 10px;
    position: sticky;
    top: 0;
    z-index: 5;
}

.tier-indicator { background: #fffccb; border: 1px solid #e6db55; padding: 4px 8px; font-size: 11px; margin-bottom: 10px; font-weight: bold; color: #555; display: inline-block; }
.nav-arrows { display: flex; align-items: center; gap: 10px; font-size: 12px; font-weight: bold; margin: 8px 0; border-bottom: 1px dotted #ccc; padding-bottom: 6px; }
.nav-arrows a { font-size: 14px; color: #660000; }

.section-block { margin-bottom: 18px; position: relative; padding-left: 45px; line-height: 1.6; }
.section-block .sec-num { position: absolute; left: 0; top: 0; color: #990000; font-size: 11px; font-weight: bold; width: 35px; text-align: right; }
.section-block .sec-num a { color: #990000; text-decoration: underline; }
.greek-text { font-size: 15px; line-height: 1.7; font-family: "Gentium Plus", "Athena", Palatino, serif; }
.english-text { font-size: 13px; color: #111; }
.note { font-size: 11px; color: #666; font-style: italic; background: #f0f0f0; padding: 0 2px; }
"""

# ── Namespace-Agnostic XML Parser Engine ───────────────────────
def extract_text_recursive(elem):
    parts = []
    if elem.text: parts.append(elem.text)
    for child in elem:
        tag = child.tag.replace('{http://www.tei-c.org/ns/1.0}', '')
        if tag == 'note':
            note_text = extract_text_recursive(child).strip()
            if note_text: parts.append(f'<span class="note">[{note_text}]</span>')
        elif tag in ('milestone', 'reg'): pass
        else: parts.append(extract_text_recursive(child))
        if child.tail: parts.append(child.tail)
    return ''.join(parts)

def parse_tei_hierarchy(path):
    if not os.path.exists(path): return None
    tree = ET.parse(path)
    root = tree.getroot()
    body = root.find('.//tei:body', NS) or root.find('.//body')
    if body is None: return None
        
    top_div = body.find('tei:div', NS) or body.find('div') or body
    data = OrderedDict()
    textparts = top_div.findall('tei:div', NS) or top_div.findall('div')
    
    if not textparts:
        textparts = top_div.findall('.//tei:div[@subtype="chapter"]', NS) or top_div.findall('.//div[@subtype="chapter"]')

    if textparts and (textparts[0].get('subtype') == 'chapter' or textparts[0].get('type') == 'chapter'):
        data['1'] = OrderedDict()
        for ch_div in textparts:
            ch_n = ch_div.get('n')
            if not ch_n: continue
            data['1'][ch_n] = OrderedDict()
            sections = ch_div.findall('tei:div[@subtype="section"]', NS) or ch_div.findall('div[@subtype="section"]')
            if not sections:
                sections = ch_div.findall('tei:div', NS) or ch_div.findall('div') or ch_div.findall('tei:p', NS) or ch_div.findall('p')
                
            for idx, sec_div in enumerate(sections):
                sec_n = sec_div.get('n') or str(idx + 1)
                paragraphs = sec_div.findall('tei:p', NS) or sec_div.findall('p')
                if paragraphs:
                    data['1'][ch_n][sec_n] = ' '.join(extract_text_recursive(p).strip() for p in paragraphs)
                else:
                    data['1'][ch_n][sec_n] = extract_text_recursive(sec_div).strip()
    else:
        for b_div in textparts:
            b_n = b_div.get('n') or '1'
            data[b_n] = OrderedDict()
            chapters = b_div.findall('tei:div[@subtype="chapter"]', NS) or b_div.findall('div[@subtype="chapter"]') or b_div.findall('tei:div', NS) or b_div.findall('div')
            for ch_div in chapters:
                ch_n = ch_div.get('n')
                if not ch_n: continue
                data[b_n][ch_n] = OrderedDict()
                sections = ch_div.findall('tei:div[@subtype="section"]', NS) or ch_div.findall('div[@subtype="section"]') or ch_div.findall('tei:p', NS) or ch_div.findall('p')
                for idx, sec_div in enumerate(sections):
                    sec_n = sec_div.get('n') or str(idx + 1)
                    paragraphs = sec_div.findall('tei:p', NS) or sec_div.findall('p')
                    if paragraphs:
                        data[b_n][ch_n][sec_n] = ' '.join(extract_text_recursive(p).strip() for p in paragraphs)
                    else:
                        data[b_n][ch_n][sec_n] = extract_text_recursive(sec_div).strip()
                        
    clean_data = OrderedDict()
    for bk, chaps in data.items():
        if chaps: clean_data[bk] = chaps
    return clean_data if clean_data else None

print("Ingesting registered textual configurations...")
MASTER_CORPUS = OrderedDict()
for v_id, config in TEXT_REGISTRY.items():
    parsed_data = parse_tei_hierarchy(config["path"])
    if parsed_data:
        MASTER_CORPUS[v_id] = parsed_data
        print(f" -> Loaded successfully: {config['label']}")

BASELINE_KEY = "grc_kassel"
baseline_structure = MASTER_CORPUS[BASELINE_KEY]

# ── Extensible Matrix Routing Filenames ───────────────────────
def get_matrix_filename(book, chapter, focus, cross1, cross2, section=None, has_no_toc=False):
    suffix = "#notoc" if has_no_toc else ""
    if section:
        return f"f_{focus}_c1_{cross1}_c2_{cross2}_b{book}_ch{chapter}_sec{section}.html{suffix}"
    return f"f_{focus}_c1_{cross1}_c2_{cross2}_b{book}_ch{chapter}.html{suffix}"

chapter_sizes = {}
for bn in baseline_structure:
    for cn in baseline_structure[bn]:
        chapter_sizes[(bn, cn)] = sum(len(txt) for txt in baseline_structure[bn][cn].values())
max_char_len = max(chapter_sizes.values()) if chapter_sizes else 1

def get_tile_width(book, chapter, max_px=30):
    size = chapter_sizes.get((book, chapter), 0)
    return max(4, int((size / max_char_len) * max_px))

# ── Component Layout Template Builders ────────────────────────
def render_sections_block(book_n, chap_n, text_version_id, focus_id, cross1_id, cross2_id, is_isolated=False, isolated_sec_n=None, has_no_toc=False):
    html = []
    version_data = MASTER_CORPUS.get(text_version_id, {})
    sections_dict = version_data.get(book_n, {}).get(chap_n, {})
    css_class = TEXT_REGISTRY[text_version_id]["class"]
    
    for sec_n, text in sections_dict.items():
        if is_isolated and sec_n != isolated_sec_n: continue
        label = sec_n if sec_n not in ('0', 'pr') else 'pr'
        sec_file_link = get_matrix_filename(book_n, chap_n, focus_id, cross1_id, cross2_id, sec_n, has_no_toc)
        
        html.append(
            f'<div class="section-block {css_class}">'
            f'<span class="sec-num"><a href="{sec_file_link}">{label}</a></span>'
            f'<p>{text if text.strip() else "<i>[Text part omitted in this edition]</i>"}</p>'
            f'</div>'
        )
    return '\n'.join(html)

def render_three_column_configurator(book_n, chap_n, active_focus, active_cross1, active_cross2, current_sec=None, has_no_toc=False):
    f_links, c1_links, c2_links = [], [], []
    for v_id, config in TEXT_REGISTRY.items():
        f_cls = ' class="active"' if v_id == active_focus else ''
        f_links.append(f'<a href="{get_matrix_filename(book_n, chap_n, v_id, active_cross1, active_cross2, current_sec, has_no_toc)}"{f_cls}>{config["label"]}</a>')
        
        c1_cls = ' class="active"' if v_id == active_cross1 else ''
        c1_links.append(f'<a href="{get_matrix_filename(book_n, chap_n, active_focus, v_id, active_cross2, current_sec, has_no_toc)}"{c1_cls}>{config["label"]}</a>')
        
        c2_cls = ' class="active"' if v_id == active_cross2 else ''
        c2_links.append(f'<a href="{get_matrix_filename(book_n, chap_n, active_focus, active_cross1, v_id, current_sec, has_no_toc)}"{c2_cls}>{config["label"]}</a>')
        
    return f"""<div id="layout-configurator">
        <div class="config-group"><strong>Focal Column:</strong> <div class="config-group-items">{' '.join(f_links)}</div></div>
        <div class="config-group"><strong>Comparison 1:</strong> <div class="config-group-items">{' '.join(c1_links)}</div></div>
        <div class="config-group"><strong>Comparison 2:</strong> <div class="config-group-items">{' '.join(c2_links)}</div></div>
    </div>"""

def render_top_navigation_matrix(book_n, chap_n, active_focus, active_cross1, active_cross2, routing_tier='chapter', has_no_toc=False):
    book_links = []
    for bn in baseline_structure:
        first_ch = list(baseline_structure[bn].keys())[0]
        cls = ' class="current"' if bn == book_n else ''
        dest = get_matrix_filename(bn, first_ch, active_focus, active_cross1, active_cross2, None, has_no_toc) if routing_tier == 'chapter' else get_matrix_filename(bn, first_ch, active_focus, active_cross1, active_cross2, list(baseline_structure[bn][first_ch].keys())[0], has_no_toc)
        book_links.append(f'<a href="{dest}"{cls} style="font-weight:bold; padding:2px 6px; margin-right:4px;">Book {bn}</a>')
    
    chap_links = []
    for cn in baseline_structure[book_n]:
        cls = ' current' if cn == chap_n else ''
        dest = get_matrix_filename(book_n, cn, active_focus, active_cross1, active_cross2, None, has_no_toc) if routing_tier == 'chapter' else get_matrix_filename(book_n, cn, active_focus, active_cross1, active_cross2, list(baseline_structure[book_n][cn].keys())[0], has_no_toc)
        pixel_width = get_tile_width(book_n, cn)
        
        chap_links.append(
            f'<div class="nav-tile-wrapper">'
            f'  <a href="{dest}" class="nav-tile-link{cls}">{cn}</a>'
            f'  <div class="size-bar" style="width:{pixel_width}px;"></div>'
            f'</div>'
        )
        
    return f"""<div id="browse-bar">
    <div class="browse-row"><label>Books:</label><div class="browse-items">{' '.join(book_links)}</div></div>
    <div class="browse-row" style="margin-top:8px;"><label>Chapters:</label><div class="browse-items" style="gap:5px 3px;">{' '.join(chap_links)}</div></div>
</div>"""

def render_sidebar_toc(current_book, current_chap, current_sec_n, active_focus, active_cross1, active_cross2, routing_tier='chapter', has_no_toc=False):
    html = ["<div id=\"sidebar-toc\">", "<h3>Table of Contents</h3>", "<ul>"]
    for bn in baseline_structure:
        if bn == current_book:
            html.append(f'<li class="active-book">Poetics Hierarchy <ul class="toc-chapters">')
            for cn in baseline_structure[bn]:
                is_active_ch = (cn == current_chap)
                cls = ' class="current"' if is_active_ch else ''
                dest = get_matrix_filename(bn, cn, active_focus, active_cross1, active_cross2, None, has_no_toc) if routing_tier == 'chapter' else get_matrix_filename(bn, cn, active_focus, active_cross1, active_cross2, list(baseline_structure[bn][cn].keys())[0], has_no_toc)
                html.append(f'<li><a href="{dest}"{cls}>Chapter {cn}</a>')
                
                if is_active_ch:
                    html.append('<ul class="toc-sections">')
                    for sn in baseline_structure[bn][cn]:
                        s_cls = ' class="current"' if (routing_tier == 'section' and sn == current_sec_n) else ''
                        s_dest = get_matrix_filename(bn, cn, active_focus, active_cross1, active_cross2, sn, has_no_toc)
                        html.append(f'<li><a href="{s_dest}"{s_cls}>{sn}</a></li>')
                    html.append('</ul>')
                html.append('</li>')
            html.append('</ul></li>')
        else:
            dest = get_matrix_filename(bn, list(baseline_structure[bn].keys())[0], active_focus, active_cross1, active_cross2, None, has_no_toc)
            html.append(f'<li><a href="{dest}">Book {bn}</a></li>')
    html.append("</ul></div>")
    return "\n".join(html)

print("Compiling global alignment matrix layout variants...")
start_time = time.time()
total_compiled = 0

for focus_id in TEXT_REGISTRY:
    for cross1_id in TEXT_REGISTRY:
        for cross2_id in TEXT_REGISTRY:
            for has_no_toc in (False, True): # Pre-build persistent state maps for open and collapsed views
                
                # ==========================================
                # PHASE 1: GENERATE CHAPTER-LEVEL SEPARATE PAGES
                # ==========================================
                for ch_idx, (book_n, chap_n) in enumerate(chapter_list):
                    prev_html, next_html = '&#x25C0;', '&#x25B6;'
                    if ch_idx > 0:
                        pb, pc = chapter_list[ch_idx - 1]
                        prev_html = f'<a href="{get_matrix_filename(pb, pc, focus_id, cross1_id, cross2_id, None, has_no_toc)}">&#x25C0; Prev Chapter</a>'
                    if ch_idx < len(chapter_list) - 1:
                        nb, nc = chapter_list[ch_idx + 1]
                        next_html = f'<a href="{get_matrix_filename(nb, nc, focus_id, cross1_id, cross2_id, None, has_no_toc)}">Next Chapter &#x25B6;</a>'
                        
                    nav_html = f'<div class="nav-arrows">{prev_html} <span style="margin:0 auto;color:#666;">Chapter View</span> {next_html}</div>'

                    focus_html = render_sections_block(book_n, chap_n, focus_id, focus_id, cross1_id, cross2_id, False, None, has_no_toc)
                    cross1_html = render_sections_block(book_n, chap_n, cross1_id, focus_id, cross1_id, cross2_id, False, None, has_no_toc)
                    cross2_html = render_sections_block(book_n, chap_n, cross2_id, focus_id, cross1_id, cross2_id, False, None, has_no_toc)

                    config_panel = render_three_column_configurator(book_n, chap_n, focus_id, cross1_id, cross2_id, None, has_no_toc)
                    browse_bar = render_top_navigation_matrix(book_n, chap_n, focus_id, cross1_id, cross2_id, 'chapter', has_no_toc)
                    sidebar_toc = render_sidebar_toc(book_n, chap_n, None, focus_id, cross1_id, cross2_id, 'chapter', has_no_toc)
                    
                    # Determine target redirection flags for persistence button link mapping
                    toggle_dest = get_matrix_filename(book_n, chap_n, focus_id, cross1_id, cross2_id, None, not has_no_toc)

                    page_content = f"""<!DOCTYPE html>
<html>
<head><meta charset="UTF-8"><title>Aristotle Poetics Ch. {chap_n}</title><style>{PERSEUS_CSS}</style></head>
<body>
<div id="app-view-root">
  <div id="outer-wrapper" id="{'notoc' if has_no_toc else 'open_toc'}">
    {sidebar_toc}
    <div id="main-container">
      <div id="focus-text" class="reading-column">
        {nav_html}
        <h2 style="font-size:14px;color:#660000;margin-bottom:10px;">Aristotle, Poetics: Chapter {chap_n}</h2>
        {focus_html}
        {nav_html}
      </div>
      <div id="cross-ref-1" class="cross-panel reading-column"><div class="panel-header">{TEXT_REGISTRY[cross1_id]["label"]}</div>{cross1_html}</div>
      <div id="cross-ref-2" class="cross-panel reading-column"><div class="panel-header">{TEXT_REGISTRY[cross2_id]["label"]}</div>{cross2_html}</div>
    </div>
  </div>
  <div id="header-container">
    <div id="perseus-banner"><h1><a href="{get_matrix_filename('1', '1', 'grc_kassel', 'eng_fyfe', 'eng_butcher', None, has_no_toc)}">Perseus Digital Library</a></h1><span class="doc-title">Aristotle, Poetics Workspace</span></div>
    <div id="nav-bar">
      <a href="{toggle_dest}" class="toc-toggle-btn"><span class="show-toc-label">Show Sidebar TOC</span><span class="hide-toc-label">Hide Sidebar TOC</span></a>
      <a href="{get_matrix_filename('1', '1', 'grc_kassel', 'eng_fyfe', 'eng_butcher', None, has_no_toc)}">Home Workspace</a>
    </div>
    {browse_bar}
    {config_panel}
  </div>
</div>
</body>
</html>"""
                    fname = "f_" + get_matrix_filename(book_n, chap_n, focus_id, cross1_id, cross2_id, None, has_no_toc).replace("#notoc", "")
                    (OUT_DIR / fname).write_text(page_content, encoding='utf-8')
                    total_compiled += 1

                # ==========================================
                # PHASE 2: GENERATE TRIPLE SECTION-LEVEL PAGES
                # ==========================================
                for sec_idx, (book_n, chap_n, sec_n) in enumerate(section_list):
                    prev_html, next_html = '&#x25C0;', '&#x25B6;'
                    if sec_idx > 0:
                        pb, pc, ps = section_list[sec_idx - 1]
                        prev_html = f'<a href="{get_matrix_filename(pb, pc, focus_id, cross1_id, cross2_id, ps, has_no_toc)}">&#x25C0; Prev Section</a>'
                    if sec_idx < len(section_list) - 1:
                        nb, nc, ns = section_list[sec_idx + 1]
                        next_html = f'<a href="{get_matrix_filename(nb, nc, focus_id, cross1_id, cross2_id, ns, has_no_toc)}">Next Section &#x25B6;</a>'
                        
                    nav_html = f'<div class="nav-arrows">{prev_html} <span style="margin:0 auto;color:#666;">Section View</span> {next_html}</div>'

                    focus_html = render_sections_block(book_n, chap_n, focus_id, focus_id, cross1_id, cross2_id, True, sec_n, has_no_toc)
                    cross1_html = render_sections_block(book_n, chap_n, cross1_id, focus_id, cross1_id, cross2_id, True, sec_n, has_no_toc)
                    cross2_html = render_sections_block(book_n, chap_n, cross2_id, focus_id, cross1_id, cross2_id, True, sec_n, has_no_toc)

                    config_panel = render_three_column_configurator(book_n, chap_n, focus_id, cross1_id, cross2_id, sec_n, has_no_toc)
                    browse_bar = render_top_navigation_matrix(book_n, chap_n, focus_id, cross1_id, cross2_id, 'section', has_no_toc)
                    sidebar_toc = render_sidebar_toc(book_n, chap_n, sec_n, focus_id, cross1_id, cross2_id, 'section', has_no_toc)
                    
                    toggle_dest = get_matrix_filename(book_n, chap_n, focus_id, cross1_id, cross2_id, sec_n, not has_no_toc)

                    page_content = f"""<!DOCTYPE html>
<html>
<head><meta charset="UTF-8"><title>Aristotle Poetics Sec. {chap_n}.{sec_n}</title><style>{PERSEUS_CSS}</style></head>
<body>
<div id="app-view-root">
  <div id="outer-wrapper" id="{"notoc" if has_no_toc else "open_toc"}">
    {sidebar_toc}
    <div id="main-container">
      <div id="focus-text" class="reading-column">
        {nav_html}
        <h2 style="font-size:14px;color:#660000;margin-bottom:10px;">Aristotle, Poetics: Chapter {chap_n}, Section {sec_n}</h2>
        {focus_html}
        {nav_html}
      </div>
      <div id="cross-ref-1" class="cross-panel reading-column"><div class="panel-header">{TEXT_REGISTRY[cross1_id]["label"]}</div>{cross1_html}</div>
      <div id="cross-ref-2" class="cross-panel reading-column"><div class="panel-header">{TEXT_REGISTRY[cross2_id]["label"]}</div>{cross2_html}</div>
    </div>
  </div>
  <div id="header-container">
    <div id="perseus-banner"><h1><a href="{get_matrix_filename('1', '1', 'grc_kassel', 'eng_fyfe', 'eng_butcher', None, has_no_toc)}">Perseus Digital Library</a></h1><span class="doc-title">Aristotle, Poetics Workspace</span></div>
    <div id="nav-bar">
      <a href="{toggle_dest}" class="toc-toggle-btn"><span class="show-toc-label">Show Sidebar TOC</span><span class="hide-toc-label">Hide Sidebar TOC</span></a>
      <a href="{get_matrix_filename('1', '1', 'grc_kassel', 'eng_fyfe', 'eng_butcher', None, has_no_toc)}">Home Workspace</a> | <a href="{get_matrix_filename(book_n, chap_n, focus_id, cross1_id, cross2_id, None, has_no_toc)}" style="color:#660000;">&uarr; Up to Full Chapter View</a>
    </div>
    {browse_bar}
    {config_panel}
  </div>
</div>
</body>
</html>"""
                    fname = "f_" + get_matrix_filename(book_n, chap_n, focus_id, cross1_id, cross2_id, sec_n, has_no_toc).replace("#notoc", "")
                    (OUT_DIR / fname).write_text(page_content, encoding='utf-8')
                    total_compiled += 1

print(f"Compilation success. Generated {total_compiled} standalone cross-reference workspace views with native layout state preservation.")
```
