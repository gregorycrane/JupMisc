```python
import os
import re
import time
from pathlib import Path
from collections import OrderedDict
import xml.etree.ElementTree as ET

# ── Local Configuration ────────────────────────────────────────
OUT_DIR = Path("/Users/gcrane/Downloads/gemsite/classical_workspace")
OUT_DIR.mkdir(parents=True, exist_ok=True)

NS = {'tei': 'http://www.tei-c.org/ns/1.0'}

TEXT_REGISTRY = {
    "grc_jones": {
        "path": "/Users/gcrane/github/canonical-greekLit/data/tlg0003/tlg001/tlg0003.tlg001.perseus-grc2.xml",
        "label": "Greek (H. S. Jones, 1942)",
        "class": "greek-text"
    },
    "eng_smith": {
        "path": "/Users/gcrane/github/canonical-greekLit/data/tlg0003/tlg001/tlg0003.tlg001.1st1K-eng1.xml",
        "label": "English (C. F. Smith, 1919)",
        "class": "english-text"
    },
    "eng_crawley": {
        "path": "/Users/gcrane/github/canonical-greekLit/data/tlg0003/tlg001/tlg0003.tlg001.perseus-eng6.xml",
        "label": "English (R. Crawley, 1914)",
        "class": "english-text"
    },
    "fre_betant": {
        "path": "/Users/gcrane/github/canonical-greekLit/data/tlg0003/tlg001/tlg0003.tlg001.1st1k-fre1.xml",
        "label": "French (E. Bétant, 1863 )",
        "class": "french-text"
    },
    "lat_haase": {
        "path": "/Users/gcrane/github/canonical-greekLit/data/tlg0003/tlg001/tlg0003.tlg001.1st1k-lat2.xml",
        "label": "Latin (Fr. Haase, 1869)",
        "class": "latin-text"
    }
}

# ── Layout CSS Matrix Framework ──────────────────────
PERSEUS_CSS = r"""
* { margin: 0; padding: 0; box-sizing: border-box; }
html, body {
    height: 100%; width: 100%; overflow: hidden; 
    font-family: "Palatino Linotype", "Book Antiqua", Palatino, Georgia, serif;
    font-size: 13px; color: #000; background: #fff;
}
a { color: #336699; text-decoration: none; }
a:hover { text-decoration: underline; }

#app-view-root { display: flex; flex-direction: column; height: 100vh; width: 100vw; overflow: hidden; }

#header-container { flex-shrink: 0; background: #fff; border-bottom: 1px solid #ccc; z-index: 10; }
#perseus-banner { background: #660000; color: #fff; padding: 6px 15px; display: flex; justify-content: space-between; align-items: center; }
#perseus-banner h1 a { color: #fff; font-size: 16px; font-weight: bold; }
#perseus-banner .doc-title { font-size: 11px; color: #ffccaa; font-weight: bold; }

#nav-bar { background: #ddddcc; border-bottom: 1px solid #999988; padding: 4px 15px; font-size: 11px; font-weight: bold; }

#browse-bar { background: #eeeeee; border-bottom: 1px solid #cccccc; padding: 6px 15px; font-size: 11px; display: flex; flex-direction: column; gap: 5px; }
.browse-row { display: flex; align-items: flex-start; }
.browse-row label { font-weight: bold; width: 80px; color: #444; flex-shrink: 0; padding-top: 1px; }
.browse-items { display: flex; flex-wrap: wrap; gap: 4px; }
.browse-items a { padding: 1px 6px; background: #e0e0d0; color: #336699; border: 1px solid #bbbb99; font-weight: bold; border-radius: 2px; }
.browse-items a.current { background: #660000; color: #fff; border-color: #330000; }

.browse-items a.sec-pill { background: #fff; color: #555; border: 1px solid #ccc; font-weight: normal; font-size: 10px; padding: 1px 5px; }
.browse-items a.sec-pill:hover { background: #660000; color: #fff; border-color: #330000; text-decoration: none; }

#outer-wrapper { display: flex; width: 100%; flex: 1; min-height: 0; }

#main-container { flex: 1; display: flex; min-width: 0; height: 100%; }
.reading-column { display: flex; flex-direction: column; flex: 1; min-width: 0; height: 100%; border-right: 1px solid #eeeeee; background: #fff; }
.reading-column:last-child { border-right: none; }
.cross-panel { background: #fafafa; }

.panel-header { background: #660000; color: #fff; padding: 4px 8px; font-size: 11px; font-weight: bold; position: sticky; top: 0; z-index: 5; display: flex; justify-content: space-between; align-items: center; flex-shrink: 0; }
.edition-select-dropdown { background: #fff; color: #333; font-size: 10px; padding: 1px 4px; border: 1px solid #ccc; border-radius: 2px; outline: none; cursor: pointer; }

.column-content-scroll { flex: 1; overflow-y: auto; padding: 15px; }

input[type="radio"] { display: none; }
.version-block { display: none; }

.section-row { border-bottom: 1px solid #f0f0f0; padding: 6px 0; display: block; scroll-margin-top: 5px; }
.section-row .sec-num { font-weight: bold; color: #990000; margin-right: 8px; display: inline-block; width: 35px; }
.greek-text { font-size: 15px; line-height: 1.7; font-family: "Gentium Plus", "Athena", serif; color: #000; }
.english-text { font-size: 13px; line-height: 1.6; color: #222; }
.note { font-size: 11px; color: #666; font-style: italic; background: #f4f4f0; padding: 0 2px; }

/* Workspace Pagination Footer Action Items */
.viewport-footer-controls { display: flex; justify-content: space-between; align-items: center; gap: 8px; margin-top: 20px; padding-top: 10px; border-top: 1px solid #ddd; }
.footer-group-left { display: flex; gap: 6px; }
.action-btn { display: inline-block; background: #660000; color: #fff !important; padding: 4px 10px; font-size: 11px; font-weight: bold; border-radius: 2px; border: none; cursor: pointer; }
.action-btn.secondary { background: #666; }
.action-btn:hover { background: #330000; text-decoration: none; }
.action-btn.secondary:hover { background: #444; }
"""

# ── Intelligent JavaScript Router with Dynamic Pagination ───────
STICKY_ROUTER_JS = r"""
<script>
document.addEventListener("DOMContentLoaded", function() {
    const prefixes = ["f", "c1", "c2"];
    
    // Global pagination variables injected dynamically during template build
    const baselineSecs = window.CHAPTER_SECTIONS || [];
    const prevChapterUrl = window.PREV_CHAPTER_URL || "";
    const nextChapterUrl = window.NEXT_CHAPTER_URL || "";

    // 1. Restore previous choices from URL query params or LocalStorage cache
    const params = new URLSearchParams(window.location.search);
    prefixes.forEach(p => {
        let savedVal = params.get(p) || localStorage.getItem("workspace_sel_" + p);
        if (savedVal && document.getElementById(p + "_" + savedVal)) {
            document.getElementById(p + "_" + savedVal).checked = true;
            document.querySelectorAll("." + p + "-select-node").forEach(selectEl => {
                selectEl.value = savedVal;
            });
        }
    });

    window.syncWorkspaceChoice = function(prefix, value) {
        document.getElementById(prefix + "_" + value).click();
        localStorage.setItem("workspace_sel_" + prefix, value);
        updateNavigationLinks();
    };

    // 2. Main engine to dynamically calculate next/prev buttons based on active target hash
    function updateNavigationLinks() {
        const currentParams = new URLSearchParams();
        prefixes.forEach(p => {
            const checkedRadio = document.querySelector(`input[name="${p}_col"]:checked`);
            if (checkedRadio) {
                currentParams.set(p, checkedRadio.id.replace(p + "_", ""));
            }
        });

        const currentHash = window.location.hash; // e.g., "#global_sec_2"
        let activeSec = "";
        if (currentHash && currentHash.startsWith("#global_sec_")) {
            activeSec = currentHash.replace("#global_sec_", "");
        }

        let prevUrl = "";
        let prevLabel = "";
        let nextUrl = "";
        let nextLabel = "";

        if (!activeSec) {
            // Full Chapter Mode View Pagination
            if (prevChapterUrl) { prevUrl = prevChapterUrl; prevLabel = "&larr; Previous Chapter"; }
            if (nextChapterUrl) { nextUrl = nextChapterUrl; nextLabel = "Next Chapter &rarr;"; }
        } else {
            // Isolated Parallel Section Mode View Pagination
            const idx = baselineSecs.indexOf(activeSec);
            if (idx !== -1) {
                // Determine Previous Section Button Dest
                if (idx > 0) {
                    prevUrl = "#global_sec_" + baselineSecs[idx - 1];
                    prevLabel = "&larr; Previous Section [" + baselineSecs[idx - 1] + "]";
                } else if (prevChapterUrl) {
                    prevUrl = prevChapterUrl; // Links to full previous chapter file
                    prevLabel = "&larr; Previous Chapter";
                }

                // Determine Next Section Button Dest
                if (idx < baselineSecs.length - 1) {
                    nextUrl = "#global_sec_" + baselineSecs[idx + 1];
                    nextLabel = "Next Section [" + baselineSecs[idx + 1] + "] &rarr;";
                } else if (nextChapterUrl) {
                    nextUrl = nextChapterUrl; // Increments to next chapter file
                    nextLabel = "Next Chapter &rarr;";
                }
            }
        }

        // Apply rewritten parameters across all reading columns
        document.querySelectorAll(".reading-column").forEach(col => {
            const btnPrev = col.querySelector(".p-btn-prev");
            const btnNext = col.querySelector(".p-btn-next");

            if (prevUrl) {
                btnPrev.style.display = "inline-block";
                btnPrev.setAttribute("href", prevUrl);
                btnPrev.innerHTML = prevLabel;
            } else {
                btnPrev.style.display = "none";
            }

            if (nextUrl) {
                btnNext.style.display = "inline-block";
                btnNext.setAttribute("href", nextUrl);
                btnNext.innerHTML = nextLabel;
            } else {
                btnNext.style.display = "none";
            }
        });

        // Append active configuration query state across all browser header links
        document.querySelectorAll("#browse-bar .browse-items a:not(.sec-pill), .p-btn-prev, .p-btn-next").forEach(link => {
            const hrefAttr = link.getAttribute("href");
            if (!hrefAttr || hrefAttr.startsWith("#")) return; // Skip internal anchor rewrites

            const baseUrl = hrefAttr.split("?")[0];
            const hashSplit = baseUrl.split("#");
            const rawUrl = hashSplit[0];
            const anchorHash = hashSplit[1] ? "#" + hashSplit[1] : "";
            link.setAttribute("href", rawUrl + "?" + currentParams.toString() + anchorHash);
        });
    }

    // Monitor internal hash routing steps to handle cross-column clicks instantly
    window.addEventListener("hashchange", updateNavigationLinks);
    updateNavigationLinks();
});
</script>
"""

# ── TEI Hierarchy Parser Engine ────────────────────────────────
def extract_text_recursive(elem):
    parts = []
    if elem.text: parts.append(elem.text)
    for child in elem:
        tag = child.tag.replace('{http://www.tei-c.org/ns/1.0}', '')
        if tag == 'note':
            note_text = extract_text_recursive(child).strip()
            if note_text: parts.append(f'<span class="note">[{note_text}]</span>')
        else:
            parts.append(extract_text_recursive(child))
        if child.tail: parts.append(child.tail)
    return ''.join(parts)

def parse_hierarchical_tei(path):
    if not os.path.exists(path): return None
    tree = ET.parse(path)
    root = tree.getroot()
    body = root.find('.//tei:body', NS) or root.find('.//body')
    if body is None: return None

    data = OrderedDict()

    def walk_divisions(node, current_path):
        tag = node.tag.replace('{http://www.tei-c.org/ns/1.0}', '')
        subtype = node.get('subtype') or node.get('type')
        n_val = node.get('n')

        if tag == 'div' and subtype in ('book', 'chapter', 'section') and n_val:
            new_path = current_path + [(subtype, n_val)]
        else:
            new_path = current_path

        paragraphs = node.findall('tei:p', NS) or node.findall('p')
        if paragraphs and new_path:
            key_map = {t: v for t, v in new_path}
            bk = key_map.get('book', '1')
            ch = key_map.get('chapter', '1')
            sec = key_map.get('section', n_val or '1')

            if bk not in data: data[bk] = OrderedDict()
            if ch not in data[bk]: data[bk][ch] = OrderedDict()
            
            combined_txt = ' '.join(extract_text_recursive(p).strip() for p in paragraphs)
            if combined_txt:
                data[bk][ch][sec] = combined_txt
            return

        for child in node:
            walk_divisions(child, new_path)

    walk_divisions(body, [])
    return data

print("Ingesting system TEI source layers...")
MASTER_CORPUS = OrderedDict()
for v_id, cfg in TEXT_REGISTRY.items():
    parsed = parse_hierarchical_tei(cfg["path"])
    if parsed:
        MASTER_CORPUS[v_id] = parsed

first_version = list(MASTER_CORPUS.keys())[0]
baseline_corpus = MASTER_CORPUS[first_version]
has_explicit_books = len(baseline_corpus.keys()) > 1 or list(baseline_corpus.keys())[0] != '1'

# Generate a global list of chapters to compute file progression bounds
CHAPTER_SEQUENCE = []
for b_k, ch_v in baseline_corpus.items():
    for c_k in ch_v.keys():
        CHAPTER_SEQUENCE.append({'book': b_k, 'chapter': c_k})

def get_page_filename(book_id, ch_id):
    return f"viewer_b{book_id}_ch{ch_id}.html"

# ── Layout Assembly Components ───────────────────────────────
def render_matrix_navigation(active_bk, active_ch, baseline_secs):
    html = ['<div id="browse-bar">']
    
    if has_explicit_books:
        html.append('<div class="browse-row"><label>Books:</label><div class="browse-items">')
        for bk in baseline_corpus.keys():
            cls = ' class="current"' if bk == active_bk else ''
            first_ch = list(baseline_corpus[bk].keys())[0]
            html.append(f'<a href="{get_page_filename(bk, first_ch)}"{cls}>{bk}</a>')
        html.append('</div></div>')

    html.append('<div class="browse-row"><label>Chapters:</label><div class="browse-items">')
    for ch in baseline_corpus[active_bk].keys():
        cls = ' class="current"' if ch == active_ch else ''
        html.append(f'<a href="{get_page_filename(active_bk, ch)}"{cls}>{ch}</a>')
    html.append('</div></div>')

    html.append('<div class="browse-row"><label>Sections:</label><div class="browse-items">')
    for sec in baseline_secs:
        html.append(f'<a href="#global_sec_{sec}" id="pill_sec_{sec}" class="sec-pill">{sec}</a>')
    html.append('</div></div></div>')
    
    return '\n'.join(html)

def render_column_viewport(bk_id, ch_id, col_class, title, radio_prefix, baseline_secs):
    html = [f'<div class="{col_class} reading-column">', f'<div class="panel-header"><span>{title}</span>']
    html.append(f'<select class="edition-select-dropdown {radio_prefix}-select-node" onchange="syncWorkspaceChoice(\'{radio_prefix}\', this.value);">')
    
    for v_id, cfg in TEXT_REGISTRY.items():
        if radio_prefix == "f" and "grc" in v_id: selected = " selected"
        elif radio_prefix == "c1" and "smith" in v_id: selected = " selected"
        elif radio_prefix == "c2" and "crawley" in v_id: selected = " selected"
        else: selected = ""
        html.append(f'<option value="{v_id}"{selected}>{cfg["label"]}</option>')
    html.append('</select></div>')

    html.append('<div class="column-content-scroll">')
    for v_id in TEXT_REGISTRY:
        html.append(f'<div class="version-block v-{v_id}">')
        ch_data = MASTER_CORPUS.get(v_id, {}).get(bk_id, {}).get(ch_id, {})
        for sec in baseline_secs:
            text = ch_data.get(sec, "<i>[Text block division missing in translation]</i>")
            html.append(
                f'<div class="section-row s-idx-{sec}">'
                f'  <span class="sec-num"><a href="#global_sec_{sec}">[{sec}]</a></span>'
                f'  <span class="text-p {TEXT_REGISTRY[v_id]["class"]}">{text}</span>'
                f'</div>'
            )
        html.append('</div>')

    # Navigation Footer Controls 
    html.append('<div class="viewport-footer-controls">')
    html.append('  <div class="footer-group-left">')
    html.append('    <a href="" class="action-btn p-btn-prev" style="display:none;"></a>')
    html.append('    <a href="#" class="action-btn secondary">Full Chapter</a>')
    html.append('  </div>')
    html.append('  <a href="" class="action-btn p-btn-next" style="display:none;"></a>')
    html.append('</div></div></div>')
    return '\n'.join(html)

# ── Processing Execution Loop ───────────────────────────────────
print("Compiling responsive interactive workspace...")
start_time = time.time()

dynamic_css = []
for v_id in TEXT_REGISTRY:
    dynamic_css.append(f"#f_{v_id}:checked ~ #app-view-root #main-container .col1-container .v-{v_id},")
    dynamic_css.append(f"#c1_{v_id}:checked ~ #app-view-root #main-container .col2-container .v-{v_id},")
    dynamic_css.append(f"#c2_{v_id}:checked ~ #app-view-root #main-container .col3-container .v-{v_id},")

base_css = PERSEUS_CSS + "\n" + "\n".join(dynamic_css).rstrip(",") + " { display: block; }\n"
base_css += "body:has(.target-marker:target) .section-row { display: none; }\n"

for c_idx, coord in enumerate(CHAPTER_SEQUENCE):
    bk_id = coord['book']
    ch_id = coord['chapter']
    baseline_secs = list(baseline_corpus[bk_id][ch_id].keys())
    
    local_css = base_css
    for sec in baseline_secs:
        local_css += f"#global_sec_{sec}:target ~ #app-view-root .s-idx-{sec} {{ display: block !important; }}\n"
        local_css += f"body:has(#global_sec_{sec}:target) #pill_sec_{sec} {{ background: #660000 !important; color: #fff !important; border-color: #330000 !important; font-weight: bold !important; }}\n"

    form_inputs = []
    for prefix in ("f", "c1", "c2"):
        for idx, v_id in enumerate(TEXT_REGISTRY):
            checked = " checked" if idx == 0 and prefix == "f" else ""
            form_inputs.append(f'<input type="radio" name="{prefix}_col" id="{prefix}_{v_id}"{checked}>')

    marker_tags = [f'<div class="target-marker" id="global_sec_{sec}"></div>' for sec in baseline_secs]
    context_string = f"Book {bk_id}, Chapter {ch_id}" if has_explicit_books else f"Chapter {ch_id}"

    # Calculate surrounding file limits for global boundary jumps
    prev_chapter_url = ""
    if c_idx > 0:
        p_c = CHAPTER_SEQUENCE[c_idx - 1]
        prev_chapter_url = get_page_filename(p_c['book'], p_c['chapter'])

    next_chapter_url = ""
    if c_idx < len(CHAPTER_SEQUENCE) - 1:
        n_c = CHAPTER_SEQUENCE[c_idx + 1]
        next_chapter_url = get_page_filename(n_c['book'], n_c['chapter'])

    master_html = f"""<!DOCTYPE html>
<html>
<head>
  <meta charset="UTF-8">
  <title>Parallel Workspace - {context_string}</title>
  <style>{local_css}</style>
  <script>
    // Inject compile-time sequence variables for the JavaScript routing engine
    window.CHAPTER_SECTIONS = {list(baseline_secs)};
    window.PREV_CHAPTER_URL = "{prev_chapter_url}";
    window.NEXT_CHAPTER_URL = "{next_chapter_url}";
  </script>
</head>
<body>
  {" ".join(form_inputs)}
  {" ".join(marker_tags)}

  <div id="app-view-root">
    <div id="header-container">
      <div id="perseus-banner">
        <h1><a href="#">Perseus Workspace Engine</a></h1>
        <span class="doc-title">Parallel Reading Environment Matrix</span>
      </div>
      <div id="nav-bar">
        <span>Active Frame Context: {context_string}</span>
      </div>
      {render_matrix_navigation(bk_id, ch_id, baseline_secs)}
    </div>

    <div id="outer-wrapper">
      <div id="main-container">
        {render_column_viewport(bk_id, ch_id, "col1-container", "Focal Axis Text Column", "f", baseline_secs)}
        {render_column_viewport(bk_id, ch_id, "col2-container cross-panel", "Comparison Version 1", "c1", baseline_secs)}
        {render_column_viewport(bk_id, ch_id, "col3-container cross-panel", "Comparison Version 2", "c2", baseline_secs)}
      </div>
    </div>
  </div>
  {STICKY_ROUTER_JS}
</body>
</html>"""

    (OUT_DIR / get_page_filename(bk_id, ch_id)).write_text(master_html, encoding='utf-8')

print(f"[SUCCESS] Multi-tier workspace modules generated inside {time.time() - start_time:.2f} seconds.")
```
