```python
import os
from bs4 import BeautifulSoup

def load_texts_from_tei(file_paths):
    """
    Parses a list of TEI XML file paths and returns a list of plain text strings.
    
    Args:
        file_paths (list): List of strings, e.g., ['file1.xml', 'file2.xml']
        
    Returns:
        list: List of strings containing the body text of each document.
    """
    docs = []
    
    for file_path in file_paths:
        if not os.path.exists(file_path):
            print(f"Warning: File not found: {file_path}")
            continue
            
        with open(file_path, 'r', encoding='utf-8') as f:
            # 'xml' parser is required for TEI tags. 
            # If you don't have lxml installed, use 'html.parser' (less accurate for XML)
            soup = BeautifulSoup(f, 'xml') 
            
            # 1. Find the <body> tag. 
            # In TEI, content is usually in <text> -> <body>
            body = soup.find('body')
            
            if body:
                # 2. Optional: Remove <note> or <figure> tags if you don't want 
                # footnotes/captions to skew your vocabulary counts.
                for tag in body.find_all(['note', 'figure', 'bibl']):
                    tag.decompose()
                
                # 3. Extract text, separating tags with a space to prevent
                # words merging (e.g., "end</line><line>Start" -> "end Start")
                text = body.get_text(separator=' ', strip=True)
                docs.append(text)
            else:
                # Fallback if no body tag is found
                print(f"Warning: No <body> tag found in {file_path}")
                docs.append("")
                
    return docs

# --- USAGE IN YOUR NOTEBOOK ---
file_list = ['/Users/gcrane/github/GRC_misc/homer.iliad.butler-aligned-eng1.xml', '/Users/gcrane/github/GRC_misc/heike.sadler1918.xml']
docs = load_texts_from_tei(file_list)
# ... continue with the rest of the notebook code
```

```python
import pandas as pd
import numpy as np
import os
from bs4 import BeautifulSoup
from sklearn.feature_extraction.text import CountVectorizer

# --- CONFIGURATION ---
files = [
    #'/Users/gcrane/github/GRC_misc/homer.iliad.butler-aligned-eng1.xml', 
    '/Users/gcrane/github/GRC_misc/heike.tokyo1933-gemini3-eng1.xml',
    '/Users/gcrane/github/GRC_misc/heike.sadler1918.xml'
]
names = ['Heike-Gemini', 'Heike']

# TOGGLE THIS: 
# True  = Ignore words starting with capitals (Filters Proper Names & Sentence Starters)
# False = Include all words and normalize to lowercase (Standard analysis)
EXCLUDE_UPPERCASE = True 

def load_texts_from_tei(file_paths):
    """
    Parses a list of TEI XML file paths and returns a list of plain text strings.
    """
    docs = []
    
    for file_path in file_paths:
        if not os.path.exists(file_path):
            print(f"Warning: File not found: {file_path}")
            # Append empty string to maintain index alignment
            docs.append("")
            continue
            
        with open(file_path, 'r', encoding='utf-8') as f:
            # 'xml' parser is required for TEI tags. 
            soup = BeautifulSoup(f, 'xml') 
            
            # 1. Find the <body> tag. 
            body = soup.find('body')
            
            if body:
                # 2. Remove notes/figures to avoid polluting counts
                for tag in body.find_all(['note', 'figure', 'bibl']):
                    tag.decompose()
                
                # 3. Extract text
                text = body.get_text(separator=' ', strip=True)
                docs.append(text)
            else:
                print(f"Warning: No <body> tag found in {file_path}")
                docs.append("")
                
    return docs

def build_word_dataframe(docs, doc_names, exclude_uppercase=False):
    """
    Takes a list of document strings and returns a DataFrame 
    where index=words and columns=documents.
    """
    # 1. Configure Vectorizer based on toggle
    if exclude_uppercase:
        print("Mode: Excluding capitalized words (Proper Names/Sentence Starters)...")
        # lowercase=False: Preserves case to check for capitals
        # pattern matches ONLY words starting with lowercase [a-z]
        vec = CountVectorizer(
            lowercase=False,
            token_pattern=r'\b[a-z]\w*\b',
            stop_words=None
        )
    else:
        print("Mode: Including all words (Normalized to lowercase)...")
        # lowercase=True: Standardizes "The" -> "the"
        # pattern matches ALL words
        vec = CountVectorizer(
            lowercase=True,
            token_pattern=r'(?u)\b\w+\b',
            stop_words=None
        )
    
    # 2. Fit and Transform
    X = vec.fit_transform(docs)
    
    # 3. Create Initial DataFrame (Rows=Docs, Cols=Words)
    df_counts = pd.DataFrame(
        X.toarray(), 
        columns=vec.get_feature_names_out(), 
        index=doc_names
    )
    
    # 4. Transpose (Rows=Words, Cols=Docs)
    df_words = df_counts.T
    df_words.columns = doc_names
    
    return df_words

def calculate_log_odds(df, col_a, col_b):
    """
    Adds a 'Log_Odds' column to the DataFrame.
    Positive score = Characteristic of col_a
    Negative score = Characteristic of col_b
    """
    # Add 1 to avoid division by zero (smoothing)
    count_a = df[col_a] + 1
    count_b = df[col_b] + 1
    
    # Calculate total size of each corpus
    total_a = count_a.sum()
    total_b = count_b.sum()
    
    # Log-Odds Ratio Formula
    df['Log_Odds'] = np.log(count_a / total_a) - np.log(count_b / total_b)
    
    return df.sort_values('Log_Odds', ascending=False)

def lookup_word(df, word):
    """
    Looks up a specific word in the DataFrame and prints its statistics.
    """
    # Ensure word is looked up in the correct format (lowercase if mode is False)
    if not EXCLUDE_UPPERCASE:
        word = word.lower()
        
    if word in df.index:
        print(f"\n--- Statistics for '{word}' ---")
        print(df.loc[word])
    else:
        print(f"\nWord '{word}' not found in vocabulary (might be filtered out or not present).")

# --- MAIN EXECUTION BLOCK ---
if __name__ == "__main__":
    # 1. Load the raw text
    print("Loading texts...")
    docs = load_texts_from_tei(files)
    
    # Check if files loaded successfully
    if not any(docs):
        print("No text loaded. Check file paths.")
    else:
        # 2. Build the DataFrame
        print("Building frequency table...")
        # Pass the global configuration variable here
        df_words = build_word_dataframe(docs, names, exclude_uppercase=False)
        
        # 3. Calculate Keyness (Log Odds)
        print("Calculating Log-Odds...")
        # Use the specific column names defined in 'names' variable
        df_final = calculate_log_odds(df_words, names[0], names[1])
        
        # 4. Show results
        label = "Lowercase Only" if EXCLUDE_UPPERCASE else "All Words"
        print(f"\n--- Top words for {names[0]} ({label}) ---")
        print(df_final.head(50))
        
        print(f"\n--- Top words for {names[1]} ({label}) ---")
        print(df_final.tail(50))

        # 5. Manual Lookup Example
        lookup_word(df_final, "women")
```

```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
from tei_loader import load_texts_from_tei

def build_word_dataframe(docs, doc_names):
    """
    Takes a list of document strings and returns a DataFrame 
    where index=words and columns=documents.
    """
    # 1. Initialize Vectorizer
    # token_pattern=r'(?u)\b\w+\b' keeps single letter words (like 'a' or 'I')
    # stop_words=None keeps common words (important for stylistic analysis)
    vec = CountVectorizer(token_pattern=r'(?u)\b\w+\b', stop_words=None)
    
    # 2. Fit and Transform
    # This counts every word in every document
    X = vec.fit_transform(docs)
    
    # 3. Create Initial DataFrame (Rows=Docs, Cols=Words)
    df_counts = pd.DataFrame(
        X.toarray(), 
        columns=vec.get_feature_names_out(), 
        index=doc_names
    )
    
    # 4. Transpose (Rows=Words, Cols=Docs)
    # We do this because it's easier to filter lists of words this way
    df_words = df_counts.T
    
    # Rename columns for clarity if needed, though index=doc_names usually handles it
    df_words.columns = doc_names
    
    return df_words

def calculate_log_odds(df, col_a, col_b):
    """
    Adds a 'Log_Odds' column to the DataFrame.
    Positive score = Characteristic of col_a
    Negative score = Characteristic of col_b
    """
    # Add 1 to avoid division by zero (smoothing)
    count_a = df[col_a] + 1
    count_b = df[col_b] + 1
    
    # Calculate total size of each corpus
    total_a = count_a.sum()
    total_b = count_b.sum()
    
    # Log-Odds Ratio Formula
    # log(frequency in A) - log(frequency in B)
    df['Log_Odds'] = np.log(count_a / total_a) - np.log(count_b / total_b)
    
    return df.sort_values('Log_Odds', ascending=False)

# --- MAIN EXECUTION BLOCK ---
if __name__ == "__main__":
    # 1. Define your files
    files = ['data/hamlet.xml', 'data/faustus.xml']
    names = ['Hamlet', 'Dr_Faustus']

    # 2. Load the raw text using your TEI loader
    # (Ensure tei_loader.py is in the same folder)
    print("Loading texts...")
    docs = load_texts_from_tei(files)
    
    # Check if files loaded successfully
    if not any(docs):
        print("No text loaded. Check file paths.")
    else:
        # 3. Build the DataFrame
        print("Building frequency table...")
        df_words = build_word_dataframe(docs, names)
        
        # 4. Calculate Keyness (Log Odds)
        print("Calculating Log-Odds...")
        df_final = calculate_log_odds(df_words, 'Hamlet', 'Dr_Faustus')
        
        # 5. Show results
        print("\n--- Top words for Hamlet ---")
        print(df_final.head(10))
        
        print("\n--- Top words for Dr. Faustus ---")
        print(df_final.tail(10))
```

```python
def calculate_log_odds(row):
    # Smoothing (+1) prevents division by zero errors
    a_count = row['Count_A'] + 1
    b_count = row['Count_B'] + 1
    
    # Total word counts per document (normalized)
    a_total = df_words['Count_A'].sum() + 1
    b_total = df_words['Count_B'].sum() + 1
    
    # Log-Odds Formula
    # log(Freq in A) - log(Freq in B)
    log_odds = np.log(a_count / a_total) - np.log(b_count / b_total)
    
    return log_odds

# Apply the formula
df_words['Log_Odds'] = df_words.apply(calculate_log_odds, axis=1)

# Sort by the score to find the extremes
df_words = df_words.sort_values('Log_Odds', ascending=False)
```

```python
# Select the top 10 distinctive words for A and bottom 10 for B
top_a = df_words.head(10)
top_b = df_words.tail(10).sort_values('Log_Odds', ascending=True) # Sort for visual

# Combine them
plot_data = pd.concat([top_a, top_b])

# Plotting
plt.figure(figsize=(10, 8))
colors = ['#3498db' if x > 0 else '#e74c3c' for x in plot_data['Log_Odds']]

plt.barh(plot_data.index, plot_data['Log_Odds'], color=colors)
plt.xlabel(f'Log-Odds Ratio\n(Left = {doc_names[1]} | Right = {doc_names[0]})')
plt.title('Most Characteristic Vocabulary per Document')
plt.axvline(0, color='black', linewidth=0.8) # Center line
plt.tight_layout()
plt.show()
```
