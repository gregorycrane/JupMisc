```python
import re

reftable = {}
notetable = {}
curpage = ''
divlist = {}

def addkey(curdict,curitem):
    if(curitem in curdict):
        curdict[curitem] = curdict[curitem] + 1
    else:
        curdict[curitem] = 1

        
fname = '/Users/gcrane/github/english_trans-dev/volumes/xenophon_1907/xenophon_1907-sent2.xml'
f = open(fname)


outfname = '/Users/gcrane/github/GRC_misc/xenophon01.watson_1854.xml'
outf = open(outfname,'w')
inindex = 0
for l in f:
    if(re.search('<div type="textpart" subtype="index">',l)):
        inindex = 1
    l = re.sub('\s+$','',l)
    if(inindex):
        print(l,file=outf)
        continue

    m = re.search('n="([^"]+watson[^"]+)"',l)
    if(m):
        curedition = m[1]
        divlist['book'] = '0'
        print('edition',curedition)
    m = re.search('n="([0-9]+)" subtype="(book|chapter|section)"',l)
    if(m):
        curdiv = m[2]
        curn = m[1]
        if(curdiv == 'book'):
            divlist['chapter'] = '0'
            divlist['section'] = '0'
            divlist['book'] = curn

        if(curdiv == 'chapter'):
            if(not int(curn) == int(divlist[curdiv]) + 1):
                print('jumped',curpage,curdiv,divlist[curdiv],curn)
            divlist[curdiv] = curn
            divlist['section'] = '0'
            baseurl = 'xml:base="'+curedition+':' + divlist['book'] + '"'
            l = re.sub(' n="',' '+baseurl+' n="',l)

        if(curdiv == 'section'):
            if(not int(curn) == int(divlist[curdiv]) + 1):
                print('jumped',curpage,curdiv,divlist[curdiv],curn)
            divlist[curdiv] = curn
            baseurl = 'xml:base="'+curedition+':' + divlist['book'] + '.' + divlist['chapter'] + '"'
            l = re.sub(' n="',' '+baseurl+' n="',l)

        
        
    while(re.search('<ref (target|n)="([0-9]+)',l)):
        m = re.search('<ref (target|n)="([0-9]+)',l)
        curkey = 'note-' + curpage + '-' + m[2]
        addkey(reftable,curkey)
        l = re.sub('<ref (target|n)="([0-9]+)','<ref target="note-'+curpage+'-\g<2>',l,1)
    m = re.search('<pb n="([^"]+)"',l)
    if(m):
        curpage = m[1]

    while(re.search('((<s |^[a-zA-Z]).+[a-z\,\.;\?”\!:\)])([0-9])',l)):
        m = re.search('((<s |^[a-zA-Z]).+[a-z\,\.;\?”\!:\)])([0-9])',l)
        if(m):
            curnum = m[1]
            curkey = 'note-' + curpage + '-' + m[3]
            l = re.sub('((<s |^[a-zA-Z]).+[a-z,\.;\?”\!:\)])([0-9])','\g<1><ref target="'+curkey+'"/>"',l,1)
            addkey(reftable,curkey)
            #print(curpage+'-' + m[3])

    m = re.search('(<note type="footnote")\s*>\s*([0-9]+)',l)
    if(m):
        curkey = 'note-' + curpage + '-' + m[2]
        addkey(notetable,curkey)
        l = re.sub('(<note type="footnote")\s*>\s*([0-9]+)','<note type="footnote" n="note-'+curpage+'-\g<2>">',l)
    print(l,file=outf)
f.close()
outf.close()

for foo in notetable:
    if(foo not in reftable):
        print(foo)

for foo in reftable:
    if(foo not in notetable):
        print(foo)
```
