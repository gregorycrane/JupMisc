```python
import re

dirname = '/Users/gcrane/github/canonical-greekLit/data/tlg0548/'
def addentry(curdict,curentry):
    if(curentry in curdict):
        curdict[curentry] = curdict[curentry] + 1
    else:
        curdict[curentry] = 1

newsources = {}

def dofile(fname):
    global newsources
    f = open(fname)
    for l in f:
        l = re.sub('<bibl [^>]+>[^<]+</bibl>',' ' ,l)
        m = re.search('<bibl>([^<,0-9]+)',l)
        if(m):
            addentry(newsources,m[1])
    f.close()
    for foo in sorted(newsources,key=newsources.get,reverse=True):
        print(foo,newsources[foo])

tgroup = 'tlg0548'
twork = 'tlg001'
fname = dirname + twork + '/' + tgroup + '.' + twork + '.perseus-eng2.xml'
dofile(fname)
```

```python
import re
pauscits = 0
herodcits = 0
apollodcits = 0
f = open('../Perseus_Encyclopedia/encyclopedia.xml')
for l in f:
    l,n = re.subn('"Hdt','',l)
    herodcits = n + herodcits
    l,n = re.subn('"Apollod','',l)
    apollodcits = n + apollodcits
    l,n = re.subn('"Paus','',l)
    pauscits = n + pauscits
f.close()
print('herod',herodcits)
print('Apollod',apollodcits)
print('pauscits',pauscits)

```
