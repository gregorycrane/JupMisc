```python
import re

def dofile(tgroup,workn):
    fname = '/Users/gcrane/github/glaux-trees/public/xml/' + tgroup + '-' + workn + '.xml'
    f = open(fname)

    greeksents = {}
    sent2cit = {}
    for l in f:
        m = re.search('subdoc="([^"]+)',l)
        if(m):
            curcit = m[1]
        m = re.search('sentence id="([^"]+)',l)
        if(m):
            cursent = m[1]
            greeksents[cursent] = ''
            #print(cursent,curcit)
            sent2cit[cursent]  = curcit
        m = re.search('form="([^"]+)',l)
        if(m):
            curform = m[1]
            if(greeksents[cursent]):
                greeksents[cursent] = greeksents[cursent] + ' ' + curform
            else:
                greeksents[cursent] = curform
    return(greeksents,sent2cit)


    

    
    f.close()
    
senttab,sent2cit = dofile('0086','034')
outfname = '/Users/gcrane/github/Poetics2.0/grc/' + 'tlg0086.tlg034.bekker-sents.tsv'
outf = open(outfname,'w')
for foo in senttab:
    print(foo,sent2cit[foo],senttab[foo],sep='\t',file=outf)

outf.close()
```
