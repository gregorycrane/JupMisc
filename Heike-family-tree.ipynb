### Helper Function to Build the Graph

We'll create a simple recursive function. This function will:
1.  Take a `graph` object and a `node` (a person dictionary).
2.  Add the person as a node to the graph.
3.  Look for a `children` key.
4.  If children exist, loop through them, add an edge from the parent to the child, and then call itself (recurse) for each child.

```python
def build_tree(graph, node):
    """
    Recursively adds nodes and edges to a Graphviz graph
    from a nested dictionary.
    """
    node_name = node['name']
    
    # Add the node itself. This is idempotent, so it's safe
    # to call even if the node was already added as a child.
    graph.node(node_name)
    
    # If the node has children, add edges and recurse
    if 'children' in node and node['children']:
        for child in node['children']:
            child_name = child['name']
            
            # Add the child node
            graph.node(child_name)
            
            # Add the edge from parent to child
            graph.edge(node_name, child_name)
            
            # Recurse for the child's children
            build_tree(graph, child)
```

### 1. Visualize the Heike Genealogy

```python
# Get the Heike tree data
heike_tree_data = data['genealogies'][0]['tree']

# Create a new directed graph
heike_graph = graphviz.Digraph(name='Heike')
heike_graph.attr(rankdir='TB', label='Genealogy of the Heike', labelloc='t', fontsize='20')

# Build the graph starting from the root node
build_tree(heike_graph, heike_tree_data)

# Render the graph in the notebook
heike_graph
```

Now, let's try visualizing just the first 8 generations of the Heike tree using our new function.

```python
def build_tree_limited(graph, node, current_depth=0, max_depth=5):
    """
    Recursively adds nodes and edges, but stops
    when current_depth equals max_depth.
    """
    node_name = node['name']
    graph.node(node_name)
    
    # Stop recursing if we've reached the max depth
    if current_depth >= max_depth:
        return

    if 'children' in node and node['children']:
        for child in node['children']:
            child_name = child['name']
            graph.node(child_name)
            graph.edge(node_name, child_name)
            
            # Call the function for the child, incrementing the depth
            build_tree_limited(graph, child, current_depth + 1, max_depth)
```

```python
# --- Heike Graph (Limited Depth) ---
heike_tree_data = data['genealogies'][0]['tree']
heike_graph_limited = graphviz.Digraph(name='Heike')

# We can use 'TB' (Top-to-Bottom) again now that the tree is smaller
heike_graph_limited.attr(
    rankdir='TB',
    label='Genealogy of the Heike (First 8 Generations)', 
    labelloc='t', 
    fontsize='20'
)

# Call our new function with max_depth=7 (0-indexed, so 0-7 is 8 levels)
build_tree_limited(heike_graph_limited, heike_tree_data, max_depth=7)
heike_graph_limited
```
