```python
import difflib

def compare_texts(text1, text2):
    # Split into words for granular comparison
    words1 = text1.split()
    words2 = text2.split()

    sm = difflib.SequenceMatcher(None, words1, words2)
    for opcode, i1, i2, j1, j2 in sm.get_opcodes():
        if opcode != 'equal':
            print(f"{opcode.upper()}:")
            print(f"  Text1: {' '.join(words1[i1:i2])}")
            print(f"  Text2: {' '.join(words2[j1:j2])}\n")
            
import requests
import re
from lxml import etree

f = open('../GRC_misc/soph.oc.campbell1879-grc.xml')
sawcits = {}
sawlines = {}
tokenlist = []
newtexts = ''
for l in f:
#start 12 οὔθ᾽ ἡδὺς οὔτ᾽ ἀλγεινὸς ἵκετ᾿ ἐξ ὅτου
#οὔθʼ
    #l = re.sub("[᾿᾽']",'ʼ',l)
    m = re.search('<l n="([^"]+)"[^>]*>(.+)</l>',l)
    if(m):
        curcit = m[1]
        curtext = m[2]
        if(curcit in sawcits):
            print('duplicate',curcit)
        sawcits[curcit] = 1
        curline = re.sub('([\.\;\,—])',' \g<1>',m[2])
        curline = re.sub('·',' ·',curline)
        sawlines[curcit] = m[2]
        newtexts = newtexts + ' ' + m[2]
        for word in curline.split():
            tokenlist.append(word + '\t' + curcit)
f.close()
print('curcit',curcit)
def extract_greek_lines_from_tei(url):
    """
    Fetches a TEI XML file from the given URL and extracts the Greek poetic lines from <l> elements.

    Parameters:
        url (str): The URL to the TEI XML resource.

    Returns:
        list of str: A list of Greek text lines.
    """
    try:
        response = requests.get(url)
        response.raise_for_status()
    except requests.RequestException as e:
        print(f"Error fetching URL: {e}")
        return []

    try:
        tree = etree.fromstring(response.content)
    except etree.XMLSyntaxError as e:
        print(f"Error parsing XML: {e}")
        return []

    ns = {'tei': 'http://www.tei-c.org/ns/1.0'}
    lines = tree.xpath('//tei:l', namespaces=ns)

    return [line.text.strip() for line in lines if line.text]

# Example usage:
url = "https://atlas.perseus.tufts.edu/library/passage/urn:cts:greekLit:tlg0011.tlg007.perseus-grc2:1-"+curcit+"/xml/"
greek_lines = extract_greek_lines_from_tei(url)
basetexts = ''
for line in greek_lines:
    basetexts = basetexts + ' ' + line
    #print(line)
basetextl = basetexts.split()


```

```python
def preptext(workline1):
    workline1 = re.sub('·','.',workline1)
    workline1 = re.sub('·','.',workline1)
    #workline1 = re.sub('\*','',workline1)
    workline1 = re.sub('([,\.;—])',' \g<1>',workline1)
    return(workline1)


basetexts = re.sub('[\n\s]+','\n',preptext(basetexts))
newtexts = re.sub('[\n\s]+','\n',preptext(newtexts))

sm = difflib.SequenceMatcher(None, basetexts.split(), newtexts.split())
print('ratio',sm.ratio())
opcodes = sm.get_opcodes()
print('opcodes',len(opcodes))
diff = difflib.Differ()
result = list(diff.compare(basetexts.split(), newtexts.split()))


```

```python

def side_by_side_diff(fout,a: str, b: str, width=60):
    from itertools import zip_longest
    global tokenlist

    a_lines = a.splitlines()
    b_lines = b.splitlines()
    sm = difflib.SequenceMatcher(None, a_lines, b_lines)

    tokennum = -1
    for tag, i1, i2, j1, j2 in sm.get_opcodes():
        for left, right in zip_longest(a_lines[i1:i2], b_lines[j1:j2], fillvalue=''):
            if(tag == 'replace' and re.sub('\*','',left) == right):
                tag = 'equal*'
            if tag == 'equal':
                marker = ' '
            elif tag == 'replace':
                marker = '|'
            elif tag == 'delete':
                marker = '<'
                right = ''
            elif tag == 'insert':
                marker = '>'
                left = ''
            #print(f"{left:<{width}} {marker} {right}",file=fout)
            if(left):
                tokennum = tokennum + 1
                curtoken = tokenlist[tokennum]
            else:
                curtoken = left
                
            print(curtoken,left,tag,right,sep='\t',file=fout)
                
fout = open('soph.oc-diffs2.txt','w')
side_by_side_diff(fout,newtexts,basetexts)
fout.close()
```
