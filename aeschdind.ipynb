```python
import re


def updatef(basen):

    fname = '/Users/gcrane/github/english_trans-dev/volumes/aeschylus_1909/aesch.' + basen + '.headlam-eng1.xml'
    f = open(fname)
    outfname = re.sub('/Users/gcrane/Downloads','/Users/gcrane/github/GRC_misc',fname)
    outfname = re.sub('-eng1.xml','_eng2.xml',outfname)
    print(outfname)
    outf = open(outfname,'w')

    curlnum = 0
    for l in f:
        l = re.sub('\n$','',l)
        l = re.sub('(.+)<note place="right" type="lyric">([^<]+)</note>','<milestone unit="lyric" n="\g<2>"/>\n\g<1>',l)

        l = re.sub('<head>[^<]+</head>','',l)
        m = re.search('<lb n="([0-9]+)([a-z]*)',l)
        if(m):
            curlnum = int(m[1])
            #print(l,file=outf)
            #print(l)
            continue
        
        if(curlnum and re.search('(<l [^>]+[0-9]">|<l>)',l)):
            curlnum = curlnum + 1
            l = re.sub('(<l [^>]+>|<l>)', '<l n="'+str(curlnum)+'">',l)

                   
        print(l,file=outf)
    f.close()
    outf.close()

updatef('pb')
```
