```python
import re

sophabbs = {
    'Aj.' : 'Aj.',
    'Ant.' : 'Ant.',
    'El.' : 'El.',
    'O.C.' : 'OC',
    'O.T.' : 'OT',
    'Phil.' : 'Phil.',
    'Tr.' : 'Trach.'
    
}

aeschabbs = {
    'Ag.' : 'Ag.' ,
    'Choeph.' : 'LB',
    'Prom.' : 'PB' ,
    'P. V.' : 'PB' ,
    'P.V.' : 'PB' ,
    'S. c. Τ.' :  'Seven',
    'Suppl.' : 'Supp.',

}


rnums = {
    'i': '1',
    'ii': '2',
    'iii': '3',
    'iv': '4',
    'v': '5',
    'vi': '6',
    'vii': '7',
    'viii': '8',
    'ix': '9',
    'x' : '10',

}
tmpfile = '/tmp/clang.xml'
f = open('../GRC_misc/campbell-sophlanguage.xml')
outf = open(tmpfile,'w')

titregex = '(Aj\.|Ant\.|El\.|O\.\s*C\.|O\.\s*T\.|Phil\.|Tr\.)\s+([0-9\-]+)'
atitregex = 'Aesch\. (Ag\.|Choeph\.|Prom\.|P\.\s*V\.|S\.\s*c\.\s*T\.|Suppl\.)\s+([0-9\-]+)'
etitregex = 'Eur\. (Alc\.|Bacch\.|Hel\.|Hippol\.|I\.\s*A\.|Ion|Or\.|Phoen\.)\s+([0-9\-]+)'
#titregex = '(Aj\.)\s+([0-9\-]+)'

eurabbs = {
    'Alc.' : 'Alc.',
    'Bacch.' : 'Ba.',
    'Hel.' : 'Hel.',
    'Hippol.' : 'Hipp.',
    'I. A.' : 'IA',
    'Ion' : 'Ion',
    'Or.' : 'Orest.',
    'Phoen.' : 'Phoen.',

}

proseregex = '(Hdt\.|Thuc\.)\s+(i|ii|iii|iv|v|vi|vii|viii|ix|x)\.\s*([0-9\-]+)'
for l in f:
    
    l = re.sub('\s+$','',l)
    l = re.sub('Ο(\.\s*[CTΤ])','O\g<1>',l)
    l = re.sub('(O\.\s*)Τ\.','\g<1>T.',l)
    l = re.sub('[OΟО]\.\s*[TΤ]\.','O. T.',l)
    l = re.sub('Αj\.','Aj.',l)
    l = re.sub('Εl\.','El.',l)
    l = re.sub('Τr\.','Tr.',l)
    l = re.sub('Ρhil\.','Phil.',l)
    l = re.sub('[ОO]\.\s*[CС]\.','O. C.',l)
    m = re.search('Ib\.\s+([0-9\-]+)',l)
    if(m):
        print(curauth,l)
        if(curauth in sophabbs.values()):
            l = re.sub('Ib\.\s+([0-9\-]+)','\n<bibl n="Soph. '+curauth+'ZZZ \g<1>">Ib.ZZZ \g<1></bibl>',l)
        else:
            l = re.sub('Ib\.\s+([0-9\-]+)','\n<bibl n="'+curauth+'ZZZ \g<1>">Ib.ZZZ \g<1></bibl>',l)

    l = re.sub('\s*(<p>|</p>|</div>|<div[^>]+>|<head>|</head>|<list>|<item>)\s*','\n\g<1>\n',l)
    
    m = re.search('Ib\.\s+([0-9\-]+)',l)
    if(m):
        
        l = re.sub('Ib\.\s+([0-9\-]+)','\n<bibl n="'+curauth+' \g<1>">Ib. \g<1></bibl>',l)

    l = re.sub('(Plat\.\s+Rep\.)\s+([0-9]|[ivx]+\.)\s+([0-9]+)\s*([A-E])','<bibl n="\g<1> \g<3>\g<4>">\g<1> \g<2> \g<3> \g<4></bibl>',l)
    l = re.sub('(Plat\.)\s+(Legg\.)\s+([0-9]|[ivx]+\.)\s+([0-9]+)\s*([A-E])','<bibl n="\g<1> Laws \g<4>\g<5>">\g<1> \g<2> \g<3> \g<4> \g<5></bibl>',l)
    l = re.sub('(Plat\.)\s+(Legg\.)\s+([0-9]+)\s*([A-E])','<bibl n="\g<1> Laws \g<3>\g<4>">\g<1> \g<2> \g<3> \g<4></bibl>',l)
    l = re.sub('(Plat\.)\s+(Theaet\.|Soph\.)\s+([0-9]+)\s*([A-E]+)','<bibl n="\g<1>ZZZ \g<2> \g<3>\g<4>">\g<1>ZZZ \g<2> \g<3> \g<4></bibl>',l)
    l = re.sub('(Plat\.)\s+(Lysis|Theaet\.|Soph\.)\s+([0-9]+)\s*([A-E]+)','<bibl n="\g<1> \g<2> \g<3>\g<4>">\g<1> \g<2> \g<3> \g<4></bibl>',l)
    while(re.search(proseregex,l)):
        m = re.search(proseregex,l)
        curbook = rnums[m[2]]
        l = re.sub(proseregex,'<bibl n="\g<1> '+curbook+'.\g<3>">\g<1>ZZZ \g<2>. \g<3></bibl>',l,1)
        
    while(re.search(titregex,l)):
        m = re.search(titregex,l)
        curauth = sophabbs[re.sub('\s+','',m[1])]
        curline = m[2]
       
        l = re.sub(titregex,'\n<bibl n="Soph. '+curauth+' ZZZ\g<2>">\g<1> ZZZ\g<2></bibl>',l,1)
  
        
    while(re.search(etitregex,l)):
        m = re.search(etitregex,l)
        curauth = eurabbs[re.sub('\s+','',m[1])]
        curline = m[2]       
        l = re.sub(etitregex,'\n<bibl n="Eur. '+curauth+' ZZZ\g<2>">Eur. \g<1> ZZZ\g<2></bibl>',l,1)
        
    while(re.search(atitregex,l)):
        m = re.search(atitregex,l)
        curauth = aeschabbs[re.sub('\s+','',m[1])]
        curline = m[2]       
        l = re.sub(atitregex,'\n<bibl n="Aesch. '+curauth+' ZZZ\g<2>">Aesch. \g<1> ZZZ\g<2></bibl>',l,1)
        
    l = re.sub('ZZZ','',l)
    l = re.sub('(Il\.|Od\.)\s+([0-9]+)\.\s*([0-9\-]+)','<bibl n="Hom. \g<1> \g<2>.\g<3>">\g<1> \g<2>.\g<3></bibl>',l)
    print(l,file=outf,end=' ')

f.close()
outf.close()

outf = open('../GRC_misc/campbell-sophlanguage-2.xml','w')

f = open(tmpfile)
for l in f:
    l = re.sub('\s+$','',l)
    l = re.sub('</bibl>\.\s+lyr\.',' lyr.</bibl>',l)
    l = re.sub('(</bibl>[,\.—\s]+)([^<]+)([—\-\s]+<gloss>)','\g<1><quote xml:lang="grc">\g<2></quote>\g<3>',l)
    l = re.sub('(</bibl>\.\s*)([^<a-zA-Z+[α-ω][^<a-zA-Z-]+)','\g<1><quote xml:lang="grc">\g<2></quote>',l)
    l = re.sub('\s+</quote>\s*','</quote> ',l)
    l = re.sub('</bibl>\.','.</bibl>',l)
    l = re.sub('(</bibl>\s+)([^\na-z<A-Z]+[α-ω][^\n<a-zA-Z]+)','\g<1><quote xml:lang="grc">\g<2></quote>',l)
    l = re.sub('\s+</quote>\s*','</quote> ',l)
    l = re.sub('\s*(<quote xml:lang="grc">)\s+',' \g<1>',l)
    print(l,file=outf)


    
f.close()
outf.close()
```

```python
import re

f = open('../GRC_misc/campbsoph1-2.xml')
outf = open('../GRC_misc/campbsoph1-3.xml','w')

prevline = 0
curdiv = 0
for l in f:
    l = re.sub('\s+$','',l)

    m = re.search('pb n="([0-9]+)',l)
    if(m):
        curpage = m[1]
    m = re.search('commline" n="([0-9]+)',l)
    if(m):
        curline = int(m[1])
        
        if(curline < prevline):
            print(curpage,'jump',prevline,l)
        prevline = curline
    if(re.search('<div',l)):
        curdiv = curdiv + 1
        if(not curdiv == 1 ):
            print(curdiv,l,curpage)

    if(re.search('</div>',l)):
        curdiv = curdiv - 1
        if(not curdiv == 0):
            print(curdiv,l,curpage)
    if(curdiv<0 or curdiv> 1):
        print(curdiv,l)

    print(l,file=outf)

f.close()
outf.close()
```
