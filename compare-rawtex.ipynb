```python
import re

def genrawtxt(fname1,outfname):
    intext = 0
    f = open(fname1)
    text = ''
    for l in f:
        if(re.search('<text',l)):
            intext = 1
        if(not intext):
            continue
        l = re.sub('\s+$','',l)
        m = re.search('<pb[^>]+n="([^"]+)"/>',l)
        #if(m):
        #    print(m[1])

        l = re.sub('<pb[^>]+n="([^"]+)"/>',' page\g<1>',l)
        l = re.sub('<[^>]+>',' ',l)
        l = re.sub('\s+','\n',l)
        text = text + '\n' + l

    text = re.sub('\n+','\n',text)
    f.close()

    outf = open(outfname,'w')
    print(text,file=outf)
    outf.close()

genrawtxt('/Users/gcrane/swanag.xml','/tmp/f2.txt')
genrawtxt('/Users/gcrane/github/GRC_misc/aesch.ag.swanwick1886.xml','/tmp/f1.txt')



```
