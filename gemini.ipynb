```python
# Install the Google Generative AI library
%pip install -q -U google-generativeai

import google.generativeai as genai
import os

def get_gemini_api_key():
    """Prompts the user for their Gemini API key."""
    api_key = input("Please enter your Gemini API key: ")
    return api_key

# Get API key from the user
api_key = get_gemini_api_key()

# Configure the Gemini API with the provided key
genai.configure(api_key=api_key)

print("Gemini API configured!")
```

```python
# Choose a Gemini model (gemini-pro is generally a good starting point for text)
# You can explore other models like 'gemini-1.5-flash' for faster responses
# or 'gemini-1.5-pro' for more capable responses.
model = genai.GenerativeModel('gemini-pro')

print("\n--- Gemini AI Chat ---")
print("Type 'exit' to quit.")

while True:
    user_prompt = input("You: Translate into English ἐγὼ δὲ θερμόνους τάχ᾽ ἐν πέδῳ βαλῶ")
    if user_prompt.lower() == 'exit':
        print("Exiting chat. Goodbye!")
        break

    try:
        # Generate content based on the user's prompt
        response = model.generate_content(user_prompt)

        # Print the model's response
        print(f"Gemini: {response.text}")

    except Exception as e:
        print(f"An error occurred: {e}")
        print("Please ensure your API key is correct and you have an active internet connection.")
```
