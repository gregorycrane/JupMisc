```python
import re

def dopage(fname,curpage,fout):
    print(fname)
    
    f = open('../../Downloads/sophocles-antigone-hermann-uva-x030731879-1748053298/'+fname)
    for l in f:
        l = re.sub('\s+$','',l)
        print(l,file=fout)

    f.close()
    
fout = open('soph.ant.hermann-int-lat.xml','w')
for i in range(5,122):
    if(i<10):
        curpage = '0000000' + str(i-4) + '.txt'
    elif(i<100):
        curpage = '000000' + str(i-4) + '.txt'
    else:
        curpage = '00000' + str(i-4) + '.txt'
    print(curpage)

    dopage(curpage,i-4,fout)
fout.close()
```
