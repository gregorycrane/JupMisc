```python
import re

fname = '../GRC_misc/nauck.tragfrag-fontes.xml'
outfname = '../GRC_misc/nauck.tragfrag-fontes-2.xml'

f = open(fname)
curbuf = f.read()

f.close()

outf = open(outfname,'w')
curbuf = re.sub('(Soph\.|Aesch\.|Eur\.|adesp\.)([\n\s]+)([0-9]+)', '\n<bibl>\g<1> \g<3></bibl>',curbuf)
print(curbuf,file=outf)
outf.close()

fraglist = {}
fragcnt = {}
f = open(outfname)
curauthw = 'NA'
for l in f:
    m = re.search('<authw>(.+)</authw>',l)
    if(m):
        curauthw = m[1]
    if(not curauthw in fraglist):
        fragcnt[curauthw] = 0
        fraglist[curauthw] = []
    while(re.search('<bibl>([^<]+)</bibl>',l)):
        m = re.search('<bibl>([^<]+)</bibl>',l)
        fragcnt[curauthw] = fragcnt[curauthw] + 1
        fraglist[curauthw].append(m[1])
        l = re.sub('<bibl>([^<]+)</bibl>',' ',l,1)
f.close()
```

```python
i = 0
for foo in sorted(fragcnt,key=fragcnt.get,reverse=True):
    i = i + 1
    print(i,foo,fragcnt[foo])
```
