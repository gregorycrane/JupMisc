```python
import re

skipped = {}

f = open('../GRC_misc/save/dind-aesch-lex-dupes.xml')
for l in f:
    m = re.search('xml:id="([^"]+)"',l)
    if(m):
        skipped[m[1]] = 1
f.close()

current = {}
f = open('../GRC_misc/aeschylus-gemini.dindorf.lexicon.xml')
for l in f:
    m = re.search('xml:id="([^"]+)"',l)
    if(m):
        current[m[1]] = 1
f.close()

for foo in skipped:
    if(not foo in current):
        print('skipped',foo)
```
